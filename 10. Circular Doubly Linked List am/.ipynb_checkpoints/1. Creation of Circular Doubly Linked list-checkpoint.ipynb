```python
class Node:
    def __init__(self,value= None):
        self.value = value
        self.prev = None
        self.next = None

class CircularDoublyLL:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def createcdll(self,value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.size = 1
        
```
