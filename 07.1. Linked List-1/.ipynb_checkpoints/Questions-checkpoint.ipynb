```python
class Node:
    def __init__(self,value = None):
        self.value = value
        self.next = None

class SLList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def insert_at_beginning(self,value):
        new_node = Node(value)
        if self.head is None:
            self.head = self.tail = new_node
        else:
            new_node.next = self.head
            self.head = new_node
        self.size += 1

    def insert_at_end(self,value):
        new_node = Node(value)
        if self.head is None:
            self.head = self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node
        self.size += 1
    def insert_at_middle(self,value,position):
        if position == 0:
            return self.insert_at_beginning(value)
        elif position == self.size:
            return self.insert_at_end(value)
        elif position > self.size:
            raise IndexError("Position out of bounds")
        new_node = Node(value)
        index = 0
        temp_node = self.head
        while index < position - 1:
            if temp_node.next is None:
                raise IndexError("Index out of bounds")
            temp_node = temp_node.next
            index += 1
        next_node = temp_node.next
        temp_node.next = new_node
        new_node.next = next_node
        self.size += 1

    def search_node_return_index(self,value):
        curr = self.head
        count = 0
        while curr:
            if curr.value == value:
                return count
            count += 1
            curr = curr.next
        return -1

sll = SLList()
sll.insert_at_middle(8,0)
sll.insert_at_middle(4,1)
sll.insert_at_middle(1,2)
sll.insert_at_middle(9,0)

sll.search_node_return_index(8)
```
