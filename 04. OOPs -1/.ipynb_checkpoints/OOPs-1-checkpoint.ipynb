## 8. Self Parameter / instance methods - in which self is passed

```python
class Student:
    passin_percentage=40 # class attribute
    def student_details(self):
        self.name="vishal"
        self.percentage=80
        print("name = ", self.name)
        print("percentage = ", self.percentage)
        print("Passin percentage is : ",self.passin_percentage)

    def is_passed(self):
        if self.percentage>Student.passin_percentage:
            print("Student is passed.")
        else:
            print("Student is not passed.")

s1=Student()
s1.student_details()
s1.is_passed()
```

## 11. Static method

```python
class Student:
    passin_percentage=40 # class attribute
    def student_details(self):
        self.name="vishal"
        self.percentage=80
        print("name = ", self.name)
        print("percentage = ", self.percentage)
        print("Passin percentage is : ",self.passin_percentage)

    def is_passed(self):
        if self.percentage>Student.passin_percentage:
            print("Student is passed.")
        else:
            print("Student is not passed.")
            
    # static method doesn't require self argument
    @staticmethod
    def welcome_to_school():
        print("Hey! Welcome to School.")

s1=Student()
s1.student_details()
s1.is_passed()
s1.welcome_to_school()
```

## Public and Private Modifiers

```python
class Student:
    def __init__(self,name,age):
        self.__name=name
        self.age=age
    def print_student_details(self):
        print(self.__name,end=" ")
        print(self.age)

s=Student("vishal",20)
print(s.age)
print(s.name)
```

```python
class Student:
    def __init__(self,name,age):
        self.__name=name
        self.age=age
    def print_student_details(self):
        print(self.__name,end=" ")
        print(self.age)

s=Student("vishal",20)
s.print_student_details()
```

```python
class Student:
    name="vishal"
    def store_details(self):
        self.age=60
    def print_age(self):
        print(self.age)

s=Student()
s.store_details()
s.print_age()
s2=Student()
s2.store_details()
s2.print_age()
s1=Student()
s1.print_age()
```

## Fraction Class

```python
class Fraction:
    def __init__(self,num=0,den=1):
        if den==0:
            den=1
        self.num=num
        self.den=den

    def print(self):
        if self.num==0:
            print(0)
        elif self.den==1:
            print(self.num)
        else:
            print(self.num, "/", self.den)
    def simplify(self):
        if self.num==0:
            self.den=1
            return
        current=min(self.num, self.den)
        while current >1:
            if self.num%current==0 and self.den % current==0:
                break
            current -=1
        self.num=self.num//current
        self.den=self.den//current

    # adding function of two objects data
    def add(self, otherFraction):
        new_num=otherFraction.den*self.num + otherFraction.num*self.den
        new_den=otherFraction.den * self.den

        self.num=new_num
        self.den=new_den
        self.simplify()

    def multiply(self,other_fraction):
        new_num=self.num*other_fraction.num
        new_den=self.den*other_fraction.den

        self.num=new_num
        self.den=new_den
        self.simplify()
```

```python
f=Fraction(0,5)
f.print()
```

```python
f.simplify()
f.print()
```

```python
f1=Fraction(14,6)
f1.print()
```

```python
f1.simplify()
```

```python
f1.print()
```

```python
f2=Fraction(14,5)
f2.print()
f2.simplify()
f2.print()
```

```python
f7=Fraction(2,3)
f8=Fraction(1,3)
```

```python
f7.add(f8)
f7.print()
```

```python
f1=Fraction(4,6)
f2=Fraction(7,8)
```

```python
f1.multiply(f2)
f1.print()
```
