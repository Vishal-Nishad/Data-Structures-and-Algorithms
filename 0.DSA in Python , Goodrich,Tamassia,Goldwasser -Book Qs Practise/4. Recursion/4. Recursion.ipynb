```python
# iterative fibonacci
def it_fib(n):
    if n==1:
        return 1
    else:
        fib=0
        a=0
        b=1
        for i in range(1,n):
            fib=a+b
            a=b
            b=fib
        return fib
it_fib(4)
            
```

```python
# recursive approach
def rec_fib(n):
    if n==1 or n==2:
        return 1
    return rec_fib(n-1)+rec_fib(n-2)

rec_fib(5)
```

```python
rec_fib(5) = rec_fib(4) + rec_fib(3)
rec_fib(4) = rec_fib(3) + rec_fib(2)
rec_fib(3) = rec_fib(2) + rec_fib(1)
rec_fib(2) = 1
rec_fib(1) = 1

Now calculating:
rec_fib(3) = 1 + 1 = 2
rec_fib(4) = 2 + 1 = 3
rec_fib(5) = 3 + 2 = 5
```

```python
# Page no. 165 , A recursive approach to ruler drawing
def draw_line(tick_length, tick_label= ''):
    line = '-'*tick_length
    if tick_label:
        line += ' ' + tick_label
    print(line)

def draw_interval(center_length):
    if center_length>0:
        draw_interval(center_length-1)
        draw_line(center_length)
        draw_interval(center_length-1)

def draw_ruler(num_inches,major_tick_length):
    draw_line(major_tick_length, '0')
    for j in range(1,num_inches+1):
        draw_interval(major_tick_length-1)
        draw_line(major_tick_length, str(j))


draw_ruler(2,4)
```

```python
# binary search using iteration
def bin(li,x):
    start=0
    end=len(li)-1
    while start<=end:
        mid=(start+end)//2
        if li[mid]==x:
            return mid
        elif li[mid]>x:
            end=mid -1
        else:
            start=mid+1
    return 'not found'
li=[1,2,3,4,5]
bin(li,6)
```

```python
# using recursion method
def bin_search(data,target,start,end):
    if start>end:
        return "Element not found"
    else:
        mid=(start+end)//2
        if data[mid]==target:
            return mid
        elif data[mid]>target:
            return bin_search(data,target,start,mid-1)
        else:
            return bin_search(data,target,mid+1,end)

li=[1,2,3,4,5,6]
bin_search(li,5,0,len(li)-1)
```

```python
# page 179, summing the elements of sequence recursively

def summ(li,n):
    if n==0:
        return 0
    return summ(li,n-1)+li[n-1]

li=[5,6,50,100]
summ(li,len(li))
```

```python
# reversing a sequence with recursion
def reverse(li,start,end):
    if start>=end:
        return
    li[start],li[end]=li[end],li[start]
    reverse(li,start+1,end-1)
li=[1,2,3,4,0]
reverse(li,0,len(li)-1)
li
```

```python
# page 181, computing powers
def power(x,n):
    if n==0:
        return 1
    return x*power(x,n-1)
power(5,3)
```

```python
# page 182, computing power using repeated squaring
def pwr(x,n):
    if n==0:
        return 1
    partial=pwr(x,n//2)
    result=partial*partial
    if n%2==1:
        result*=x
    return result

pwr(2,4)
```

```python
#4.3 recursively computing product of two numbers , using only addition and subtraction concept
def product_(num1,num2,value):
    if num2==0:
        return value
    else:
        num2-=1
        value+=num1
        return product_(num1,num2,value)

if __name__=="__main__":
    num1=int(input())
    num2=int(input())
    print(product_(num1,num2,0))
```

```python
# 4.4 recursively check if string is pallindrome or not
def is_pallindrom(s:str,start,end):
    if start>=end:
        return True
    if s[start]==s[end]:
        return is_pallindrom(s,start+1,end-1)
    else:
        return False

if __name__=="__main__":
    text=str(input())
    print(is_pallindrom(text,0,len(text)-1))
```

```python
# 4.5 rearranges a arr of int values , even befor odd values
def rearrange(ar,start,end):
    if start>=end:
        return ar
    if ar[start]%2==0:
        return rearrange(ar,start+1,end)
    elif ar[end]%2!=0:
        return rearrange(ar,start,end-1)
    else:
        ar[start],ar[end]=ar[end],ar[start]
        return rearrange(ar,start+1,end-1)

if __name__=="__main__":
    li=[1,2,3,4,5,6]
    print(rearrange(li,0,len(li)-1))
```
