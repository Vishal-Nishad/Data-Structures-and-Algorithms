```python
# Illustrative Programs
# 5.1 copy to element of array A to B
def copy_arr(a):
    b=[]
    for i in range(len(a)):
        b.append(a[i])
    return b

a=[4,3,2]

copy_arr(a)
```

```python
# book solution
def copy(a):
    b=[None]*len(a)
    for i in range(len(a)):
        b[i]=a[i]
    return b
a=[1,23,'a']
copy(a)
```

```python
# 5.2 find frequency of each element in arr a
def freq(a):
    frequency={}
    for i in a:
        if i not in frequency.keys():
            frequency[i]=1
        else:
            frequency[i] +=1
    return frequency

if __name__=="__main__":
    a=[4,23,43245,2,2,23,65,1,1,1]
    frequency=freq(a)
    print(frequency)
    for i in frequency.keys():
        print("\n",i, " : ", frequency[i])
```
