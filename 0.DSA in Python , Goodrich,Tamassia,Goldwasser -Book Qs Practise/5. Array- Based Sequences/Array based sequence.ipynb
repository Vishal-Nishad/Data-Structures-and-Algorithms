```python
# Illustrative Programs
# 5.1 copy to element of array A to B
def copy_arr(a):
    b=[]
    for i in range(len(a)):
        b.append(a[i])
    return b

a=[4,3,2]

copy_arr(a)
```

```python
# book solution
def copy(a):
    b=[None]*len(a)
    for i in range(len(a)):
        b[i]=a[i]
    return b
a=[1,23,'a']
copy(a)
```

```python
# 5.2 find frequency of each element in arr a
def freq(a):
    frequency={}
    for i in a:
        if i not in frequency.keys():
            frequency[i]=1
        else:
            frequency[i] +=1
    return frequency

if __name__=="__main__":
    a=[4,23,43245,2,2,23,65,1,1,1]
    frequency=freq(a)
    print(frequency)
    for i in frequency.keys():
        print("\n",i, " : ", frequency[i])
```

```python
def count_freq(ls):
    ans = {}
    for i in ls:
        count = 0
        for j in range (len(ls)):
            if i == ls[j]:
                count +=1
        ans[i] = count
    return ans
ls = [43,3,43,2,2,4,4554,2,3,2343,5,53,2,2,4,5]
count_freq(ls)
```

```python
# 5.3 remove array element present immediately to the right of the element entered by the user
def remove_right(ar,element):
    try:
        element_index=ar.index(element)
    # if not present in list then raise exception
    except ValueError:
        print("Element is not present in the list.")
    if element_index!=len(ar)-1:
        ar.pop(element_index+1) 
    else:
        print("Element is in the last cell of the list.")
    return ar

if __name__=="__main__":
    ar=[1,2,1,3]
    print(remove_right(ar,1))
```

```python
# 5.4 remove array element with max frequency
def count_freq(ar):
    frequency={}
    for i in ar:
        if i not in frequency.keys():
            frequency[i]=1
        else:
            frequency[i]+=1
    return frequency

def remove_max_freq(ar):
    frequency=count_freq(ar)
    element=-1
    max_freq=0
    for i in frequency.keys():
        if max_freq<frequency[i]:
            max_freq=frequency[i]
            element=i
    for i in range(max_freq):
        ar.remove(element)
    return ar

if __name__=="__main__":
    ar=[1,2,1,4]
    print(remove_max_freq(ar))
    
```

```python
# 5.5 remove duplicates
def remove_dup(ar):
    return list(set(ar))

if __name__=="__main__":
    ar=[1,2,1,2,1,3]
    print(remove_dup(ar))
```

```python
def remove_duplicate(ar):
    ar.sort()
    temp=[]
    for i in range(len(ar)-1):
        if ar[i]!=ar[i+1]:
            temp.append(ar[i])
    temp.append(ar[len(ar)-1])
    return temp

if __name__=="__main__":
    ar=[1,2,1,2,1,3]
    print(remove_duplicate(ar))
```

```python
# 5.6 program to sort the aray elements in increasing order of their frequencies.
def freq_count(ar):
    freq={}
    for i in ar:
        if i in freq.keys():
            freq[i]+=1
        else:
            freq[i]=1
    return freq

def freq_sort(ar):
    freq_=freq_count(ar)
    return [key for key, value in sorted(freq_.items(), key=lambda item: item[1])]

if __name__=="__main__":
    ar=[5,5,3,8,7,8,5,2,1]
    print(freq_sort(ar))
```

```python
# create a list of prime elements form given list
def is_prime(n):
    flag=True
    for i in range(2,(n//2)+1):
        if n%i==0:
            flag=False
            return flag
    return flag

is_prime(4)
def prime_elements(ar):
    prime_arr=[]
    for i in ar:
        prime=is_prime(i)
        if prime:
            prime_arr.append(i)

    return prime_arr

if __name__=="__main__":
    ar=[5,4,8,10,2,3]
    print(prime_elements(ar))
```
