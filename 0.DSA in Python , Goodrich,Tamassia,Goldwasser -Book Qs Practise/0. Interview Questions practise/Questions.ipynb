```python
# GCD of two numbers
def gcd(a,b):
    while b!=0:
        c=b
        b=a%b
        a=c
    return a

gcd(100,20)
```

```python
# multiple numbers
def gcd_n(arr):
    smallest=arr[0]
    for i in arr:
        if i<smallest:
            smallest=i

    for i in range(smallest,0,-1):
        is_gcd=True
        for j in arr:
            if j%i!=0:
                is_gcd=False
                break
        if is_gcd:
            return i

ar=[6,15,24]
gcd_n(ar)
```

```python
# is prime or not
def is_prime(n):
    flag=True
    for i in range(2,n//2+1):
        if n%i==0:
            flag=False
    if flag:
        print(f"given number {n} is Prime number.")
    else:
        print(f"given number {n} is not Prime number.")

is_prime(11)
```

```python
# to display the distinct element of array
def distinct(li):
    new_list=[]
    for i in range(len(li)):
        count=0
        for j in range(len(li)):
            if li[i]==li[j]:
                count+=1
        if count==1:
            new_list.append(li[i])
    return new_list

lis=[4,5,4,6,7,5]
distinct(lis)
```

```python
# to remove vowels from an input strings
def remove_vowels(st):
    new_st=''
    for i in st:
        if i !='a' and i !='e' and i !='i' and i !='o' and i !='u':
            new_st+=i
    return new_st

remove_vowels('vishal')
```
