```python
# GCD of two numbers
def gcd(a,b):
    while b!=0:
        c=b
        b=a%b
        a=c
    return a

gcd(100,20)
```

```python
# is prime or not
def is_prime(n):
    flag=True
    for i in range(2,n//2+1):
        if n%i==0:
            flag=False
    if flag:
        print(f"given number {n} is Prime number.")
    else:
        print(f"given number {n} is not Prime number.")

is_prime(11)
```
