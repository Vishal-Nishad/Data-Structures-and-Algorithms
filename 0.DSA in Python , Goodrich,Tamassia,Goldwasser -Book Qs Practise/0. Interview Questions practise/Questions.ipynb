```python
# GCD of two numbers
def gcd(a,b):
    while b!=0:
        c=b
        b=a%b
        a=c
    return a

gcd(100,20)
```

```python
# multiple numbers
def gcd_n(arr):
    smallest=arr[0]
    for i in arr:
        if i<smallest:
            smallest=i

    for i in range(smallest,0,-1):
        is_gcd=True
        for j in arr:
            if j%i!=0:
                is_gcd=False
                break
        if is_gcd:
            return i

ar=[6,15,24]
gcd_n(ar)
```

```python
# is prime or not
def is_prime(n):
    flag=True
    for i in range(2,n//2+1):
        if n%i==0:
            flag=False
    if flag:
        print(f"given number {n} is Prime number.")
    else:
        print(f"given number {n} is not Prime number.")

is_prime(11)
```

```python
# to display the distinct element of array
def distinct(li):
    new_list=[]
    for i in range(len(li)):
        count=0
        for j in range(len(li)):
            if li[i]==li[j]:
                count+=1
        if count==1:
            new_list.append(li[i])
    return new_list

lis=[4,5,4,6,7,5]
distinct(lis)
```

```python
# to remove vowels from an input strings
def remove_vowels(st):
    new_st=''
    for i in st:
        if i !='a' and i !='e' and i !='i' and i !='o' and i !='u':
            new_st+=i
    return new_st

remove_vowels('vishal')
```

**Q-Maximum and Minimum Element in an Array**

```python
def max_and_min(arr):
    min=arr[0]
    for i in range (1,len(arr)):
        if arr[i]<min:
            min=arr[i]
    max=arr[0]
    for i in range(1,len(arr)):
        if arr[i]>max:
            max=arr[i]
    return min,max

arr=[4,5,2,1,0]
max_and_min(arr)
```

**Q-Reverse the Array**

```python
def rev_arr(arr):
    new_arr=[]
    for i in range(len(arr)-1,-1,-1):
        new_arr.append(arr[i])
    return new_arr
arr=[4,5,2,0,1]
rev_arr(arr)
```

```python
def rev_arr(arr):
    return arr[::-1]
arr=[5,3,2,8,0]
rev_arr(arr)
```

```python
# by swapping 
def rev_arr(arr):
    start=0
    end=len(arr)-1
    while start<end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
    return arr
arr=[2,6,4,2,3]
rev_arr(arr)
```

**Q-	Maximum-Subarray**

```python
Given an integer array nums, find the subarray with the largest sum, and return its sum.

Example 1:

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.
```

```python
def max_subarray(arr):
    max_sum=arr[0]
    current_sum=arr[0]
    for i in range(1,len(arr)):
        if current_sum+arr[i]>arr[i]:
            current_sum=current_sum+arr[i]
        else:
            current_sum=arr[i]
        if current_sum>max_sum:
            max_sum=current_sum
    return max_sum

arr=[-2,1,-3,4,-1,2,1,-5,4]
max_subarray(arr)
```

**Q-Contains Duplicate**

```python
def contains_duplicate(arr):
    seen=set()
    for num in arr:
        if num in seen:
            return True
        seen.add(num)
    return False
arr=[1,2,3,5,3]
contains_duplicate(arr)
```

```python

def cont_dupl(arr):
    return len(set(arr))!=len(arr)

arr=[1,2,3]
cont_dupl(arr)
```

**Q-	Chocolate Distribution Problem**

```python
def choco_min(arr,m):
    arr.sort()
    return arr[(m-1)]-arr[0]

arr=[7, 3, 2, 4, 9, 12, 56]
m = 5 
choco_min(arr,m)
```
