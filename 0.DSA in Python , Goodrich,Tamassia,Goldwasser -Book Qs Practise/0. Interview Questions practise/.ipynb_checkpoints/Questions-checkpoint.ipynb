```python
# GCD of two numbers
def gcd(a,b):
    while b!=0:
        c=b
        b=a%b
        a=c
    return a

gcd(100,20)
```

```python
# multiple numbers
def gcd_n(arr):
    smallest=arr[0]
    for i in arr:
        if i<smallest:
            smallest=i

    for i in range(smallest,0,-1):
        is_gcd=True
        for j in arr:
            if j%i!=0:
                is_gcd=False
                break
        if is_gcd:
            return i

ar=[6,15,24]
gcd_n(ar)
```

```python
# is prime or not
def is_prime(n):
    flag=True
    for i in range(2,n//2+1):
        if n%i==0:
            flag=False
    if flag:
        print(f"given number {n} is Prime number.")
    else:
        print(f"given number {n} is not Prime number.")

is_prime(11)
```

```python
# to display the distinct element of array
def distinct(li):
    new_list=[]
    for i in range(len(li)):
        count=0
        for j in range(len(li)):
            if li[i]==li[j]:
                count+=1
        if count==1:
            new_list.append(li[i])
    return new_list

lis=[4,5,4,6,7,5]
distinct(lis)
```

```python
# to remove vowels from an input strings
def remove_vowels(st):
    new_st=''
    for i in st:
        if i !='a' and i !='e' and i !='i' and i !='o' and i !='u':
            new_st+=i
    return new_st

remove_vowels('vishal')
```

**Q-Maximum and Minimum Element in an Array**

```python
def max_min(ar):
    max=ar[0]
    min=ar[0]
    for i in range(1,len(ar)):
        if ar[i]>max:
            max=ar[i]
        if ar[i]<min:
            min=ar[i]
    return max,min

ar=[3,25,23,1,44,56,0,454]
max_min(ar)
```

```python
def max_and_min(arr):
    min=arr[0]
    for i in range (1,len(arr)):
        if arr[i]<min:
            min=arr[i]
    max=arr[0]
    for i in range(1,len(arr)):
        if arr[i]>max:
            max=arr[i]
    return min,max

arr=[4,5,2,1,0]
max_and_min(arr)
```

**Q-Reverse the Array**

```python
# with the help of new variable
def rev_arr(arr):
    new_arr=[]
    for i in range(len(arr)-1,-1,-1):
        new_arr.append(arr[i])
    return new_arr
arr=[4,5,2,0,1]
rev_arr(arr)
```

```python
# slicing 
def rev_arr(arr):
    return arr[::-1]
arr=[5,3,2,8,0]
rev_arr(arr)
```

```python
# by swapping 
def rev_arr(arr):
    start=0
    end=len(arr)-1
    while start<end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
    return arr
arr=[2,6,4,2,3]
rev_arr(arr)
```

**Q-	Maximum-Subarray**

```python
Given an integer array nums, find the subarray with the largest sum, and return its sum.

Example 1:

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.
```

```python
def max_subarray(arr):
    max_sum=arr[0]
    current_sum=arr[0]
    for i in range(1,len(arr)):
        if current_sum+arr[i]>arr[i]:
            current_sum=current_sum+arr[i]
        else:
            current_sum=arr[i]
        if current_sum>max_sum:
            max_sum=current_sum
    return max_sum

arr=[-2,1,-3,4,-1,2,1,-5,4]
max_subarray(arr)
```

**Q-Contains Duplicate**

```python
def dup(arr):
    dict={}
    for i in arr:
        if i in dict:
            return True
        dict[i]=1
    return False
arr=[1,2,5,3]
dup(arr)
```

```python
# return true if any num array has dublicate elements
def contains_duplicate(arr):
    seen=set()
    for num in arr:
        if num in seen:
            return True
        seen.add(num)
    return False
arr=[1,2,3,5,3]
contains_duplicate(arr)
```

```python

def cont_dupl(arr):
    return len(set(arr))!=len(arr)

arr=[1,2,3]
cont_dupl(arr)
```

**Q-	Chocolate Distribution Problem**

```python
# arr, m is no of packets
def min_diff(arr,m):
    n=len(arr)
    min_diff=float("inf")
    for i in range(n-m+1):
        diff=arr[i+m-1]-arr[i]
        if diff<min_diff:
            min_diff=diff
    return min_diff

arr=[7, 3, 2, 4, 9, 12, 56]
m = 5 
min_diff(arr,m)
```

**Q-	Search in Rotated Sorted Array**

```python
def search(arr,target):
    start=0
    end=len(arr)-1
    while start<=end:
        mid=(start+end)//2
        if arr[mid]==target:
            return mid
        if arr[start]<=arr[mid]:
            if arr[start]<=target<arr[mid]:
                end=mid-1
            else:
                start=mid+1
        else:
            if arr[mid]<=target<arr[end]:
                start=mid+1
            else:
                end=end-1
            
    return -1
arr=[4,5,6,7,0,1,2]
search(arr,4)
```

**Q-	Next Permutation**

```python
def next_permutation(arr):
    n=len(arr)
    i=-1
    # finding the breakpoint 
    for k in range(n-2,0,-1):
        if arr[k]<arr[k+1]:
            i=k
            break
    # swapping the ith element with the smallest element larger then arr[i]
    if i!=-1:
        for j in range(n-1,i,-1):
            if arr[j]>arr[i]:
                arr[i],arr[j]=arr[j],arr[i]
    # reversing the rest array 
    start=i+1
    end=n-1
    while start<end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
    return arr

a=[1,2,3]
b=[3,2,1]
next_permutation(b)
```

**Q-	Best time to Buy and Sell Stock**

```python
def max_profit(arr):
    min_price=float("inf")
    max_profit=0
    for price in arr:
        if price<min_price:
            min_price=price
        profit=price-min_price
        if profit>max_profit:
            max_profit=profit
    return max_profit

prices = [7,1,5,3,6,4]
max_profit(prices)
```

**Q-	Repeat and Missing Number Array**

```python
def r_m_num(arr):
    n=len(arr)
    repeating=-1
    missing=-1
    for i in range(1,n+1):
        count=0
        for j in range(n):
            if arr[j]==i:
                count+=1
        if count==2:
            repeating=i
        if count==0:
            missing=i
        if repeating != -1 and missing !=-1:
            break
    return repeating,missing

a=[3,1,2,5,3]
b=[4,3,6,2,1,1]
c=[2,2]
r_m_num(c)
```

optimal solution 
We used basic math:

A - B = diff1

A^2 - B^2 = diff2

From that, found A + B = diff2 / diff1

Solved to find A (repeated number), B (missing number)

```python
def mismatch(arr):
    n=len(arr)
    s_n=n*(n+1)//2
    s_sq=n*(n+1)*(2*n+1)//6

    arr_s=0
    arr_s_sq=0
    for num in arr:
        arr_s+=num
        arr_s_sq+=num*num
    diff1=arr_s-s_n  #(A-B)
    diff2=arr_s_sq-s_sq  #(A^2-B^2)  > (A+B)(A-B)
    sum_ab=diff2//diff1 #(A+B)

    # for A > (A+B)+(A-B)=diff1+sum_ab ,, A=diff1+sum_ab)//2
    A=(diff1+sum_ab)//2
    B=A-diff1  # (A-B)=diff1
    return [A,B]

arr = [3, 1, 2, 5, 3]
mismatch(arr)
```
