```python
# multiple numbers
def gcd_n(arr):
    smallest=arr[0]
    for i in arr:
        if i<smallest:
            smallest=i

    for i in range(smallest,0,-1):
        is_gcd=True
        for j in arr:
            if j%i!=0:
                is_gcd=False
                break
        if is_gcd:
            return i

ar=[6,15,24]
gcd_n(ar)
```

**Q-Maximum and Minimum Element in an Array**

```python
def max_min(ar):
    max=ar[0]
    min=ar[0]
    for i in range(1,len(ar)):
        if ar[i]>max:
            max=ar[i]
        if ar[i]<min:
            min=ar[i]
    return max,min

ar=[3,25,23,1,44,56,0,454]
max_min(ar)
```

```python
def max_and_min(arr):
    min=arr[0]
    for i in range (1,len(arr)):
        if arr[i]<min:
            min=arr[i]
    max=arr[0]
    for i in range(1,len(arr)):
        if arr[i]>max:
            max=arr[i]
    return min,max

arr=[4,5,2,1,0]
max_and_min(arr)
```

**Q-	Maximum-Subarray**

```python
Given an integer array nums, find the subarray with the largest sum, and return its sum.

Example 1:

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.
```

```python
def max_subarray(arr):
    max_sum=arr[0]
    current_sum=arr[0]
    for i in range(1,len(arr)):
        if current_sum+arr[i]>arr[i]:
            current_sum=current_sum+arr[i]
        else:
            current_sum=arr[i]
        if current_sum>max_sum:
            max_sum=current_sum
    return max_sum

arr=[-2,1,-3,4,-1,2,1,-5,4]
max_subarray(arr)
```

```python
# arr, m is no of packets
def min_diff(arr,m):
    n=len(arr)
    min_diff=float("inf")
    for i in range(n-m+1):
        diff=arr[i+m-1]-arr[i]
        if diff<min_diff:
            min_diff=diff
    return min_diff

arr=[7, 3, 2, 4, 9, 12, 56]
m = 5 
min_diff(arr,m)
```

**Q-	Search in Rotated Sorted Array**

```python
def search(arr,target):
    start=0
    end=len(arr)-1
    while start<=end:
        mid=(start+end)//2
        if arr[mid]==target:
            return mid
        if arr[start]<=arr[mid]:
            if arr[start]<=target<arr[mid]:
                end=mid-1
            else:
                start=mid+1
        else:
            if arr[mid]<=target<arr[end]:
                start=mid+1
            else:
                end=end-1
            
    return -1
arr=[4,5,6,7,0,1,2]
search(arr,4)
```

**Q-	Next Permutation**

```python
def next_permutation(arr):
    n=len(arr)
    i=-1
    # finding the breakpoint 
    for k in range(n-2,0,-1):
        if arr[k]<arr[k+1]:
            i=k
            break
    # swapping the ith element with the smallest element larger then arr[i]
    if i!=-1:
        for j in range(n-1,i,-1):
            if arr[j]>arr[i]:
                arr[i],arr[j]=arr[j],arr[i]
    # reversing the rest array 
    start=i+1
    end=n-1
    while start<end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
    return arr

a=[1,2,3]
b=[3,2,1]
next_permutation(b)
```

**Q-	Best time to Buy and Sell Stock**

```python
def max_profit(arr):
    min_price=float("inf")
    max_profit=0
    for price in arr:
        if price<min_price:
            min_price=price
        profit=price-min_price
        if profit>max_profit:
            max_profit=profit
    return max_profit

prices = [7,1,5,3,6,4]
max_profit(prices)
```

**Q-	Repeat and Missing Number Array**

```python
def r_m_num(arr):
    n=len(arr)
    repeating=-1
    missing=-1
    for i in range(1,n+1):
        count=0
        for j in range(n):
            if arr[j]==i:
                count+=1
        if count==2:
            repeating=i
        if count==0:
            missing=i
        if repeating != -1 and missing !=-1:
            break
    return repeating,missing

a=[3,1,2,5,3]
b=[4,3,6,2,1,1]
c=[2,2]
r_m_num(c)
```

optimal solution 
We used basic math:

A - B = diff1

A^2 - B^2 = diff2

From that, found A + B = diff2 / diff1

Solved to find A (repeated number), B (missing number)

```python
def mismatch(arr):
    n=len(arr)
    s_n=n*(n+1)//2
    s_sq=n*(n+1)*(2*n+1)//6

    arr_s=0
    arr_s_sq=0
    for num in arr:
        arr_s+=num
        arr_s_sq+=num*num
    diff1=arr_s-s_n  #(A-B)
    diff2=arr_s_sq-s_sq  #(A^2-B^2)  > (A+B)(A-B)
    sum_ab=diff2//diff1 #(A+B)

    # for A > (A+B)+(A-B)=diff1+sum_ab ,, A=diff1+sum_ab)//2
    A=(diff1+sum_ab)//2
    B=A-diff1  # (A-B)=diff1
    return [A,B]

arr = [3, 1, 2, 5, 3]
mismatch(arr)
```

```python
# brute force method
def gcd(a,b):
    gcd=1
    min=a
    if b<a:
        min=b
    for i in range(2,min+1):
        if a%i ==0 and b%i==0:
            gcd=i
    return gcd
gcd(30,15)
```

```python
# euclidean algorithm
def gcd(a,b):
    while a>0 and b>0:
        if a>b:
            a=a%b
        else:
            b=b%b
    if a==0:
        return b
    return a
gcd(30,15)
```

```python
'''
Version	   Time Complexity	   Space Complexity	     Remarks
2nd	      O(log(min(a, b)))	           O(1)	           Best & cleanest
1st	      O(log(min(a, b)))	           O(1)	           Slightly more comparisons
'''
def gcd(a,b):
    while a!=0:
        c=a
        a=b%a
        b=c
    return b
gcd(20,15)
```

**Check if a number is Armstrong Number or not**

```python

def armstrong(n):
    cnt=int(math.log10(n)+1)
    sum=0
    temp=n
    while temp>0:
        rem=temp%10
        sum=sum+(rem**cnt)
        temp=temp//10
    if sum==n:
        return True
    else:
        return False
print(armstrong(153))
armstrong(371)
```

**Print all Divisors of a given Number**

```python
def divisors(n):
    div=[]
    for i in range(1,n//2+1):
        if n%i==0:
            div.append(i)
    div.append(n)
    return div
print(divisors(12))
divisors(36)
```

```python
# optimal solution
'''
Complexity Analysis

Time Complexity: O(sqrt(N)) where N is the input number. The algorithm iterates 
through each number from 1 to the square root of N once to check if it is a divisor.

Space Complexity : O(2*sqrt(N))where N is the input number. This approach allocates memory for an array to hold all the divisors. 
The size of this array could go to be 2*(sqrt(N)).
'''

def divi(n):
    div=[]
    for i in range(1,int(math.sqrt(n))+1):
        if n%i==0:
            div.append(i)
            if i!=n//i:
                div.append(n//i)
    return div

divi(36)
```

**Check if a number is prime or not**

```python
'''
Complexity Analysis

Time Complexity: O(sqrt(N))where N is the input number. The loop iterates up to the square root of n performing constant time operations at each step.

Space Complexity : O(1) as the space complexity remains constant and independent of the input size. Only a fixed amount of memory is required to store 
the integer variables.
'''
def is_prime(n):
    if n==1 or n==0:
        return False
    for i in range(2,n//2+1):
        if n%i==0:
            return False
    return True
is_prime(5)
```

```python
def c_freq(n):
    freq={}
    for i in n:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    return freq
a=[4,2,1,2,3,4,4,10,9,9,9,9,9]
c_freq(a)
```

```python
# using iterative method and boolean to print element and its count
def c_freq(n):
    ln=len(n)
    visited=[False]*ln
    for i in range(ln):
        if visited[i]==True:
            continue
        count=1
        for j in range(i+1,ln):
            if n[i]==n[j]:
                visited[j]=True
                count+=1
        print(n[i]," : ", count)
a=[4,2,1,2,3,4,4,10,9,9,9,9,9]
c_freq(a)
```

**Find the highest/lowest frequency element**

```python
def high_low(arr):
    freq=c_freq(arr)
    high=0
    low=float("inf")
    for i in freq.keys():
        if freq[i]<low:
            low=freq[i]
        if freq[i]>high:
            high=freq[i]
    high_f_el=[key for key,val in freq.items() if val==high]
    low_f_el=[key for key,val in freq.items() if val==low]
    print(high_f_el , " : ", high)
    print(low_f_el, " : ", low)
    return 
    

a=[4,2,1,2,3,4,4,10,9,9,9,9,9]
high_low(a)
```

**https://leetcode.com/problems/frequency-of-the-most-frequent-element/description/**

```python
def freq(n,k):
    n.sort()
    left=total=max_freq=0
    for i in range(len(n)):
        total+=n[i]
        while n[i]*(i-left+1)>total+k:
            total-=n[left]
            left+=1
        max_freq=max(max_freq,i-left+1)
    return max_freq
a=[1,2,4]
b=[3,9,7]
freq(b,2)
```

```python
# leetcode most optimal
class Solution:
    def maxFrequency(self, nums: List[int], k: int) -> int:
        i = 0
        nums.sort()

        for j in range(len(nums)):
            k += nums[j]

            if k < nums[j] * (j - i + 1):
                k -= nums[i]
                i += 1
        return j - i + 1
```

**Selection Sort Algorithm**

What is Selection Sort?

*-
-Selection Sort is a simple sorting algorithm that sorts an array by repeatedly finding the minimum element (for ascending order) from the unsorted part and moving it to the front.

```python
'''
⏱️ Time & Space Complexity
Time Complexity:
Worst/Average/Best = O(n²)
Because for every element, you're comparing it with all remaining elements.

Space Complexity:
O(1) (In-place sorting, no extra memory)

'''

def sele_sort(ar):
    for i in range(len(ar)):
        min_ind=i
        for j in range(i+1,len(ar)):
            if ar[j]<ar[min_ind]:
                min_ind=j
        ar[i],ar[min_ind]=ar[min_ind],ar[i]
    return ar

a=[5,3,1,4]
sele_sort(a)
```

**Bubble Sort Algorithm**

What is Bubble Sort?

-Bubble Sort is a simple sorting algorithm that repeatedly swaps adjacent elements if they are in the wrong order.

It’s called “bubble” sort because the largest element bubbles up to the end of the array in each pass.



```python
'''
📊 Time & Space Complexity:
Case	Time Complexity
Best Case	O(n) (with optimization)
Average Case	O(n²)
Worst Case	O(n²)
Space	O(1) (in-place)
'''
def bubble_sort(ar):
    for i in range(len(ar)-1,-1,-1):
        for j in range(i):
            if ar[j]>ar[j+1]:
                ar[j],ar[j+1]=ar[j+1],ar[j]
    return ar
a=[5,3,1,4]
bubble_sort(a)
```

```python
def bub_sort(ar):
    n=len(ar)
    for i in range(n):
        for j in range(n-i-1):
            if ar[j]>ar[j+1]:
                ar[j],ar[j+1]=ar[j+1],ar[j]
    return ar
a=[5,3,1,4]
bub_sort(a)
```

```python
Aspect	                                               Selection Sort	                                            Bubble Sort
How it works	         Finds the minimum (or maximum) and places it in the correct position	   Compares adjacent pairs and swaps them if out of order
Selection criteria	    Selects the min element in each outer loop and places it at beginning	   Bubbles the largest element to the end in each pass
Swaps per pass	         Only 1 swap per pass (after finding min)	                               Multiple swaps per pass (for every adjacent pair)
Best case optimization	 Cannot detect if the array is already sorted	                           Can be optimized to stop early if no swaps happen
Direction	             Sorts from start to end (puts smallest first)	                           Sorts from end to start (puts largest last)
```

**Insertion Sort Algorithm**

-Take the first value from the unsorted part of the array.

Move the value into the correct place in the sorted part of the array
.-
Go through the unsorted part of the array again as many times as there are values.

```python
'''
⏱️ Time Complexity
Case	Time
Best	O(n)
Average	O(n²)
Worst	O(n²)
'''
def insertion_sort(ar):
    n=len(ar)
    for i in range(1,n):
        curren_value=ar[i]
        insert_index=i
        for j in range(i-1,-1,-1):
            if ar[j]>curren_value:
                ar[j+1]=ar[j]
                insert_index=j
            else:
                break
        ar[insert_index]=curren_value
    return ar

my_array = [64, 34, 25, 12, 22, 11, 90, 5]
insertion_sort(my_array)
```

**Find the Largest element in an array**

```python
def largest(ar):
    max=ar[0]
    for i in range(1,len(ar)):
        if ar[i]>max:
            max=ar[i]
    return max
a=[4,5,3,1]
largest(a)
```

**Find Second Smallest and Second Largest Element in an array**

```python
'''
Complexity Analysis

Time Complexity: O(N), We do two linear traversals in our array

Space Complexity: O(1)
'''
def sec_s_l(arr):
    n=len(arr)
    smallest=arr[0]
    largest=arr[0]
    for i in range(1,n):
        if arr[i]>largest:
            largest=arr[i]
        if arr[i]<smallest:
            smallest=arr[i]
    sec_small=arr[0]
    sec_largest=arr[0]
    for i in range(n):
        if arr[i]>sec_largest and arr[i]<largest:
            sec_largest=arr[i]
        if arr[i]<sec_small and arr[i]>smallest:
            sec_small=arr[i]
    return sec_largest, sec_small
a=[2,4,1,2,3,5]
sec_s_l(a)
```

```python
# optimal approach , single pass solution
def sec(arr):
    small=float("inf")
    sec_small=float("inf")
    largest=float("-inf")
    sec_largest=float("-inf")
    n=len(arr)
    for i in range(n):
        if arr[i]<small:
            sec_small=small
            small=arr[i]
        elif arr[i]<sec_small and arr[i]!=small:
            sec_small=arr[i]
        if arr[i]>largest:
            sec_largest=largest
            largest=arr[i]
        elif arr[i]>sec_largest and arr[i]!=largest:
            sec_largest=arr[i]
    return sec_largest, sec_small

a=[5,2,4,3,1]
sec(a)
```

**Remove Duplicates in-place from Sorted Array**

```python
def rem(arr):
    st=set()
    for i in range(len(arr)):
        st.add(arr[i])
    k=len(st)
    j=0
    for l in st:
        arr[j]=l
        j+=1
    return arr, k
a=[1,1,2,3,3,4,5]
rem(a)
```

```python
'''optimal soltion
Solution 2: Two pointers
Intuition: We can think of using two pointers ‘i’ and ‘j’, we move ‘j’ till we don't get a number arr[j] which is different from arr[i]. As we got a unique number we will increase the i pointer and update its value by arr[j]. 

Approach:
Take a variable i as 0;
Use a for loop by using a variable ‘j’ from 1 to length of the array.
If arr[j] != arr[i], increase ‘i’ and update arr[i] == arr[j].
 After completion of the loop return i+1, i.e size of the array of unique elements.
'''

def rem_dup(arr):
    i=0
    for j in range(1,len(arr)):
        if arr[i]!=arr[j]:
            i+=1
            arr[i]=arr[j]
    return i+1
a=[1,1,2,3,3,4,5]
rem_dup(a)
```

**Left Rotate the Array by One**

```python
def left_one(arr):
    n=len(arr)
    temp=arr[0]
    for i in range(n-1):
        arr[i]=arr[i+1]
    arr[n-1]=temp
    return arr
a=[1,2,3,4,5]
left_one(a)
```

```python
# left rotate by kth elements
def lef_ro(arr,k):
    n=len(arr)
    k=k%n
    def reverse(arr,st,end):
        while st<end:
            arr[st],arr[end]=arr[end],arr[st]
            st+=1
            end-=1
    reverse(arr,0,n-1)
    reverse(arr,0,n-1-k)
    reverse(arr,n-k,n-1)
a=[1,2,3,4,5]
lef_ro(a,4)
a
```
