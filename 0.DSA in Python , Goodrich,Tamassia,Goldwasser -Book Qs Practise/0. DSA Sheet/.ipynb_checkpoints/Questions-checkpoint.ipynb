```python
# GCD of two numbers
def gcd(a,b):
    while b!=0:
        c=b
        b=a%b
        a=c
    return a

gcd(100,20)
```

```python
# multiple numbers
def gcd_n(arr):
    smallest=arr[0]
    for i in arr:
        if i<smallest:
            smallest=i

    for i in range(smallest,0,-1):
        is_gcd=True
        for j in arr:
            if j%i!=0:
                is_gcd=False
                break
        if is_gcd:
            return i

ar=[6,15,24]
gcd_n(ar)
```

```python
# is prime or not
# Use i * i <= n in code.
# Think in terms of √n in math.
def is_prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0:
        return False
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i +=2
    return True

is_prime(11)


# now optimize 6k +- 1 form 
# “After removing 2 and 3, only numbers shaped like 6k ± 1 can divide n, and i * i <= n will be followed here too, as there should be atleast one divisor
# less then sq.root of n
# only rule to remember, i = 5, i+2, then i += 6 walks through all numbers not divisible by 2 or 3 — nothing else is worth checking.
# below algo why we are starting with i = 5 and doing i +2
# 6k + 5 = 6k + 6 − 1
#        = 6(k + 1) − 1
# 6k+5=6(k+1)−1 

# What exactly are we SKIPPING?
# Between 6k − 1 and next 6(k+1) − 1 lie:
# Number	Why skipped
# 6k	divisible by 6 ❌
# 6k+1	checked
# 6k+2	even ❌
# 6k+3	divisible by 3 ❌
# 6k+4	even ❌
# 6k+5	checked (as next block’s −1)

# So we skip:
# 6k, 6k+2, 6k+3, 6k+4

def is_prime_opt(n):
    if n <=1:
        return False
    if n <=3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <=n:
        if n % i == 0 or n % (i+2) == 0:
            return False
        i += 6
    return True
is_prime_opt(10)
```

```python
# to display the distinct element of array
def distinct(li):
    new_list=[]
    for i in range(len(li)):
        count=0
        for j in range(len(li)):
            if li[i]==li[j]:
                count+=1
        if count==1:
            new_list.append(li[i])
    return new_list

lis=[4,5,4,6,7,5]
distinct(lis)
```

**Q-Maximum and Minimum Element in an Array**

```python
def max_min(ar):
    max=ar[0]
    min=ar[0]
    for i in range(1,len(ar)):
        if ar[i]>max:
            max=ar[i]
        if ar[i]<min:
            min=ar[i]
    return max,min

ar=[3,25,23,1,44,56,0,454]
max_min(ar)
```

**Q-	Maximum-Subarray**

```python
Given an integer array nums, find the subarray with the largest sum, and return its sum.

Example 1:

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.
```

```python
# At every index, you only ask one question:

# ❓ Is it better to
# extend the previous subarray
# OR
# start a new subarray from here?

def max_subarray(arr):
    max_sum=arr[0]
    current_sum=arr[0]
    for i in range(1,len(arr)):
        if current_sum+arr[i]>arr[i]:
            current_sum=current_sum+arr[i]
        else:
            current_sum=arr[i]
        if current_sum>max_sum:
            max_sum=current_sum
    return max_sum

arr=[-2,1,-3,4,-1,2,1,-5,4]
max_subarray(arr)
```

```python
# prefix sum
# example: arr = [3, -2, 5, -1]
# Prefix sum precomputes cumulative sums so that any subarray sum can be computed 
# using one subtraction.
# At each index of prefix sum, store the sum of all elements before it from the original array that's why it contains one element etc then original array.
# index:   0   1   2   3   4 
# prefix:  0   3   1   6   ]

#  WHY subtraction works (MOST IMPORTANT

# sum(i→j)=prefix[j+1]−prefix[i])

# prefix[4] = 3 + (-2) + 5 + (-1)
# prefix[1] = 3 # 
Subtrat: e# (3 + -2 + 5 + -1) - 3
# = -2 + 5 + -1 # 
🔥 Everything before index i is caned outc# l
* [ SUM BEFORE | SUBARRAY YOU WA]# * 
Subtracting removes the “sum before”d t.

```

```python
arr = [3, -2, 5, -1]

def prefix_sum(arr):
    prefix_sum = [0]

    for i in range(1,len(arr) + 1):
        cumulative_sum = 0
        for j in range(0,i):
            cumulative_sum += arr[j]
        prefix_sum.append(cumulative_sum)
    return prefix_sum

prefix_sum(arr)

def max_subarray_prefix(arr):
    n = len(arr)
    prefix = [0] * (n + 1)
    for i in range(n):
        prefix[i + 1] = prefix[i] + arr[i]

    max_sum = arr[0]
    for i in range(n):
        for j in range(i,n):
            current_sum = prefix[j+1] - prefix[i]
            if current_sum > max_sum:
                max_sum = current_sum
    return max_sum
a=[-2,1,-3,4,-1,2,1,4]
ar = [1,1,1,-100,100]
print(max_subarray_prefix(a))
print(max_subarray_prefix(arr))
max_subarray_prefix(ar)
```

**Q-Contains Duplicate**

```python
def dup(arr):
    dict={}
    for i in arr:
        if i in dict:
            return True
        dict[i]=1
    return False
arr=[1,2,5,3]
dup(arr)
```

```python
# return true if any num array has dublicate elements
def contains_duplicate(arr):
    seen=set()
    for num in arr:
        if num in seen:
            return True
        seen.add(num)
    return False
arr=[1,2,3,5,3]
contains_duplicate(arr)
```

**Q-	Chocolate Distribution Problem**

```python
# arr, m is no of packets
# arr should be sorted here
def min_diff(arr,m):
    n=len(arr)
    min_diff=float("inf")
    for i in range(n-m+1):
        diff=arr[i+m-1]-arr[i]
        if diff<min_diff:
            min_diff=diff
    return min_diff

arr=[7, 3, 2, 4, 9, 12, 56]
m = 5 
min_diff(arr,m)
```

```python
def choc_dist(arr,m):
    n = len(arr)
    for i in range(n):
        smallest = arr[i]
        s_index = i
        for j in range(i,n):
            if arr[j] < smallest:
                smallest = arr[j]
                s_index = j
        arr[i], arr[s_index] = arr[s_index], arr[i]

    min_diff = float("inf")
    for i in range(n - m +1):
        diff = arr[i + m -1] - arr[i]
        if diff < min_diff:
            min_diff = diff
    return min_diff
arr=[12, 4, 7, 9, 2, 23]
m = 3 
choc_dist(arr,m)

# used selection sort here Sorting (Selection Sort)
# Selection Sort: O(n²),  Sliding window:  O(n),  Total: O(n²)
```

**Q-	Search target element index in Rotated Sorted Array**

```python
def search(arr,target):
    start=0
    end=len(arr)-1
    while start<=end:
        mid=(start+end)//2
        if arr[mid]==target:
            return mid
        if arr[start]<=arr[mid]:
            if arr[start]<=target<arr[mid]:
                end=mid-1
            else:
                start=mid+1
        else:
            if arr[mid]<=target<arr[end]:
                start=mid+1
            else:
                end=end-1
            
    return -1
arr=[4,5,6,7,0,1,2]
search(arr,4)

ar = [6,7,8,1,2,3,4,5]
search(ar,8)
```

**Q-	Next Permutation**

```python
def next_permutation(arr):
    n=len(arr)
    i=-1
    # finding the breakpoint 
    for k in range(n-2,0,-1):
        if arr[k]<arr[k+1]:
            i=k
            break
    # swapping the ith element with the smallest element larger then arr[i]
    if i!=-1:
        for j in range(n-1,i,-1):
            if arr[j]>arr[i]:
                arr[i],arr[j]=arr[j],arr[i]
    # reversing the rest array 
    start=i+1
    end=n-1
    while start<end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
    return arr

a=[1,2,3]
b=[3,2,1]
next_permutation(b)
```

**Q-	Best time to Buy and Sell Stock**

```python
def max_profit(arr):
    min_price=float("inf")
    max_profit=0
    for price in arr:
        if price<min_price:
            min_price=price
        profit=price-min_price
        if profit>max_profit:
            max_profit=profit
    return max_profit

prices = [7,1,5,3,6,4]
max_profit(prices)
```

**Q-	Repeat and Missing Number Array**

```python
def r_m_num(arr):
    n=len(arr)
    repeating=-1
    missing=-1
    for i in range(1,n+1):
        count=0
        for j in range(n):
            if arr[j]==i:
                count+=1
        if count==2:
            repeating=i
        if count==0:
            missing=i
        if repeating != -1 and missing !=-1:
            break
    return repeating,missing

a=[3,1,2,5,3]
b=[4,3,6,2,1,1]
c=[2,2]
r_m_num(c)
```

optimal solution 
We used basic math:

A - B = diff1

A^2 - B^2 = diff2

From that, found A + B = diff2 / diff1

Solved to find A (repeated number), B (missing number)

```python
def mismatch(arr):
    n=len(arr)
    s_n=n*(n+1)//2
    s_sq=n*(n+1)*(2*n+1)//6

    arr_s=0
    arr_s_sq=0
    for num in arr:
        arr_s+=num
        arr_s_sq+=num*num
    diff1=arr_s-s_n  #(A-B)
    diff2=arr_s_sq-s_sq  #(A^2-B^2)  > (A+B)(A-B)
    sum_ab=diff2//diff1 #(A+B)

    # for A > (A+B)+(A-B)=diff1+sum_ab ,, A=diff1+sum_ab)//2
    A=(diff1+sum_ab)//2
    B=A-diff1  # (A-B)=diff1
    return [A,B]

arr = [3, 1, 2, 5, 3]
mismatch(arr)
```

**Check if a number is Palindrome or Not**

```python
def palindrome(n):
    temp=n
    rev_num=0
    while temp>0:
        ld=temp%10
        rev_num=(rev_num*10)+ld
        temp=temp//10
    if rev_num==n:
        return True
    else:
        return False

palindrome(1221)
```

**Find GCD of two numbers**

```python
# brute force method
def gcd(a,b):
    gcd=1
    min=a
    if b<a:
        min=b
    for i in range(2,min+1):
        if a%i ==0 and b%i==0:
            gcd=i
    return gcd
gcd(30,15)
```

```python
# reverse itrating
def gcd(a,b):
    gcd=1
    min=a
    if b<a:
        min=b
    for i  in range(min,1,-1):
        if a%i ==0 and b%i ==0:
            gcd=i
            return gcd
    return gcd
gcd(30,15)
```

```python
# euclidean algorithm
def gcd(a,b):
    while a>0 and b>0:
        if a>b:
            a=a%b
        else:
            b=b%b
    if a==0:
        return b
    return a
gcd(30,15)
```

```python
'''
Version	   Time Complexity	   Space Complexity	     Remarks
2nd	      O(log(min(a, b)))	           O(1)	           Best & cleanest
1st	      O(log(min(a, b)))	           O(1)	           Slightly more comparisons
'''
def gcd(a,b):
    while a!=0:
        c=a
        a=b%a
        b=c
    return b
gcd(20,15)
```

**Check if a number is Armstrong Number or not**

```python

def armstrong(n):
    cnt=int(math.log10(n)+1)
    sum=0
    temp=n
    while temp>0:
        rem=temp%10
        sum=sum+(rem**cnt)
        temp=temp//10
    if sum==n:
        return True
    else:
        return False
print(armstrong(153))
armstrong(371)
```

**Print all Divisors of a given Number**

```python
def divisors(n):
    div=[]
    for i in range(1,n//2+1):
        if n%i==0:
            div.append(i)
    div.append(n)
    return div
print(divisors(12))
divisors(36)
```

```python
# optimal solution
'''
Complexity Analysis

Time Complexity: O(sqrt(N)) where N is the input number. The algorithm iterates 
through each number from 1 to the square root of N once to check if it is a divisor.

Space Complexity : O(2*sqrt(N))where N is the input number. This approach allocates memory for an array to hold all the divisors. 
The size of this array could go to be 2*(sqrt(N)).
'''

def divi(n):
    div=[]
    for i in range(1,int(math.sqrt(n))+1):
        if n%i==0:
            div.append(i)
            if i!=n//i:
                div.append(n//i)
    return div

divi(36)
```

**Check if a number is prime or not**

```python
'''
Complexity Analysis

Time Complexity: O(sqrt(N))where N is the input number. The loop iterates up to the square root of n performing constant time operations at each step.

Space Complexity : O(1) as the space complexity remains constant and independent of the input size. Only a fixed amount of memory is required to store 
the integer variables.
'''
def is_prime(n):
    if n==1 or n==0:
        return False
    for i in range(2,n//2+1):
        if n%i==0:
            return False
    return True
is_prime(5)
```

**Selection Sort Algorithm**

What is Selection Sort?

*-
-Selection Sort is a simple sorting algorithm that sorts an array by repeatedly finding the minimum element (for ascending order) from the unsorted part and moving it to the front.

```python
'''
⏱️ Time & Space Complexity
Time Complexity:
Worst/Average/Best = O(n²)
Because for every element, you're comparing it with all remaining elements.

Space Complexity:
O(1) (In-place sorting, no extra memory)

'''

def sele_sort(ar):
    for i in range(len(ar)):
        min_ind=i
        for j in range(i+1,len(ar)):
            if ar[j]<ar[min_ind]:
                min_ind=j
        ar[i],ar[min_ind]=ar[min_ind],ar[i]
    return ar

a=[5,3,1,4]
sele_sort(a)
```

**Bubble Sort Algorithm**

What is Bubble Sort?

-Bubble Sort is a simple sorting algorithm that repeatedly swaps adjacent elements if they are in the wrong order.

It’s called “bubble” sort because the largest element bubbles up to the end of the array in each pass.



```python
'''
📊 Time & Space Complexity:
Case	Time Complexity
Best Case	O(n) (with optimization)
Average Case	O(n²)
Worst Case	O(n²)
Space	O(1) (in-place)
'''
def bub_sort(ar):
    n=len(ar)
    for i in range(n):
        for j in range(n-i-1):
            if ar[j]>ar[j+1]:
                ar[j],ar[j+1]=ar[j+1],ar[j]
    return ar
a=[5,3,1,4]
bub_sort(a)
```

```python
#best case
def bub_sort(ar):
    n=len(ar)
    swap=0
    for i in range(n):
        for j in range(n-i-1):
            if ar[j]>ar[j+1]:
                ar[j],ar[j+1]=ar[j+1],ar[j]
                swap+=1
        if swap==0:
            return ar
    return ar,swap
a=[5,3,1,4]
bub_sort(a)
```

```python
Aspect	                                               Selection Sort	                                            Bubble Sort
How it works	         Finds the minimum (or maximum) and places it in the correct position	   Compares adjacent pairs and swaps them if out of order
Selection criteria	    Selects the min element in each outer loop and places it at beginning	   Bubbles the largest element to the end in each pass
Swaps per pass	         Only 1 swap per pass (after finding min)	                               Multiple swaps per pass (for every adjacent pair)
Best case optimization	 Cannot detect if the array is already sorted	                           Can be optimized to stop early if no swaps happen
Direction	             Sorts from start to end (puts smallest first)	                           Sorts from end to start (puts largest last)
```

**Insertion Sort Algorithm**

-Take the first value from the unsorted part of the array.

Move the value into the correct place in the sorted part of the array
.-
Go through the unsorted part of the array again as many times as there are values.

```python
'''
⏱️ Time Complexity
Case	Time
Best	O(n)
Average	O(n²)
Worst	O(n²)
'''
def insertion_sort(ar):
    n=len(ar)
    for i in range(1,n):
        curren_value=ar[i]
        insert_index=i
        for j in range(i-1,-1,-1):
            if ar[j]>curren_value:
                ar[j+1]=ar[j]
                insert_index=j
            else:
                break
        ar[insert_index]=curren_value
    return ar

my_array = [64, 34, 25, 12, 22, 11, 90, 5]
insertion_sort(my_array)
```

**Merge Sort Algorithm**

```python
'''
⬇️ Going Down the Stack (recursive calls)
merge_sort([4, 2, 1, 3])
├── merge_sort([4, 2])
│   ├── merge_sort([4]) → returns [4]
│   └── merge_sort([2]) → returns [2]
│   └── merge([4], [2]) → returns [2, 4]
├── merge_sort([1, 3])
│   ├── merge_sort([1]) → returns [1]
│   └── merge_sort([3]) → returns [3]
│   └── merge([1], [3]) → returns [1, 3]
└── merge([2, 4], [1, 3]) → returns [1, 2, 3, 4]




⌛ Time and Space Complexity :
Time: O(n log n) — because each split is log n levels deep, and merging at each level takes O(n)

Space: O(n) — because we need extra space to store the merged results
'''
def merge_sort(arr):
    if len(arr)<=1:
        return arr
    mid=len(arr)//2
    left=merge_sort(arr[:mid])
    right=merge_sort(arr[mid:])
    return merge(left,right)

def merge(left,right):
    i=j=0
    result=[]
    while i <len(left) and j <len(right):
        if left[i]<= right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1
    while i < len(left):
        result.append(left[i])
        i+=1
    while j < len(right):
        result.append(right[j])
        j+=1
    return result

a=[4,2,1,3]
merge_sort(a)


'''
🧠 Summary
Best-Case Time Complexity: O(n) (with optimization for already sorted data)

Average and Worst-Case Time Complexity: O(n log n)

Space Complexity: O(n)​
'''

def ms(ar):
    n=len(ar)
    if n <=1:
        return ar
    mid=n//2
    left = ms(ar[:mid])
    right = ms(ar[mid:])
    if left[-1]<=right[0]:
        return left + right
    return merge(left,right)

def mrg(left,right):
    result=[]
    i=j=0
    while i<len(left) and j<len(right):
        if left[i]<=right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1
    while i < len(left):
        result.append(left[i])
        i+=1
    while j< len(right):
        result.append(right[j])
        j+=1
    return result
b=[5,4,1,3,2]
c=[1,2,3,4]
ms(c)
```

**Quick Sort**

```python
'''
qs([4, 3, 1, 2], 0, 3)
│
├── partition → [1, 2, 4, 3] → pivot_index = 1
│
├── qs([1, 2, 4, 3], 0, 0) → base case → returns [1, 2, 4, 3]
│
└── qs([1, 2, 4, 3], 2, 3)
    │
    ├── partition → [1, 2, 3, 4] → pivot_index = 2
    │
    ├── qs([1, 2, 3, 4], 2, 1) → base case → returns [1, 2, 3, 4]
    │
    └── qs([1, 2, 3, 4], 3, 3) → base case → returns [1, 2, 3, 4]

'''

def qs(ar,low,high):
    if low<high:
        partition_ind=partition(ar,low,high)
        qs(ar,low,partition_ind-1)
        qs(ar,partition_ind+1,high)
    return ar

def partition(ar,low,high):
    pivot_element=ar[high]
    i=low-1
    for j in range(low,high):
        if ar[j]<pivot_element:
            i+=1
            ar[i],ar[j]=ar[j],ar[i]
    ar[i+1],ar[high]=ar[high],ar[i+1]
    return i+1
a=[4,3,1,2]
qs(a,0,len(a)-1)
```

```python
🔎 Time Complexity Analysis
👀 Partition Step:
The partition function loops once from low to high-1.

It takes O(n) time per call (where n = high - low + 1).

🧠 Now analyze the recursion tree:
Let’s say we are sorting an array of size n.

✅ Best Case (Balanced Partition):
Pivot always splits array into equal halves.

Each level does O(n) work, and there are log₂n levels.

⏱ Time Complexity: O(n log n)

📊 Average Case:
Pivot doesn't perfectly split, but reasonably balanced (e.g., 30%-70% split).

Still O(n) per level × O(log n) levels.

⏱ Time Complexity: O(n log n)

❌ Worst Case (Unbalanced Partition):
Pivot is always the smallest/largest element.

Array becomes: [1, 2, 3, ..., n] (sorted or reverse sorted)

Recursive calls look like:

qs(n)

qs(n-1)

qs(n-2)

...

qs(1)

This is like a skewed tree of height n.

⏱ Time Complexity: O(n²)

📦 Space Complexity
🔹 In-place Sort:
No extra array or data structure used → O(1) auxiliary space

🔹 But Recursion Stack?
Best/Average Case: Height of tree is O(log n)

Worst Case: Height of tree is O(n)
```

**Find Second Smallest and Second Largest Element in an array**

```python
'''
Complexity Analysis

Time Complexity: O(N), We do two linear traversals in our array

Space Complexity: O(1)
'''
def sec_s_l(arr):
    n=len(arr)
    smallest=arr[0]
    largest=arr[0]
    for i in range(1,n):
        if arr[i]>largest:
            largest=arr[i]
        if arr[i]<smallest:
            smallest=arr[i]
    sec_small=float("inf")
    sec_largest=float("-inf")
    for i in range(n):
        if arr[i]>sec_largest and arr[i]<largest:
            sec_largest=arr[i]
        if arr[i]<sec_small and arr[i]>smallest:
            sec_small=arr[i]
    return sec_largest, sec_small
a=[2,4,1,2,3,5]
sec_s_l(a)
```

```python
# optimal approach , single pass solution
def sec(arr):
    small=float("inf")
    sec_small=float("inf")
    largest=float("-inf")
    sec_largest=float("-inf")
    n=len(arr)
    for i in range(n):
        if arr[i]<small:
            sec_small=small
            small=arr[i]
        elif arr[i]<sec_small and arr[i]!=small:
            sec_small=arr[i]
        if arr[i]>largest:
            sec_largest=largest
            largest=arr[i]
        elif arr[i]>sec_largest and arr[i]!=largest:
            sec_largest=arr[i]
    return sec_largest, sec_small

a=[5,2,4,3,1]
sec(a)
```

```python
'''
It creates a circular check like this:
[2, 1, 3, 4]
  ↑         ↑
 i=0     i=3, compare to i+1=0 using modulo
So we connect the last element back to the first and make it a circle.

🧠 Purpose of this Code?
It counts how many times arr[i] > arr[i+1] (even including end to start):

If only 1 or 0 violations, then it's rotated sorted

More than 1 = not rotated sorted
'''


'''
rotated sorted -- arr = [3, 4, 5, 1, 2]
'''
def is_sorted(arr):
    order_violate=0
    for i in range(len(arr)):
        if arr[i]>arr[(i+1)%len(arr)]:
            order_violate+=1
    return order_violate <=1
a=[2,1,3,4]
is_sorted(a)
```

**Remove Duplicates in-place from Sorted Array**

```python
def rem(arr):
    st=set()
    for i in range(len(arr)):
        st.add(arr[i])
    k=len(st)
    j=0
    for l in st:
        arr[j]=l
        j+=1
    return arr, k
a=[1,1,2,3,3,4,5]
rem(a)
```

```python
'''optimal soltion
Solution 2: Two pointers
Intuition: We can think of using two pointers ‘i’ and ‘j’, we move ‘j’ till we don't get a number arr[j] which is different from arr[i]. 
As we got a unique number we will increase the i pointer and update its value by arr[j]. 

Approach:
Take a variable i as 0;
Use a for loop by using a variable ‘j’ from 1 to length of the array.
If arr[j] != arr[i], increase ‘i’ and update arr[i] == arr[j].
 After completion of the loop return i+1, i.e size of the array of unique elements.
'''

def rem_dup(arr):
    i=0
    for j in range(1,len(arr)):
        if arr[i]!=arr[j]:
            i+=1
            arr[i]=arr[j]
    return i+1
a=[1,1,2,3,3,4,5]
rem_dup(a)
```

**Move all Zeros to the end of the array**

```python
# Optimal Approach(Using 2 pointers): 
'''
🧠 Time Complexity
First loop (for j in range(n)):
Scans the whole array once → O(n)

Second loop (for i in range(i, n)):
At most fills n positions → O(n) in worst case

But the two loops are separate and linear, so:

✅ Total Time Complexity = O(n)
(where n = len(ar))

🧠 Space Complexity
No extra data structure is used.

All changes are made in-place inside the original array.

✅ Space Complexity = O(1)
'''
def move_zer(ar):
    i=0
    n=len(ar)
    for j in range(n):
        if ar[j]!=0:
            ar[i]=ar[j]
            i+=1
    for i in range(i,n):
        ar[i]=0
    return ar
a=[1,0,0,1,2,0,3,0]
move_zer(a)
```

**Union of Two Sorted Arrays**

```python
# This function will only work correctly if both ar1 and ar2 are sorted.
def union(ar1,ar2):
    i=j=0
    union=[]
    while i<len(ar1) and j<len(ar2):
        if ar1[i]<=ar2[j]:
            if len(union)==0 or union[-1]!=ar1[i]:
                union.append(ar1[i])
            i+=1
        else:
            if len(union)==0 or union[-1]!=ar2[j]:
                union.append(ar2[j])
            j+=1
    while i<len(ar1):
        if union[-1]!=ar1[i]:
            union.append(ar1[i])
        i+=1
    while j<len(ar2):
        if union[-1]!=ar2[j]:
            union.append(ar2[j])
        j+=1    
    return union
a=[1,2,5,7]
b=[1,2,3]
union(a,b)
```

**Find the number that appears once, and the other numbers twice**

```python
'''
This function has a time complexity of O(n), where n is the number of elements in the array. 
This is because we traverse the list once to count frequencies, and once again through the dictionary to find the first unique element.

The space complexity is O(n) as well, since we use a dictionary to store the count of each unique element in the array.

The dictionary operations are on average O(1) due to Python's efficient hash table implementation.
'''
def freq_one(arr):
    freq={}
    for i in arr:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    for key, value in freq.items():
        if value==1:
            return key
    return -1

a= [4,1,2,1,2]
freq_one(a)
```

**Longest Subarray with given Sum K(Positives and negatives)**

```python
'''
Time Complexity: O(N2) approx., where N = size of the array.
Reason: We are using two nested loops, each running approximately N times.

Space Complexity: O(1) as we are not using any extra space.
'''

def lon_sub(arr,k):
    s=0
    n=len(arr)
    length=0
    for i in range(n):
        s=0
        for j in range(i,n):
            s+=arr[j]
            if s==k:
                count=j-i+1
                if count>length:
                    length=count
    return length

a=[2,3,5,1,9]
b=[2,2,1]
c=[-1,1,1]
lon_sub(c,1)
```

```python
'''
Optimal Approach (Using 2 pointers): 

Time Complexity: O(2*N), where N = size of the given array.
Reason: The outer while loop i.e. the right pointer can move up to index n-1(the last index). Now, the inner while loop i.e. the left pointer can move up to the right pointer at most. So, every time the inner loop does not run for n times rather it can run for n times in total. So, the time complexity will be O(2*N) instead of O(N2).

Space Complexity: O(1) as we are not using any extra space.

Space Complexity: O(1) as we are not using any extra space.
'''

def sub(arr,k):
    left=right=0
    n=len(arr)
    max_len=0
    sum=arr[0]
    while right < n:
        while left <= right and sum >k:
            sum-=arr[left]
            left+=1
        if sum==k:
            max_len=max(max_len,right-left+1)
        right+=1
        if right < n:
            sum+= arr[right]
    return max_len
a=[2,3,5,1,9]
b=[2,2,1]
c=[-1,1,1]
sub(c,1)
```

```python
def fun(st):
    opened=0
    result=[]
    for p in st:
        if p=="(" and opened >0:
            result.append(p)
        if p==")" and opened >1:
            result.append(p)
        if p=="(":
            opened+=1
        else:
            opened-=1
    return "".join(result)
st="(()()())(())"
fun(st)
```

**Reverse Words in a String**

```python
def fun(s):
    n=len(s)
    words=[]
    i=0
    while i<n:
        if s[i]==" ":
            i+=1
            continue
        word=""
        while i<n and s[i]!=" ":
            word+=s[i]
            i+=1
        words.append(word)
    reversed_string=""
    for j in range(len(words)-1,-1,-1):
        reversed_string +=words[j]
        if j!=0:
            reversed_string+=" "
    return reversed_string
s="the sky is blue"
fun(s)
```

**Largest odd number in a string**

```python
def fun(st):
    for i in range(len(st)-1,-1,-1):
        if int(st[i])%2==1:
            return st[:i+1]
    return ""
st='5234'
s="2222"
fun(s)
```

**Longest Common Prefix**

```python
#Input : str = ["flowers" , "flow" , "fly", "flight" ]
#Output : "fl"
def fun(st):
    ans=""
    st=sorted(st)
    first=st[0]
    last=st[-1]
    for i in range(min(len(first),len(last))):
        if first[i]==last[i]:
            ans+=first[i]
    return ans
input =["flowers" , "flow" , "fly", "flight" ]
fun(input)  
```

```python
def freq(ar):
    fre={}
    for i in ar:
        if i not in fre:
            fre[i]=1
        else:
            fre[i]+=1
    return fre
ar = "vishal nishad"
freq(ar)
```

```python
def rd(st):
    result=""
    seen=set()
    for i in st:
        if i not in seen:
            result+=i
            seen.add(i)
    return result
s="aabbccd"
rd(s)
```

```python
def fst(st):
    freq={}
    for i in st:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    for i in st:
        if freq[i]==1:
            return i

a="aabbcddef"
fst(a)
```

```python
def st_rev(st):
    a=""
    for i in st:
        a=i+a
    return a
a="vishal"
st_rev(a)
    
```

```python
def power(x,n):
    ans=x
    for i in range(n-1):
        ans*=x
    return ans
power(2,3)
```

```python
def rec(x,n):
    if n==1 :
        return x
    return x*rec(x,n-1)
rec(2,3)
```

```python
def union(a,b):
    un=[]
    for i in range(len(a)):
        if a[i] not in un:
            un.append(a[i])
    for j in range(len(b)):
        if b[j] not in un:
            un.append(b[j])
    return un

a=[1,2,3]
b=[3,4,5]
union(a,b)
```

```python
def inter(a,b):
    int=set()
    for i in range(len(a)):
        for j in range(len(b)):
            if a[i]==b[j]:
                int.add(a[i])

    return list(int)
a=[1,2,3]
b=[3,4,5]
inter(a,b)
```
