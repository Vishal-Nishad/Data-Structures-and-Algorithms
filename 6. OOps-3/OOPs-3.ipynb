## Abstract classes
**Abstract classes have abstract methods i.e Declared, but they have no implementation ,, to be implemented in extended class i.e child class**

1. Object of abstract class cannot be created.
2. Implementation of all the abstract methods is mandatory in child class.

```python
from abc import ABC, abstractmethod
class Automobile(ABC):
    def __init__(self):
        print("Automobile created.")
    @abstractmethod
    def start(self):
        pass
    @abstractmethod    
    def stop(self):
        pass
    @abstractmethod
    def drive(self):
        pass

# need to imlement those methods in child class, cant create object directly from that abstract class

class Car(Automobile):
    # def __init__(self,name):
    #     print("Car created.")
    #     self.name=name

    def start(self):
        pass
    def stop(self):
        pass
    def drive(self):
        pass
```
