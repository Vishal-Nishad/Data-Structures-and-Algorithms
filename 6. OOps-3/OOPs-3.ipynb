## Abstract classes
**Abstract classes have abstract methods i.e Declared, but they have no implementation ,, to be implemented in extended class i.e child class**

1. Object of abstract class cannot be created.
2. Implementation of all the abstract methods is mandatory in child class.

```python
from abc import ABC, abstractmethod
class Automobile(ABC):
    def __init__(self):
        print("Automobile created.")
    @abstractmethod
    def start(self):
        pass
    @abstractmethod    
    def stop(self):
        pass
    @abstractmethod
    def drive(self):
        pass

# need to imlement those methods in child class, cant create object directly from that abstract class

class Car(Automobile):
    # def __init__(self,name):
    #     print("Car created.")
    #     self.name=name

    def start(self):
        pass
    def stop(self):
        pass
    def drive(self):
        pass
```

## Exception Handling

```python
try:
    m=int(input("Enter the numerator:"))
    n=int(input("Enter the denominator:"))
    output=m/n
    print(output)
except ValueError:
    print("Numerator and Denominator should be integer.")
```

```python
try:
    a=5
    b=0
    output=a/b
    print(output)
except ValueError:
    print("Exception Occured")
```

```python
try:
    a=5
    b=0
    output=a/b
    print(output)
except Exception as e:  # Catches all exceptions
    print("Exception Occurred: ",e)
```

## Custom Exceptions

```python
try:
    m=int(input("Enter the numerator:"))
    n=int(input("Enter the denominator:"))
    if n==0:
        raise ValueError # we can throw any exceptin i.e custom-user defined
    output=m/n
    print(output)
except ValueError:
    print("num and den should be integer.")
except ZeroDivisionError:
    print("division by zero is not allowed.")
```

```python
# Creating a custom Exception class
class ZeroDenominatorError(Exception):
    pass
try:
    m=int(input("Enter the numerator:"))
    n=int(input("Enter the denominator:"))
    if n==0:
        raise ZeroDenominatorError("Denominator should not be zero.")
    output=m/n
    print(output)
except ValueError:
    print("num and den should be integer.")
except ZeroDivisionError:
    print("division by zero is not allowed.")
except ZeroDenominatorError:
    print("ZeroDenominatorError is raised.")
```

```python
# order of writing 
class ZeroDenominatorError(ZeroDivisionError):
    pass
try:
    m=int(input("Enter the numerator:"))
    n=int(input("Enter the denominator:"))
    if n==0:
        raise ZeroDenominatorError("Denominator should not be zero.")
    output=m/n
    print(output)
except ValueError:
    print("num and den should be integer.")
except ZeroDivisionError:
    print("division by zero is not allowed.")
except ZeroDenominatorError:
    print("ZeroDenominatorError is raised.")
except:  #for any other expception which are not specified above , should be in last
    print("some exception is raised")
```

```python
## else and finally
# else: will be executed only if there is no exception raised
# finally: will always get executed whether any exceptions may raised or not

class ZeroDenominatorError(ZeroDivisionError):
    pass
try:
    m=int(input("Enter the numerator:"))
    n=int(input("Enter the denominator:"))
    if n==0:
        raise ZeroDenominatorError("Denominator should not be zero.")
    output=m/n
except ValueError:
    print("num and den should be integer.")
except ZeroDivisionError:
    print("division by zero is not allowed.")
except ZeroDenominatorError:
    print("ZeroDenominatorError is raised.")
except:  #for any other expception which are not specified above , should be in last
    print("some exception is raised")

else:
    print("Output : ",output)
finally:
    print("Hey, I am finally block.")
```
