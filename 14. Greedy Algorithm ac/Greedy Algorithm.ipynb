# Greedy Algorithm
- It is an algorithmic paradigm that builds the solution piece by piece
- In each step it chooses the piece that offers most obvious and immediate benefit
- it fits perfectly for those solutions in which local optimal solutions lead to global solution

## Some Greedy Alogrithm are-
- Insertion Sort
- Selection Sort
- Topological Sort
- Prim's Algorithm
- Kruskal Algorithm

2. **https://www.geeksforgeeks.org/problems/fractional-knapsack-1587115620/1**

```python
# time complexity = nlogn
# space complexity = n
def fractionalKnapsack(val, wt, capacity):
        #code here
        ratio = []
        for i in range(len(val)):
            ratio.append([val[i]/wt[i],i])
        
        ratio.sort(key = lambda x:x[0])
        
        ratio.reverse()
        
        max_value = 0
        
        j = 0
        while (capacity > 0 and j < len(ratio)):
            if wt[ratio[j][1]] <= capacity:
                max_value += val[ratio[j][1]]
                capacity -= wt[ratio[j][1]]
                j += 1
            else:
                max_value += ratio[j][0]*capacity
                capacity -= capacity
        return max_value

val = [60, 100, 120]
wt = [10, 20, 30]
capacity = 50
fractionalKnapsack(val,wt,capacity)
```

3. **https://www.geeksforgeeks.org/problems/minimum-sum-of-absolute-differences-of-pairs/1**

```python
# time = nlogn
# space = 1

def findMinSum(A,B,N):
        A.sort()
        B.sort()
        min_dif = 0
        for i in range(N):
            min_dif += abs(A[i]-B[i])
        return min_dif

N = 4
A = [4,1,8,7]
B = [2,3,6,5]
findMinSum(A,B,N)
```

```python
def findMin(n):
       # code here 
        count = 0
        if n>=10:
           count += n//10
           n = n%10
        if n>=5:
            count += n//5
            n = n%5
        if n>=2:
            count += n//2
            n = n%2
        count += n
        return count

n = 121
findMin(n)
```

🔴 6. **https://www.geeksforgeeks.org/problems/job-sequencing-problem-1587115620/1**

```python
# another brute greedy
# time n2
# space n
def jobSequencing(deadline, profit):
        # code here
        dln_pr = []
        for i in range(len(profit)):
            dln_pr.append([deadline[i],profit[i]])
            
        dln_pr.sort(key=lambda x:x[1])
        dln_pr.reverse()
        
        slots = [False]* (max(deadline)+1)
        count = 0
        m_profit = 0
        for i in range(len(dln_pr)):
            for j in range(dln_pr[i][0],0,-1):
                if not slots[j]:
                    slots[j] = True
                    count += 1
                    m_profit += dln_pr[i][1]
                    break
        return [count,m_profit]

deadline = [4, 1, 1, 1]
profit = [20, 10, 40, 30]
jobSequencing(deadline,profit)
```
