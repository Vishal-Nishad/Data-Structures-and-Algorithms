```python
class Node:
    def __init__(self,value = None):
        self.value = value
        self.prev = None
        self.next = None

class DoublyLL:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
    def __iter__(self):
        curr = self.head
        while curr:
            yield curr
            curr = curr.next
    def create_dll(self,value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.size = 1
        return "DLL CREATED SUCCESSFULLY"

    def insertion_at_beginning(self,value):
        if self.head is None:
            return self.create_dll(value)
        else:
            new_node = Node(value)
            new_node.next = self.head
            new_node.prev = None
            self.head.prev = new_node
            self.head = new_node
            self.size += 1
    def insertion_at_end(self,value):
        if self.head is None:
            return self.create_dll(value)
        else:
            new_node = Node(value)
            new_node.next = None
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node
            self.size += 1
    def insertion_at_middle(self,value,position):
        if position < 0 or position > self.size:
            raise IndexError("Index out of bound")
        if self.head is None:
            return self.create_dll(value)
        if position == 0:
            return self.insertion_at_beginning(value)
        elif position == self.size:
            return self.insertion_at_end(value)
        else:
            new_node = Node(value)
            tempnode = self.head
            index = 0
            while index < position - 1:
                tempnode = tempnode.next
                index += 1
            nextnode = tempnode.next
            new_node.prev = tempnode
            tempnode.next = new_node
            new_node.next = nextnode
            nextnode.prev = new_node
            self.size += 1
    def traverse_dll(self):
        curr = self.head
        while curr:
            print(curr.value)
            curr = curr.next
    def reverse_traverse_dll(self):
        curr = self.tail
        while curr:
            print(curr.value)
            curr = curr.prev

    def search_by_value(self,value):
        if self.head is None:
            return "Linked list does not exist"
        else:
            curr = self.head
            while curr:
                if curr.value == value:
                    print(curr.value)
                    return 
                curr = curr.next
            return "value does not exist in linked list"
    def search_by_index(self,position):
        if position < 0 or position > self.size - 1:
            raise IndexError("Index out of bound")
        if self.head is None:
            return "Linked list does not exist"
        else:
            index = 0
            curr = self.head
            while index < position:
                curr = curr.next
                index += 1
            return curr.value

    def deletion_of_node(self,position):
        if self.head is None:
            return "Linked list does not exist"
        if position < 0 or position >= self.size:
            raise IndexError("Index out of bound")
        if self.head == self.tail:
            self.head = None
            self.tail = None
            self.size = 0
            return
        elif position == 0:
            self.head = self.head.next
            self.head.prev = None
        elif position == self.size - 1:
            self.tail = self.tail.prev
            self.tail.next = None
        else:
            index = 0
            curr = self.head
            while index < position - 1:
                curr = curr.next
                index += 1
            curr.next = curr.next.next
            curr.next.prev = curr
        self.size -= 1
            
            
doubly_ll = DoublyLL()
doubly_ll.create_dll(3)
# print([(i.value,i.next,i.prev,i.head.next,i.tail.next) for i in doubly_ll])  # i is object of Node class so we can only print Node class attributes
print([(i.value,i.next,i.prev) for i in doubly_ll])  # i is object of Node class so we can only print Node class attributes
print("Nodes:")
for i in doubly_ll:
    print(i.value, i.prev, i.next)

print("List head next:", doubly_ll.head.next)
print("List tail next:", doubly_ll.tail.next)
print("**********************")
doubly_ll.traverse_dll()
print("***********************")
doubly_ll.insertion_at_beginning(5)
doubly_ll.insertion_at_end(23)
doubly_ll.insertion_at_middle(44,1)
doubly_ll.insertion_at_middle(99,4)
doubly_ll.insertion_at_middle(67,0)

doubly_ll.traverse_dll()
print("*****************")
doubly_ll.reverse_traverse_dll()
print("**************")
doubly_ll.search_by_value(67)
doubly_ll.search_by_index(5)
doubly_ll.deletion_of_node(0)
doubly_ll.deletion_of_node(5)
print("************after deletion")
doubly_ll.traverse_dll()
```
