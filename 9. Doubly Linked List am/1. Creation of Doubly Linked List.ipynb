```python
class Node:
    def __init__(self,value = None):
        self.value = value
        self.prev = None
        self.next = None

class DoublyLL:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
    def __iter__(self):
        curr = self.head
        while curr:
            yield curr
            curr = curr.next
    def create_dll(self,value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.size = 1
        return "DLL CREATED SUCCESSFULLY"

    def insertion_at_beginning(self,value):
        if self.head is None:
            return self.create_dll(value)
        else:
            new_node = Node(value)
            new_node.next = self.head
            new_node.prev = None
            self.head.prev = new_node
            self.head = new_node
            self.size += 1
    def insertion_at_end(self,value):
        if self.head is None:
            return self.create_dll(value)
        else:
            new_node = Node(value)
            new_node.next = None
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node
            self.size += 1
    def insertion_at_middle(self,value,position):
        if position < 0 or position > self.size:
            raise IndexError("Index out of bound")
        if self.head is None:
            return self.create_dll(value)
        if position == 0:
            return self.insertion_at_beginning(value)
        elif position == self.size:
            return self.insertion_at_end(value)
        else:
            new_node = Node(value)
            tempnode = self.head
            index = 0
            while index < position - 1:
                tempnode = tempnode.next
                index += 1
            nextnode = tempnode.next
            new_node.prev = tempnode
            tempnode.next = new_node
            new_node.next = nextnode
            nextnode.prev = new_node
            self.size += 1
doubly_ll = DoublyLL()
doubly_ll.create_dll(3)
# print([(i.value,i.next,i.prev,i.head.next,i.tail.next) for i in doubly_ll])  # i is object of Node class so we can only print Node class attributes
print([(i.value,i.next,i.prev) for i in doubly_ll])  # i is object of Node class so we can only print Node class attributes
print("Nodes:")
for i in doubly_ll:
    print(i.value, i.prev, i.next)

print("List head next:", doubly_ll.head.next)
print("List tail next:", doubly_ll.tail.next)
```
