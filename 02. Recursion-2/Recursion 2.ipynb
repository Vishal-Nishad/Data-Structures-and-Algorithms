```python
# 1. Replacing character a with b
def replace_char(st,a,b):
    if len(st)==0:
        return st
    small_output=replace_char(st[1:],a,b)
    if st[0]==a:
        return b+small_output
    else:
        return st[0]+small_output

replace_char("ccceeessccc","c","x")
```

```python
# 2. Remove x from char
def remove_x(st):
    if len(st)==0:
        return st
    small_output=remove_x(st[1:])
    if st[0]=="x":
        return small_output
    else:
        return st[0]+small_output

remove_x("axaaxx")
```

```python
# 3. replace pi with 3.14
def replace_pi(st):
    if len(st)==0 or len(st)==1:
        return st
    if st[0]=='p' and st[1]=='i':
        small_output=replace_pi(st[2:])
        return '3.14'+small_output
    else:
        small_output=replace_pi(st[1:])
        return st[0]+small_output
replace_pi("ddpid")
```

```python
# 4. remove duplicates recursively
def remove_cons_dup(st):
    if len(st)==0 or len(st)==1:
        return st
    if st[0]==st[1]:
        sm_output=remove_cons_dup(st[1:])
        return sm_output
    else:
        sm_output=remove_cons_dup(st[1:])
        return st[0]+sm_output
remove_cons_dup("aabccba")
```

```python
# 5. Binary Search using recursion
def bin_search(arr,x,si,ei):
    if si>ei:
        return -1
    mid=(si+ei)//2
    if arr[mid]==x:
        return mid
    elif arr[mid]>x:
        return bin_search(arr,x,si,mid-1)
    else:
        return bin_search(arr,x,mid+1,ei)

bin_search([1,2,3,4,5],4,0,4)
```

```python
# 6,7,8 Merge sort
def merge (a1,a2,a):
    i=0
    j=0
    k=0
    while i<len(a1) and j<len(a2):
        if (a1[i]<a2[j]):
            a[k]=a1[i]
            i+=1
            k+=1
        else:
            a[k]=a2[j]
            k+=1
            j+=1
    while i<len(a1):
        a[k]=a1[i]
        k+=1
        i+=1
    while j<len(a2):
        a[k]=a2[j]
        k+=1
        j+=1

def merge_sort(a):
    if len(a)==0 or len(a)==1:
        return
    mid=len(a)//2
    a1=a[0:mid]
    a2=a[mid:]

    merge_sort(a1)
    merge_sort(a2)
    merge(a1,a2,a)


a=[10,5,3,1,7,9,4]
merge_sort(a)
a
```

```python
# 9,10,11 Quick Sort
def partition(arr,si,ei):
    pivot=arr[si]
    c=0
    for i in range(si,ei+1):
        if arr[i]<pivot:
            c+=1
    arr[si+c],arr[si]=arr[si],arr[si+c]
    pivot_index=si+c
    i=si
    j=ei
    while i<j:
        if arr[i]<pivot:
            i+=1
        elif arr[j]>pivot:
            j-=1
        else:
            arr[i],arr[j]=arr[j],arr[i]
            i+=1
            j-=1
    return pivot_index



def quick_sort(arr,si,ei):
    if si>=ei:
        return
    pivot_index=partition(arr,si,ei)
    quick_sort(arr,si,pivot_index-1)
    quick_sort(arr,pivot_index+1,ei)


a=[6,7,4,3,2,1,9]
quick_sort(a,0,len(a)-1)
a
```

```python
# Tower of hanoi 
def tower_hanoi(n,a,b,c):
    if n==1:
        print("move 1st disk from ", a, " to ", c)
        return
    tower_hanoi(n-1,a,c,b)
    print("move ", n,"th disk from ", a, " to ", c)
    tower_hanoi(n-1,b,a,c)
```
