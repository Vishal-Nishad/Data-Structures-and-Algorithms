# Q. Balanced Paranthesis, return True or False

```python
# you have to check every paranthesis or curly braces should be balanced, if yes return True else False

def balanced_paranthesis(string):
    lt = []
    for i in range(len(string)):
        if string[i] == "(":
            lt.append(string[i])
        elif string[i] == ")":
            lt.pop()
    return len(lt) == 0

st = "(()()(()))"
balanced_paranthesis(st)
```
