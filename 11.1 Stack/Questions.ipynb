# Q. Balanced Paranthesis, return True or False

````python
# you have to check every paranthesis or curly braces should be balanced, if yes return True else False

def balanced_paranthesis(string):
    lt = []
    for i in range(len(string)):
        if (string[i] == "(" or string[i] == "{" or string[i] == "["):
            lt.append(string[i])
        elif string[i] == ")":
            if len(lt) == 0 or lt[-1] != "(":
                return False
            lt.pop()
        elif string[i] == "}":
            if len(lt) == 0 or lt[-1] != "{":
                return False
            lt.pop()
        elif string[i] == "]":
            if len(lt) == 0 or lt[-1] != "[":
                return False
            lt.pop()
    return len(lt) == 0

st = "(()()(()))"
balanced_paranthesis(st)
````

# Reverse Stack, only two stack is given one is with value, other is empty, we want reversed value in same s1 stack

````python
def reversestck(s1,s2):
    def reversestack(s1,s2):
        if len(s1) == 0:
            return
        lastelement = s1.pop()
        reversestack(s1,s2)

        s2.append(lastelement)
    reversestack(s1,s2)
    while len(s2) > 0:
        s1.append(s2.pop())

    return s1

s1 = [1,2,3,4]
s2 = []
reversestck(s1,s2)
````

# 🧠 HOW TO DRAW TWO COLUMNS

On paper, draw two vertical columns like this:

```
CALL PHASE (↓)              RETURN PHASE (↑)
--------------------        --------------------
```

Now we fill them.

---

# 🔽 LEFT COLUMN — CALL PHASE

You only write:

* Current stack
* What got popped
* Next recursive call

Do NOT write s2 yet.

---

```
CALL PHASE (↓)              RETURN PHASE (↑)
-------------------------------------------------

Depth 1:
s1 = [1,2,3,4]
pop → 4
call reversestack([1,2,3])

Depth 2:
s1 = [1,2,3]
pop → 3
call reversestack([1,2])

Depth 3:
s1 = [1,2]
pop → 2
call reversestack([1])

Depth 4:
s1 = [1]
pop → 1
call reversestack([])

Depth 5:
s1 = []
BASE CASE → return
```

STOP.

Do not think about pushing yet.

---

# 🔼 RIGHT COLUMN — RETURN PHASE

Now we go upward one by one.

We resume each paused frame.

---

```
CALL PHASE (↓)              RETURN PHASE (↑)
-------------------------------------------------

Depth 4 resumes:
lastelement = 1
s2 = [1]

Depth 3 resumes:
lastelement = 2
s2 = [1,2]

Depth 2 resumes:
lastelement = 3
s2 = [1,2,3]

Depth 1 resumes:
lastelement = 4
s2 = [1,2,3,4]
```

Notice carefully:

We removed:

```
4 → 3 → 2 → 1
```

We inserted:

```
1 → 2 → 3 → 4
```

Reversal happened automatically.

---

# 🔁 FINAL STEP (Outside Recursion)

Now move s2 back to s1:

```
pop 4 → s1 = [4]
pop 3 → s1 = [4,3]
pop 2 → s1 = [4,3,2]
pop 1 → s1 = [4,3,2,1]
```

Final:

```
s1 = [4,3,2,1]
```

---

# 🧠 THE KEY VISUAL MEMORY TRICK

Think of recursion like stacking boxes:

```
[4]
[3]
[2]
[1]
```

When base case hits, boxes start coming off:

```
1
2
3
4
```

That reversal is automatic because recursion stack is LIFO.

---

# 🏆 The Exact Way You Should Draw Every Recursion

Always draw like this:

```
CALL PHASE ↓
Depth 1:
Depth 2:
Depth 3:
Base Case
-----------------
RETURN PHASE ↑
Depth 3:
Depth 2:
Depth 1:
```

Never mix phases.

---

# 🔥 Why This Method Never Confuses You

Because:

* Left column = removing
* Right column = rebuilding

You separate them mentally.

# check if any paranthesis is redundant, if redundant return True else return False

````python
# space complexity = O(n)
# time complexity = O(n)
# Even though there is a nested while,
# this is not O(n²) because:
# Elements are not repeatedly processed —
# each character is pushed and popped only once.
def checkvalidparanthesis(exp):
    stack = []
    for i in range(len(exp)):
        if exp[i] != ")":
            stack.append(exp[i])
        else:
            count = 0
            while len(stack) != 0 and stack[len(stack) - 1] != "(":
                stack.pop()
                count += 1
            stack.pop()
            if count <= 1:
                return True
    return False

a = "(a+b)"
b = "((a+b))"
c = "(a)"
print(checkvalidparanthesis(a))
print(checkvalidparanthesis(b))
print(checkvalidparanthesis(c))
````

# Stock span

**Problem Description: The span si of a stock’s price on a certain day i is the
minimum number of consecutive days (up to the current day) the price of the
stock has been less than its price on that ith day. If for a particular day, if no
stock price is greater then just count the number of days till 0th day including
current day. Given an input array with all stock prices, you need to return the
array with corresponding spans of each day.**


**Sample Input:
8
60 70 80 100 90 75 80 120
Sample Output:
1 2 3 4 1 1 2 8**

````python
# brute force
# space complexity = O(n)
# time complexity = O(n2)
def stock_span(ar):
    res = []
    for i in range(len(ar)):
        count = 1
        for j in range(i-1,-1,-1):
            if ar[j]>ar[i]:
                break
            else:
                count += 1
        res.append(count)
    return res

a = [60,70,80,100,90,75,80,120]
print(stock_span(a))
````

````python
# optimal using stack
# time complexity = O(n), we are pushing only once and poping only once, hence n + n = 2n
# space complexity = O(n), if the price in decreasing monotonic order, then in stack all the indices get push not pop
def span_stock_optimal(prices):
    stack = []
    span = []
    for i in range(len(prices)):
        while len(stack) != 0 and prices[stack[-1]] <= prices[i]:
            stack.pop()
        if len(stack) == 0:
            span.append(i+1)
        else:
            span.append(i - stack[-1])
        stack.append(i)
    return span

a = [60,70,80,100,90,75,80,120]
print(span_stock_optimal(a))
````

# minimum bracket reversal

````python
# time complexity = O(n)
# space complexity = O(n), in worst case we are pushing all the elements in stack
def minimum_braces_reversal(st):
    stack = []
    if len(st) % 2 != 0: # early return if len is odd, because it is impossible to balance odd len braces, there is no pair available to balance
        return -1
    for i in range(len(st)):
        if st[i] == "{":
            stack.append(st[i])
        else:
            if len(stack) == 0:
                stack.append(st[i])
            elif len(stack) != 0 and stack[-1] != "{":
                stack.append(st[i])
            else:
                stack.pop()

    count = 0
    while len(stack) >1:
        c1 = stack.pop()
        c2 = stack.pop()
        if c1 == c2:
            count += 1
        else:
            count += 2
    return count

s = "}}}{"
minimum_braces_reversal(s)
````
