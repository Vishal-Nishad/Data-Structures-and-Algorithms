## Two Pointers:
**https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/submissions/1876732526/**

```python
# 167. Two Sum II - Input Array Is Sorted
# brute force approach, time complexity O(n2)
def two_sum(arr,target):
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] + arr[j] == target:
                return [i+1,j+1]

ar = [2,7,11,15]
two_sum(ar,9)
```

```python
# Hashmap approach, time complexity = O(n), space complexity = O(n)
def two_sum_hash(ar, target):
    seen = {}
    n = len(ar)
    for i in range(n):
        needed = target - ar[i]
        if needed in dict:
            return [seen[needed]+1, i+1]
        else:
            dict[ar[i]] = i

ar = [2,7,11,15]
two_sum_hash(ar, 26)
```

```python
# Optimal two pointer approach, time complexity = O(n) if array is already sorted, otherwise if sorting needed then O(nlogn)
# space complexity O(1) constant we are not using any etc space 
def two_sum_optimal(ar, target):
    i,j = 0, len(ar) - 1
    while i < j:
        s = ar[i] + ar[j]
        if s == target:
            return [i+1, j+1]
        # if sum is greater than target then we will move j to the left to decrease the sum
        elif s > target:
            j -= 1
        # else sum is less than target then we will move i to the right to increase the sum
        else:
            i += 1
    return -1
ar = [2,7,11,15]
two_sum_optimal(ar, 26)
```

2. **https://leetcode.com/problems/remove-duplicates-from-sorted-list/description/**

```python
# Remove duplicates from sorted array
# Brute force, space complexity = O(n), time complexity = O(n)
def rem_duplicated(arr):
    res = [arr[0]]
    for j in range(1,len(arr)):
        if res[-1] != arr[j]:
            res.append(arr[j])
    return len(res)
ar = [2,3,3,4,4,5,6]
rem_duplicated(ar)
```

```python
# Optimal two pointer
def rem_duplicate_optimal(arr):
    i,j,result= 0,1,1
    while j<len(arr):
        if arr[j] != arr[i]:  # we can compare j with j-1 means with previos element as well,
            arr[i+1] = arr[j]
            i += 1
            j += 1
            result += 1
        else:
            j +=1
    return i+1, result
ar = [2,3,3,4,4,5,6]
print(rem_duplicate_optimal(ar))

def rem_duplicate_optimal_easier(arr):
    i = 0
    for j in range(1,len(arr)):
        if arr[i] != arr[j]: 
            arr[i+1] = arr[j]
            i +=1
    return i+1

ar = [2,3,3,4,4,5,6]
rem_duplicate_optimal_easier(ar)
```

3. **https://leetcode.com/problems/squares-of-a-sorted-array/description/**

```python
# return squares of sorted array, array can contain negative positive elements
# Brute force, just do the square and sort and return, time complexity = O(nlogn), because of doing sorting
def sq_sorted_brute(arr):
    sq_sorted = [i**2 for i in arr]
    arr = sq_sorted
    copy = sq_sorted
    # doing manually, O(n2) time complexity
    for i in range(len(sq_sorted)):
        min_value = copy[i]
        min_ind = i     
        for j in range(i+1,len(sq_sorted)):
            if min_value > copy[j]:
                min_value = copy[j]
                min_ind = j
        copy[min_ind], copy[i] = copy[i], copy[min_ind]
    return sq_sorted, arr,copy
ar = [-4, -1, 0, 4,3,5]
sq_sorted_brute(ar)
```

```python
# optimal approach using two pointer
def sq_sorted_optimal(arr):
    neg = []
    pos = []
    for i in range(len(arr)):
        if arr[i]<0:
            neg.append(arr[i]**2)
        else:
            pos.append(arr[i]**2)
    if len(neg) == 0:
        return pos
    neg = neg[::-1]
    
    res_sorted = []
    i,j = 0, 0
    while i<len(neg) and j < len(pos):
        if neg[i] <= pos[j]:
            res_sorted.append(neg[i])
            i +=1
        else:
            res_sorted.append(pos[j])
            j +=1
    # if neg element still remainin
    while i < len(neg):
        res_sorted.append(neg[i])
        i += 1
    while j < len(pos):
        res_sorted.append(pos[j])
        j +=1
    return res_sorted
ar = [-4, -1, 0, 3,4,5]
sq_sorted_optimal(ar)
```

```python
# Triplet sum to zero
# brute force, time complexity = O(n3), space complexity = O(k) i.e total number of triplets
def sum_to_zero(arr):
    n = len(arr)
    result = set()
    for i in range(n):
        for j in range(i+1, n):
            for k in range(j+1,n):
                if arr[i] + arr[j] + arr[k] == 0:
                    element = tuple(sorted([arr[i],arr[j],arr[k]]))
                    result.add(element)
    return list(result)

ar = [-1,0,1,2,-1,-4]
sum_to_zero(ar)
```

```python
# Optimal using tow pointers
def sum_to_zero_optimal(arr):
    arr = sorted(arr)
    result = []
    for i in range(len(arr) - 2):
        if (i>0 and arr[i] == arr[i-1]):
            continue
        target = -arr[i]
        j,k = i+1, len(arr)-1
        while (j<k):
            if arr[j] + arr[k] == target:
                result.append([arr[i],arr[j],arr[k]])
                # print(result)
                j += 1
                k -= 1
                while (j<k and arr[j] == arr[j-1]):
                    j += 1
                while (j<k and arr[k] == arr[k+1]):
                    k -=1
            elif arr[j] + arr[k] < target:
                j +=1
            else:
                k -= 1
    return result
ar = [-1,0,1,2,-1,-4]
sum_to_zero_optimal(ar)
```
