1.**https://leetcode.com/problems/linked-list-cycle/description/**

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        s = f = head
        while f is not None and f.next is not None:
            s = s.next
            f = f.next.next
            if s == f:
                return True
        return False
````

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                break

        if fast is None or fast.next is None:
            return None
        slow = head
        while slow != fast:
            slow = slow.next
            fast = fast.next
        return slow
        
````

3. **https://leetcode.com/problems/find-the-duplicate-number/submissions/1902925837/j**

````python
# brute force
# space complexity = O(n)
# time complexity = O(n)
def findDuplicate(nums):
        freq = {}
        for i in range(len(nums)):
            if nums[i] not in freq:
                freq[nums[i]] = 1
            else:
                freq[nums[i]] += 1

        for key in freq.keys():
            if freq[key] >1:
                return key

nums = [3,1,3,4,2]
findDuplicate(nums)
````

````python
# brute force but cleaner version
def finddupcleaner(nums):
    freq = {}
    for i in range(len(nums)):
        if nums[i] in freq:
            return nums[i]
        freq[nums[i]] = 1
nums = [3,1,3,4,2]
finddupcleaner(nums)
````
