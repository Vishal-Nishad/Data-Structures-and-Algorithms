````python
# print sum of n natural numbers by recursion
def rec_sum(n):
    if n == 1:
        return 1
    ans = rec_sum(n-1)
    return ans + n
rec_sum(5)
````

# 1. First understand what recursion *really is*

Think of recursion as:

> “I don’t solve everything. I solve **one step**, and delegate the rest.”

For your function:

```python
def rec_sum(n):
    if n == 1:
        return 1
    ans = rec_sum(n-1)
    return ans + n
```

Each call says:

> “Give me sum till (n-1), I’ll just add `n`.”

---

# 2. The SMART way to draw recursion (this is the key technique)

Instead of messy trees, always use this **2-column method**:

| Call | Returns |
| ---- | ------- |

---

## Step-by-step for `rec_sum(5)`

### Step 1: Write all calls first (top → bottom)

```
rec_sum(5)
rec_sum(4)
rec_sum(3)
rec_sum(2)
rec_sum(1)
```

👉 Just keep expanding until base case.

---

### Step 2: Now resolve from bottom (VERY IMPORTANT)

Start filling returns **backwards**:

| Call       | Returns     |
| ---------- | ----------- |
| rec_sum(1) | 1           |
| rec_sum(2) | 1 + 2 = 3   |
| rec_sum(3) | 3 + 3 = 6   |
| rec_sum(4) | 6 + 4 = 10  |
| rec_sum(5) | 10 + 5 = 15 |

# 3. Visualizing call stack (how interviewer expects you to think)

Think in **stack behavior (LIFO)**:

```

  → rec_sum(4)
      → rec_sum(3)
          → rec_sum(2)
              → rec_sum(1)  ✅ base case

Now returning:

```

# 4. The REAL trick for complex problems (this is what you're asking)

For hard recursion problems, follow this **4-step system**:

---

## ✅ Step 1: Define “what does function return”

Ask:

> “What exactly does `rec_sum(n)` return?”

👉 Answer:

> Sum of numbers from `1 → n`

This avoids confusion later.

---

## ✅ Step 2: Assume smaller problem works

Never expand everything.

Instead say:

```
Assume rec_sum(4) = 10
```

Then:

```

```

👉 This is called **faith in recursion**

---

## ✅ Step 3: Only expand when stuck

If confused:

* Expand only 2–3 levels
* Not the entire tree

---

## ✅ Step 4: Track variables per call

For complex problems (like trees, DP):

Write like this:

```

# find maximum number in array recursively

````python
def max_rec(arr,n):
    if n == 1:
        return arr[0]
    last = arr[n-1]
    max_no = max_rec(arr,n-1)
    return max(last,max_no)

a = [3,4,1]
max_rec(a,3)
````

# Check Palindrome recursively

````python
# time complexity = O(n)
# space = O(n) i.e because of recursive call stack,
def check_palindrome(str,st,end):
    if(st>=end):
        return True
    if(str[st] != str[end]):
        return False
    return check_palindrome(str,st+1,end -1)

st = "namal"
check_palindrome(st,0,len(st)-1)
````

# Q. recursively print sum of digits


````python
# iterative
def iter_digit_sum(n):
    ans = 0
    while n>0:
        rem = n%10
        ans += rem
        n = n//10
    return ans

iter_digit_sum(123)
````

````python
# recursive
def rec_digit_sum(n,ans):
    if(n==0):
        return ans
    ans += n%10
    return rec_digit_sum(n//10,ans)

rec_digit_sum(123,0)
````

# Q. check if array is sorted or not, recursively

````python
def rec_is_sorted(ar,i):
    if(i==len(ar)-1):
        return True
    if(ar[i]>ar[i+1]):
        return False
    return rec_is_sorted(ar,i+1)

a = [1,4,2,3]
rec_is_sorted(a,0)
````

# Q. remove char from given string

````python
def rec_remove_ch(str,ans_str,ch,ind):
    if(ind == len(str)):
        return ans_str
    if(str[ind] != ch):
        ans_str += str[ind]
    return rec_remove_ch(str,ans_str,ch,ind+1)

st = "apple"
rec_remove_ch(st,"","e",0)
````

🔴 1. **https://leetcode.com/problems/combination-sum/description/**

````python
# time complexity = O(N^target)
'''Branching Factor -> Maximum recursive calls created by ONE call
Depth -> Maximum recursion chain length
Total Work -> branching_factor^depth

because every level multiplies possibilities.'''

# Space Complexity:
'''O(target) auxiliary space due to recursion stack
and current path storage.

If output storage is included,
additional space is required for storing combinations.'''

def combination_sum(candidates,target):
    result = []
    def backtrack(start,res_sum,temp_arr):
        # base case
        if(res_sum == target):
            result.append(temp_arr[:])
            return

        # invalid case
        if(res_sum > target):
            return

        # recurse, all choices
        for i in range(start,len(candidates)):
            num = candidates[i]
            # work, do
            temp_arr.append(num)
            backtrack(i,res_sum+num,temp_arr)

            # undo
            temp_arr.pop()
    backtrack(0,0,[])
    return result

candidates = [2,3,6,7]
target = 7
combination_sum(candidates,target)
````

2. **https://leetcode.com/problems/combination-sum-ii/description/**

````python
def combination_sum(candidates,target):
    # sorting
    candidates.sort()
    result = []
    def backtrack(start,res_sum,temp_arr):
        # base case
        if(res_sum == target):
            result.append(temp_arr[:])
            return

        # invalid case
        if(res_sum > target):
            return

        # recurse, all choices
        for i in range(start,len(candidates)):
            if(i>start and candidates[i]==candidates[i-1]):
                continue
            num = candidates[i]
            # work, do
            temp_arr.append(num)
            backtrack(i+1,res_sum+num,temp_arr)

            # undo
            temp_arr.pop()
    backtrack(0,0,[])
    return result

candidates = [10,1,2,7,6,1,5]
target = 8
combination_sum(candidates,target)
````

# 1. `candidates.sort()`

```python
candidates.sort()
```

Purpose:

```text id="4ylmzz"
Bring duplicate numbers together.
```

Why needed?

Because duplicate skipping condition works only when equal elements are adjacent.

Example:

Without sorting:

```python id="gzaybq"
[1,2,1]
```

duplicates are separated.

After sorting:

```python id="0b7k4f"
[1,1,2]
```

now duplicates are easy to detect.

---

# 2. `backtrack(i+1, ...)`

```python
backtrack(i+1, ...)
```

Purpose:

```text id="9qj0jg"
Move forward to next index.
```

Why needed?

Because in Combination Sum II:

```text id="hkwy1u"
Each element can be used only ONCE.
```

So after choosing current element at index `i`,
next recursion starts from:

```python id="dwng1d"
i+1
```

which prevents reusing same index again.

---

# 3. Duplicate Skip Condition

```python
if i > start and candidates[i] == candidates[i-1]:
    continue
```

Purpose:

```text id="l5v1v8"
Skip duplicate choices at SAME recursion level.
```

Why needed?

To avoid duplicate combinations like:

```python id="0dgz0s"
[1,7]
[1,7]
```

coming from different duplicate `1`s.

---

# Meaning of `i > start`

```text id="1u85m4"
Skip duplicates only among siblings,
NOT parent-child recursion.
```

This allows:

```python id="bh6ik2"
[1,1,6]
```

but prevents duplicate branches starting from same level.

---

4. **https://leetcode.com/problems/word-search/description/**

````python
def exist(board, word):
        def backtrack(col,row,index):
            # base case
            if(index == len(word)):
                return True
                
            #invalid case
            if(row<0 or col < 0 or row>=len(board) or col>=len(board[0]) or
            board[row][col] != word[index]):
                return False

            # mark visited/ do
            temp = board[row][col]
            board[row][col] = "#"

            # down
            if(backtrack(col,row+1,index+1)):
                return True

            # up
            if(backtrack(col,row-1,index+1)):
                return True
            
            # right
            if(backtrack(col+1,row,index+1)):
                return True

            # left
            if(backtrack(col-1,row,index+1)):
                return True

            # undo
            board[row][col] = temp
            return False
        
        for r in range(len(board)):
            for c in range(len(board[0])):
                if(backtrack(c,r,0)):
                    return True
        return False

board = [["A","B","C","E"],["S","F","C","S"],["A","D","E","E"]]
word = "ABCCED"
exist(board,word)
````
