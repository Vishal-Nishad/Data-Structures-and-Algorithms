1. **https://leetcode.com/problems/first-unique-character-in-a-string/description/**

```python
# brute force
# time complexity = O(n2)
# space complexity = O(1)
def firstUniqChar(s: str) -> int:
        is_dup = False
        for i in range(len(s)):
            is_dup = False
            for j in range(len(s)):
                if i == j:
                    continue
                if s[i] == s[j]:
                    is_dup = True
                    break
            if not is_dup:
                return i
        return -1

s = "loveleetcode"
firstUniqChar(s)
```

```python
# optimal using hashmap
# time complexity = O(n)
# space complexity = O(n)
def first_unique_char(st):
    freq = {}
    for i in range(len(st)):
        freq[st[i]] = freq.get(st[i], 0) + 1
    for i in range(len(st)):
        if freq[st[i]] == 1:
            return i
    return -1

s = "loveleetcode"
first_unique_char(s)
```

2. **https://leetcode.com/problems/maximum-number-of-balloons/description/**

```python
# hashmap
# time complexity = O(n)
# space = O(1), as only for char, max it can store upto 26
def maxNumberOfBalloons(text: str) -> int:
        freq = {}
        for i in range(len(text)): # O(n)
            freq[text[i]] = freq.get(text[i],0) + 1
        
        count = 0
        while True: # in worst case O(n)
            st = "balloon"
            for i in range(len(st)): # constant as always run for only 7 iteration
                if st[i] in freq and freq[st[i]] >=1:
                    freq[st[i]] -= 1
                else:
                    return count
            count += 1

text = "loonbalxballpoon"
maxNumberOfBalloons(text)
```

```python
# without using hashmap
# time = O(n)
def max_number_of_ballons(text):
    b = a = l = o = n = 0
    for i in range(len(text)):
        if text[i] == "b":
            b += 1
        elif text[i] == "a":
            a += 1
        elif text[i] == "l":
            l += 1
        elif text[i] == "o":
            o += 1
        elif text[i] == "n":
            n += 1
    return min(b,a,l//2, o//2,n)

text = "loonbalxballpoon"
max_number_of_ballons(text)
```

3. **https://leetcode.com/problems/longest-palindrome/description/**

```python
# time complexity = O(n)
# space complexity = O(1)
def longestPalindrome(s: str) -> int:
        freq = {}
        for i in range(len(s)):
            freq[s[i]] = freq.get(s[i],0) + 1
        
        count = 0
        for key, value in freq.items():
            if value >= 2:
                if value %2 ==0:
                    count += value
                else:
                    count += value - 1
        if count < len(s):
            count += 1
        return count

s = "abccccdd"
longestPalindrome(s)
```
