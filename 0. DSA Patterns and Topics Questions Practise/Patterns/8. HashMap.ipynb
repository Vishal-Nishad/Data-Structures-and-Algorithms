```python
# brute force
# time complexity = O(n2)
# space complexity = O(1)
def firstUniqChar(s: str) -> int:
        is_dup = False
        for i in range(len(s)):
            is_dup = False
            for j in range(len(s)):
                if i == j:
                    continue
                if s[i] == s[j]:
                    is_dup = True
                    break
            if not is_dup:
                return i
        return -1

s = "loveleetcode"
firstUniqChar(s)
```

```python
# optimal using hashmap

def first_unique_char(st):
    freq = {}
    for i in range(len(st)):
        freq[st[i]] = freq.get(st[i], 0) + 1
    for i in range(len(st)):
        if freq[st[i]] == 1:
            return i
    return -1

s = "loveleetcode"
first_unique_char(s)
```
