```python
def sorting_2d_arr(ar):
    for i in range(len(ar)):
        min_ind = i
        # min_value = ar[i][0]
        for j in range(i+1, len(ar)):
            if ar[j][0] < ar[min_ind][0]:
                min_ind = j
                # min_value = ar[j][0]
        ar[i],ar[min_ind] = ar[min_ind], ar[i]
    return ar

ar = [[1,4],[3,2],[2,5],[8,1],[4,2]]

sorting_2d_arr(ar)
```

```python
def merge_interval_brute(ar):
    i = 0
    while i < len(ar):
        j = i+1
        while j < len(ar):
            if ar[i][1] >= ar[j][0] and ar[j][1] >= ar[i][0]:
                ar[i][0] = min(ar[i][0],ar[j][0])
                ar[i][1] = max(ar[i][1],ar[j][1])

                ar.pop(j)
                j = i+1
            else:
                j += 1
        i += 1
    return ar

ar = [[1,3],[2,6],[8,10],[15,18]]
merge_interval_brute(ar)
```

```python
# optimal but taking space complexity etc
# time complexity = O(nlogn)  because of sorting
# space complexity = O(n)
def merge_interval(nums):
    res = []
    nums.sort(key = lambda x:x[0])
    # for i in range(len(nums)):
    #     min_ind = i
    #     for j in range(i+1,len(nums)):
    #         if nums[j][0]<nums[min_ind][0]:
    #             min_ind = j
    #     nums[i],nums[min_ind] = nums[min_ind],nums[i]
    start_1 = nums[0][0]
    end_1 = nums[0][1]
    for i in range(1,len(nums)):
        start_2 = nums[i][0]
        end_2 = nums[i][1]
        if (end_1 >=start_2):
            end_1 = max(end_1,end_2)
            continue
        res.append([start_1,end_1])
        start_1 = start_2
        end_1 = end_2
    res.append([start_1,end_1])
    return res

ar = [[1,3],[2,6],[8,10],[15,18]]
merge_interval(ar)
```

2. **https://leetcode.com/problems/insert-interval/description/**

```python
# one traversal clear explainable logic
# space complexity O(n)
# time complexity O(n)
def onepass_insert_interval(intervals,new_interval):
    result = []
    i = 0
    n = len(intervals)

    # adding non overlapping intervals into result, end_1 < start_2(new_interval), because intervals is already sorted and non-overlapping so we only
    # check new_interval
    while i<n and intervals[i][1] < new_interval[0]:
        result.append(intervals[i])
        i += 1

    # as now if this condition - end_1 < start_2(new_interval),, becomes false means , now end_1 >= start_2(new_interval) now we have to handle overlapping
    # scenarios, so we will convert all overlapping in single new_interval then we will add it to result
    while i<n and intervals[i][0] <= new_interval[1]:
        new_interval[0] = min(intervals[i][0], new_interval[0])
        new_interval[1] = max(intervals[i][1], new_interval[1])
        i += 1
    result.append(new_interval)

    while i<n:
        result.append(intervals[i])
        i += 1
    return result

intervals = [[1,2],[3,5],[6,7],[8,10],[12,16]]
newInterval = [4,8]
onepass_insert_interval(intervals, newInterval)
```
