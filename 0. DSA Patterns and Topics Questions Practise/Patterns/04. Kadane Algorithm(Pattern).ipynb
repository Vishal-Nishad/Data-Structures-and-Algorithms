1. **https://leetcode.com/problems/maximum-subarray/description/?utm_source=chatgpt.com**

````python
# brute force
# space complexity = O(1)
# time complexity = O(n2)
def maxsubarray_brute(nums):
    result = float("-inf")
    n = len(nums)
    for i in range(n):
        max_sum = 0
        for j in range(i,n):
            max_sum += nums[j]
            if max_sum > result :
                result = max_sum
    return result

nums = [5,4,-1,7,8]
maxsubarray_brute(nums)
````

````python
# optimal kadane algo
# time complexity = O(n)
# space complexity = O(1)
def maxsubarray_optimal(nums):
    result = nums[0]
    best = nums[0]
    for i in range(1,len(nums)):
        if best + nums[i] > nums[i]:
            best += nums[i]
        else:
            best = nums[i]
        if best > result:
            result = best
    return result

nums = [5,4,-1,7,8]
maxsubarray_optimal(nums)
````

3. **https://leetcode.com/problems/maximum-product-subarray/?utm_source=chatgpt.com**

````python
# space complexity = O(1)
# time complexity = O(n2)
def maxproduct_brute(nums):
    result = float("-inf")
    n = len(nums)
    for i in range(n):
        max_sum = 1
        for j in range(i,n):
            max_sum *= nums[j]
            if max_sum > result :
                result = max_sum
    return result

nums = [2,3,-2,4]
maxproduct_brute(nums)
````

````python
# optimal
# space complexity = O(1)
# time complexity = O(n)

def maxprod_optimal(nums):
    max_prod = nums[0]
    min_prod = nums[0]
    result = nums[0]
    for i in range(1,len(nums)):
        temp_max = max(nums[i],nums[i]*max_prod,nums[i]*min_prod)
        min_prod = min(nums[i],nums[i]*max_prod,nums[i]*min_prod)
        max_prod = temp_max
        if max_prod > result:
            result = max_prod
    return result

nums = [2,3,-2,4]
maxprod_optimal(nums)
````

4. **https://leetcode.com/problems/maximum-subarray-sum-with-one-deletion/description/**

````python
# brute force
# space complexity = 
# time complexity = 
def mx_sum_one_del(arr):
    if len(arr) == 1:
        return arr[0]
    res = float("-inf")
    for i in range(len(arr)):
        subarr_sum = 0
        onedeletion_sum = 0
        count = 0
        min_value = float("inf")
        for j in range(i,len(arr)):
            subarr_sum += arr[j]
            count += 1
            min_value = min(min_value,arr[j])
            if count > 1:
                onedeletion_sum = subarr_sum - min_value
                res = max(res,onedeletion_sum,subarr_sum)
            else:
                res = max(res,subarr_sum)
    return res

arr = [1,-2,0,3]
mx_sum_one_del(arr)
````

````python
# one deletion + Kadane = Kadane with TWO lanes 

# two states in this question
# One delete → either extend or delete current.
# No delete → normal Kadane.

# time complexity O(n)
# space complexity O(1)
def mx_sum_one_del_optimal(arr):
    no_del = arr[0]
    one_del = float("-inf")
    result = arr[0]  # suppose if there is a single element, then just return that,bcz loop will not run for single element

    for i in range(1,len(arr)):
        one_del = max(one_del + arr[i], no_del)
        no_del = max(no_del + arr[i],arr[i])
        result = max(result,one_del,no_del)
    return result
arr = [1,-2,0,3]
mx_sum_one_del_optimal(arr)
````

5 **https://leetcode.com/problems/maximum-absolute-sum-of-any-subarray/submissions/1908582039/**

````python
# brute force
# time complexity = O(n2)
# space complexity = O(1)
def mx_abs_sum_brute(nums):
    result = nums[0]
    for i in range(len(nums)):
        curr_mx = 0
        for j in range(i,len(nums)):
            curr_mx += nums[j]
            result  = max(result, abs(curr_mx))
    return result
nums = [1,-3,2,3,-4]
mx_abs_sum_brute(nums)
````

````python
# optima
# time complexity = O(n)
# space complexity = O(1)
def mx_abs_sum_optimal(nums):
    result = nums[0]
    min_best = nums[0]
    max_best = nums[0]
    for i in range(1,len(nums)):
        min_best = min(min_best + nums[i], nums[i])
        max_best = max(max_best + nums[i], nums[i])
        result = max(abs(result), abs(min_best), abs(max_best))
    return result

nums = [1,-3,2,3,-4]
mx_abs_sum_optimal(nums)
````

6.**https://leetcode.com/problems/maximum-sum-circular-subarray/description/?utm_source=chatgpt.com**

````python
# brute force 
# space complexity = O(1)
# time complexity = O(n2)
def circular_max_sum_sub(nums):
    res = nums[0]
    n = len(nums)
    for i in range(n): # O(n)
        curr_sum = 0
        length = 0
        ind = i
        while length < n: #O(n)  , n*n = n2
            curr_sum += nums[ind]
            res = max(res,curr_sum)
            ind = (ind + 1)%n
            length += 1
    return res
nums = [5,-3,5]
circular_max_sum_sub(nums)
````

````python
# optimal 
# space complexity = O(1)
# time complexity = O(n)
def circular_max_sum_sub_optimal(nums):
    total = nums[0]
    res_max = max_end = nums[0]
    res_min = min_end = nums[0]
    for i in range(1,len(nums)):
        max_end = max(nums[i], max_end + nums[i])
        res_max = max(res_max , max_end)

        min_end = min(nums[i], min_end + nums[i])
        res_min = min(min_end, res_min)
        total += nums[i]
    if res_max < 0: # if all numbers are negative return early, because, total-res_min always give 0 in this case, as total and res_min will always same
        return res_max
    return max(res_max, total - res_min)

nums = [5,-3,5]
circular_max_sum_sub_optimal(nums)
````

Great question — this is the **exact line** where most people feel “okay I see the formula, but WHY?”

Let’s remove the formula completely and **build the idea from scratch with examples**.
No Kadane. No shortcuts. Just logic.

---

## 1️⃣ What does a *circular* subarray really mean?

In a **circular array**, a subarray can:

* Start near the end
* Wrap around
* Continue at the beginning

Example:

```
nums = [5, -3, 5]
```

A wrapping subarray looks like:

```
[5] (end) + [5] (start) = 10
```

So wrapping = **suffix + prefix**

---

## 2️⃣ Key realization (THIS is the turning point)

If a subarray wraps around, then:

> It is equivalent to **taking the whole array**
> and **removing one contiguous middle part**

That’s it. That’s the whole idea.

Let’s visualize.

---

## 3️⃣ Visual example (very important)

### nums = [5, -3, 5]

Think of the array as a circle:

```
5 ---- -3 ---- 5
^                 |
|-----------------|
```

If you want the wrapping subarray `[5, 5]`, what did you do?

You **excluded** `[-3]`.

So:

```
wrap_sum = total_sum - excluded_part
```

Here:

```
total_sum = 5 + (-3) + 5 = 7
excluded_part = -3
wrap_sum = 7 - (-3) = 10
```

---

## 4️⃣ Which part should we exclude?

Ask this question carefully:

> If I must remove **one contiguous subarray**,
> which one should I remove to make the remaining sum as large as possible?

Answer:
👉 **The subarray with the minimum sum**

Because:

* Removing a very negative chunk **increases** the remaining sum the most

---

## 5️⃣ Another example (more complex)

### nums = [8, -1, -3, 8]

Let’s compute:

```
total_sum = 8 - 1 - 3 + 8 = 12
```

Now consider wrapping subarray:

Best wrap is:

```
[8] (end) + [8] (start) = 16
```

What did we exclude?

```
[-1, -3] = -4
```

So:

```
wrap_sum = 12 - (-4) = 16
```

And `[-1, -3]` is exactly the **minimum subarray**.

---

## 6️⃣ One more example (to really lock it in)

### nums = [4, -6, 2, 3, -2, 3]

```
total_sum = 4 - 6 + 2 + 3 - 2 + 3 = 4
```

Minimum subarray?

```
[-6] → sum = -6
```

Exclude it:

```
wrap_sum = 4 - (-6) = 10
```

Remaining wrap subarray:

```
[2, 3, -2, 3, 4] → 10
```

---

## 7️⃣ Why this logic is ALWAYS valid

For any wrapping subarray:

* You take some suffix
* You take some prefix
* The only part you don’t take is **one contiguous middle section**

That middle section:

* Must be contiguous
* Must be excluded exactly once

Therefore:

```
wrap_sum = total_sum - (sum of excluded middle part)
```

To maximize wrap_sum:

* Minimize excluded sum

Hence:

```
wrap_sum = total_sum - min_subarray_sum
```

---

## 8️⃣ Why we still compare with normal Kadane

Because sometimes **not wrapping is better**.

Example:

```
nums = [1, -2, 3, -2]
```

```
total = 0
min_subarray = -2
wrap_sum = 0 - (-2) = 2
```

But normal Kadane gives:

```
[3] → 3
```

So final answer is:

```
max(normal, wrap)
```

---

## 9️⃣ The all-negative edge case (IMPORTANT)

Example:

```
nums = [-3, -2, -3]
```

```
total = -8
min_subarray = -8
wrap_sum = -8 - (-8) = 0 ❌
```

But `0` means **empty subarray**, which is not allowed.

So rule:

> If all numbers are negative → return normal Kadane

---

## 🔑 Final cheat-code (remember this forever)

> **Wrapping subarray = everything except the worst contiguous part**

Or even shorter:

> **total − min_subarray = best wrap**

If this sentence makes sense, you truly understand the logic.
