1. **https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/description/**

````python
def remove_all_adjacent_brute(st):
    run_loop_again = True
    while run_loop_again:# in worst case it will cost O(n)
        run_loop_again = False
        new_st = "" # space complexity = O(n)
        i = 0
        while i < len(st):  # everytime it will run O(n)  , so worst case time complexity = O(n2)
            if i < len(st) - 1 and st[i] == st[i+1]:
                run_loop_again = True
                i += 2
            else:
                new_st += st[i]
                i += 1
        st = new_st
    return st
s = "abbaca"
remove_all_adjacent_brute(s)
````

````python
# time complexity = O(n2)
# space complexity = O(n)
def remove_all_adjacent_duplicates(st):
    stack = []  # O(n) space
    for i in range(len(st)): # O(n)
        if len(stack) != 0 and stack[-1] == st[i]:
            stack.pop()
        else:
            stack.append(st[i])
    res = "" # O(n) space
    while len(stack) != 0:    # in worst case it can be O(n),,,, so this while loop in string concatenation will cose O(n2) in worst case
        res += str(stack.pop())  # String concatenation in Python is NOT O(1).
# Strings are immutable. Each concatenation copies the whole string. So this part costs: 1 + 2 + 3 + ... + n
    return res[::-1]   # this reversal also cost O(n)

s = "abbaca"
remove_all_adjacent_duplicates(s)
````

````python
# time complexity = O(n)
def remove_all_adjacent(st):
    stack = []
    for i in range(len(st)):
        if len(stack) != 0 and stack[-1] == st[i]:
            stack.pop()
        else:
            stack.append(st[i])
    return "".join(stack)

# How join Is Faster
# Calculates total required size
# Allocates memory once
# Copies all elements in one pass

# So:
# O(n)
# Much faster.

s = "abbaca"
remove_all_adjacent(s)
````

2. **https://leetcode.com/problems/valid-parentheses/description/**

````python
def isValid(s):
        st = []
        for i in range(len(s)):
            if s[i] == "(" or s[i] == "[" or s[i] == "{":
                st.append(s[i])
            elif s[i] == ")":
                if len(st) == 0 or st[-1] != "(":
                    return False
                st.pop()

            elif s[i] == "}":
                if len(st) == 0 or st[-1] != "{":
                    return False
                st.pop()
            elif s[i] == "]":
                if len(st) == 0 or st[-1] != "[":
                    return False
                st.pop()
        return len(st) == 0

s = "()()()"
isValid(s)
````

4. **https://leetcode.com/problems/next-greater-element-i/description/**

````python
# brute force
def nextGreaterElement(nums1,nums2):
        res = [] # space O(n)
        for i in range(len(nums1)): # time O(m)
            for j in range(len(nums2)): # in worst case time O(n), so if m ~n = n2 time complexity
                if nums1[i] == nums2[j]:
                    k = j + 1
                    isfound = False
                    while k <len(nums2):
                        if nums2[k] > nums1[i]:
                            res.append(nums2[k])
                            isfound = True
                            break
                        k += 1
                    if isfound:
                        break
                    res.append(-1)

        return res

nums1 = [4,1,2]
nums2 = [1,3,4,2]
nextGreaterElement(nums1,nums2)
````

5. **https://leetcode.com/problems/next-greater-element-ii/**

````python
def nextGreaterElements(self, nums: List[int]) -> List[int]:
        res = []
        for i in range(len(nums)):
            is_append = False
            j = i + 1
            while j != i:
                if j == len(nums):
                    j = 0
                if nums[j] > nums[i]:
                    res.append(nums[j])
                    is_append = True
                    break

                if j == len(nums) -1:
                    j = 0
                else:
                    j += 1
            if not is_append:
                res.append(-1)

        return res
````

````python
def nextgreaterelements(nums):
    n = len(nums)
    res = [-1]*n
    stack = []
    # n = len(nums)
    for i in range(2*n, -1,-1):
        current = nums[i%n]
        while len(stack) != 0 and stack[-1] <= current:
            stack.pop()
        if i <n:
            if len(stack) == 0:
                res[i] = -1
            else:
                res[i] = stack[-1]
        stack.append(current)
    return res

nums = [1,2,3,4,3]
nextgreaterelements(nums)
````

6.**https://leetcode.com/problems/daily-temperatures/**

````python
# brute force
def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        res = []
        for i in range(len(temperatures)):
            count = 0
            for j in range(i+1, len(temperatures)):
                count += 1
                is_found = False
                if temperatures[j] > temperatures[i]:
                    is_found = True
                    break
            if is_found:
                res.append(count)
            else:
                res.append(0)
        return res
````

````python
# optimal 
def daily_temp_optimal(tem):
    res = [0] * len(tem)
    stack = []
    for i in range(len(tem)-1,-1,-1):
        while len(stack) != 0 and tem[stack[-1]] <= tem[i]:
            stack.pop()
        if len(stack) == 0:
            res[i] = 0
        else:
            res[i] = stack[-1] - i
        stack.append(i)
    return res

temperatures = [73,74,75,71,69,72,76,73]
daily_temp_optimal(temperatures)
````

Excellent. Now we’re entering real algorithm analysis territory.
Let’s go slowly and make this crystal clear.

You asked two things:

1. Why does the runtime become
   n + (n-k) + (n-2k) + ... ≈ O(n²)?
2. What are upper bound and lower bound?

Let’s handle them cleanly.

---

# PART 1 — Why n + (n-k) + (n-2k) + ... ≈ O(n²)

Your algorithm:

* Scan entire string
* Remove blocks of size k
* Rebuild string
* Repeat until no changes

Let initial length = n.

---

## Case: Worst shrinking behavior

Take worst case input:

```
s = "aaaaaaaaaa..." (n times)
k = 2
```

Each iteration removes only 2 characters.

So:

Iteration 1 → length = n
Iteration 2 → length = n-2
Iteration 3 → length = n-4
...
Until 0

Number of iterations ≈ n/2 ≈ O(n)

---

## Work per iteration

Each iteration scans the entire current string.

So work looks like:

```
n
+ (n-2)
+ (n-4)
+ ...
+ 2
```

This is an arithmetic series.

Formula for:

```
n + (n-2) + (n-4) + ... + 2
```

This equals approximately:

```
n² / 2
```

In Big-O we ignore constants:

⭐ O(n²)

---

## Why not O(n³) here?

Now here is the subtle part.

Inside your loop, you have:

```python
len(set(s[i:i+k]))
```

This costs O(k).

So inner scanning cost is actually:

O(n * k)

If k ≈ n (example: k = n/2)

Then:

O(n * n) = O(n²) per iteration

And if outer loop runs O(n) times

Then worst theoretical bound:

⭐ O(n³)

That is a *strict* worst upper bound.

But in practical shrinking behavior,
string length decreases, so total sums closer to O(n²).

---

# PART 2 — What Is Upper Bound?

Upper bound means:

> The maximum possible running time in worst case.

It guarantees:

```
T(n) ≤ f(n)
```

Example:

If we say:

```
T(n) = O(n²)
```

It means:

Algorithm will NEVER grow faster than n².

It might grow slower.

---

# Lower Bound

Lower bound means:

> The minimum possible running time in best case.

It guarantees:

```
T(n) ≥ g(n)
```

Example:

If best case always scans full string once:

Lower bound = Ω(n)

---

# Tight Bound (Theta)

If:

Upper bound = O(n²)
Lower bound = Ω(n²)

Then:

```
T(n) = Θ(n²)
```

Meaning exact growth rate.

---

# Simple Example

Consider:

```
for i in range(n):
    for j in range(n):
        print(i,j)
```

Upper bound = O(n²)
Lower bound = Ω(n²)
So exact = Θ(n²)

---

# Your Case

Best case:

If no duplicates found:

Only one pass.

⭐ Ω(n)

Worst case:

Repeated shrinking:

Between O(n²) and O(n³) depending on k.

So safe upper bound:

⭐ O(n³)

More realistic practical bound:

⭐ O(n²)

---

# Visual Summary

Upper bound → worst possible time
Lower bound → best possible time
Theta → exact growth

---

# Why We Care

In interviews, we usually state worst-case upper bound.

So your brute-force solution:

Safe answer:

⭐ Time = O(n²)
(or mention potential O(n³) due to slicing)

Space = O(n)

````python
# optimal 
# time complexity = O(n)
# space complexity = O(n)
def remove_all_adjacent_optimal(s,k):
    n = len(s)
    st = []
    for i in range(n):
        if len(st) == 0:
            st.append([s[i],1])
        elif st[-1][0] != s[i]:
            st.append([s[i],1])
        elif st[-1][1] < k-1 and st[-1][0] == s[i]:
            temp = st.pop()
            st.append([temp[0],temp[1] + 1])
        else:
            st.pop()
    # res = ""
    res_list = []
    while len(st) != 0:
        curr = st.pop()
        for i in range(curr[1]):
            res_list.append(curr[0])
    return "".join(res_list[::-1])
    # join() copies n characters once
    # += copies 1 + 2 + 3 + ... + n characters   , That repeated copying is why the complexity becomes O(n²).

a = "deeedbbcccbdaabd"
remove_all_adjacent_optimal(a,3)
````
