1. **https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/description/**

```python
def remove_all_adjacent_brute(st):
    run_loop_again = True
    while run_loop_again:# in worst case it will cost O(n)
        run_loop_again = False
        new_st = "" # space complexity = O(n)
        i = 0
        while i < len(st):  # everytime it will run O(n)  , so worst case time complexity = O(n2)
            if i < len(st) - 1 and st[i] == st[i+1]:
                run_loop_again = True
                i += 2
            else:
                new_st += st[i]
                i += 1
        st = new_st
    return st
s = "abbaca"
remove_all_adjacent_brute(s)
```

```python
# time complexity = O(n2)
# space complexity = O(n)
def remove_all_adjacent_duplicates(st):
    stack = []  # O(n) space
    for i in range(len(st)): # O(n)
        if len(stack) != 0 and stack[-1] == st[i]:
            stack.pop()
        else:
            stack.append(st[i])
    res = "" # O(n) space
    while len(stack) != 0:    # in worst case it can be O(n),,,, so this while loop in string concatenation will cose O(n2) in worst case
        res += str(stack.pop())  # String concatenation in Python is NOT O(1).
# Strings are immutable. Each concatenation copies the whole string. So this part costs: 1 + 2 + 3 + ... + n
    return res[::-1]   # this reversal also cost O(n)

s = "abbaca"
remove_all_adjacent_duplicates(s)
```

```python
# time complexity = O(n)
def remove_all_adjacent(st):
    stack = []
    for i in range(len(st)):
        if len(stack) != 0 and stack[-1] == st[i]:
            stack.pop()
        else:
            stack.append(st[i])
    return "".join(stack)

# How join Is Faster
# Calculates total required size
# Allocates memory once
# Copies all elements in one pass

# So:
# O(n)
# Much faster.

s = "abbaca"
remove_all_adjacent(s)
```

2. **https://leetcode.com/problems/valid-parentheses/description/**

```python
def isValid(s):
        st = []
        for i in range(len(s)):
            if s[i] == "(" or s[i] == "[" or s[i] == "{":
                st.append(s[i])
            elif s[i] == ")":
                if len(st) == 0 or st[-1] != "(":
                    return False
                st.pop()

            elif s[i] == "}":
                if len(st) == 0 or st[-1] != "{":
                    return False
                st.pop()
            elif s[i] == "]":
                if len(st) == 0 or st[-1] != "[":
                    return False
                st.pop()
        return len(st) == 0

s = "()()()"
isValid(s)
```

3. **https://www.geeksforgeeks.org/problems/previous-greater-element/1**

```python
#optimal
# time complexity = O(n)
#space complexity = O(n)
def preGreaterEle(arr):
		res = [] # space O(n)
		stack = []  # space (n) in worst case
		res.append(-1)
		stack.append(arr[0])
		for i in range(1,len(arr)):  # time O(n)
		    while len(stack) != 0 and stack[-1] <= arr[i]: # n + n = 2n, one pop and only one push
		        stack.pop()
		    if len(stack) == 0:
		        res.append(-1)
		    else:
		        res.append(stack[-1])
		    stack.append(arr[i])
		return res

arr = [10, 4, 2, 20, 40, 12, 30]
preGreaterEle(arr)
```

4. **https://leetcode.com/problems/next-greater-element-i/description/**

```python
# brute force
def nextGreaterElement(nums1,nums2):
        res = [] # space O(n)
        for i in range(len(nums1)): # time O(m)
            for j in range(len(nums2)): # in worst case time O(n), so if m ~n = n2 time complexity
                if nums1[i] == nums2[j]:
                    k = j + 1
                    isfound = False
                    while k <len(nums2):
                        if nums2[k] > nums1[i]:
                            res.append(nums2[k])
                            isfound = True
                            break
                        k += 1
                    if isfound:
                        break
                    res.append(-1)

        return res

nums1 = [4,1,2]
nums2 = [1,3,4,2]
nextGreaterElement(nums1,nums2)
```

```python
# space = O(n)
# time = O(n)
def nextgreaterelementoptimal(nums1,nums2):
    hashm = {}  # space O(n)
    res = []  # space O(k)
    stack = []
    stack.append(nums2[-1])
    hashm[nums2[-1]] = -1
    for i in range(len(nums2) - 2 , -1, -1): # O(n)
        while len(stack) != 0 and stack[-1] <= nums2[i]:  # O(n) in worst case, only pushing and poping once, so n + n = 2n
            stack.pop()
        if len(stack) == 0:
            hashm[nums2[i]] = -1
        else:
            hashm[nums2[i]] = stack[-1]
        stack.append(nums2[i])
    # print(stack)
    for j in range(len(nums1)): # time O(n)
        if nums1[j] in hashm:
            res.append(hashm[nums1[j]])
    return res

nums1 = [4,1,2]
nums2 = [1,3,4,2]
nextgreaterelementoptimal(nums1,nums2)
```

5. **https://leetcode.com/problems/next-greater-element-ii/**

```python
def nextGreaterElements(self, nums: List[int]) -> List[int]:
        res = []
        for i in range(len(nums)):
            is_append = False
            j = i + 1
            while j != i:
                if j == len(nums):
                    j = 0
                if nums[j] > nums[i]:
                    res.append(nums[j])
                    is_append = True
                    break

                if j == len(nums) -1:
                    j = 0
                else:
                    j += 1
            if not is_append:
                res.append(-1)

        return res
```

```python
def nextgreaterelements(nums):
    n = len(nums)
    res = [-1]*n
    stack = []
    # n = len(nums)
    for i in range(2*n, -1,-1):
        current = nums[i%n]
        while len(stack) != 0 and stack[-1] <= current:
            stack.pop()
        if i <n:
            if len(stack) == 0:
                res[i] = -1
            else:
                res[i] = stack[-1]
        stack.append(current)
    return res

nums = [1,2,3,4,3]
nextgreaterelements(nums)
```

6.**https://leetcode.com/problems/daily-temperatures/**

```python
# brute force
def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        res = []
        for i in range(len(temperatures)):
            count = 0
            for j in range(i+1, len(temperatures)):
                count += 1
                is_found = False
                if temperatures[j] > temperatures[i]:
                    is_found = True
                    break
            if is_found:
                res.append(count)
            else:
                res.append(0)
        return res
```

```python
# optimal 
def daily_temp_optimal(tem):
    res = [0] * len(tem)
    stack = []
    for i in range(len(tem)-1,-1,-1):
        while len(stack) != 0 and tem[stack[-1]] <= tem[i]:
            stack.pop()
        if len(stack) == 0:
            res[i] = 0
        else:
            res[i] = stack[-1] - i
        stack.append(i)
    return res

temperatures = [73,74,75,71,69,72,76,73]
daily_temp_optimal(temperatures)
```
