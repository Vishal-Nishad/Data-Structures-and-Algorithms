# 1. Subsets / Power Set

1. **https://leetcode.com/problems/subsets/description/**

````python
# time: O(2^n)
# space: auxiliary that is call stack here, O(n)  deep
class Solution:
    def subsets(self, nums):
        result = []
        def backtrack(start,temp_ar):
            if(start == len(nums)):
                result.append(temp_ar)
                return
            
            # include
            backtrack(start+1,temp_ar + [nums[start]])

            # exclude
            backtrack(start+1,temp_ar)
        backtrack(0,[])
        return result

nums = [1,2,3]
Solution().subsets(nums)
````

2. **https://leetcode.com/problems/subsets-ii/description/**

````python
class Solution:
    def subsetsWithDup(self, nums):
        nums.sort()
        result = []
        def backtrack(start,temp_ar):
            if(start == len(nums)):
                if temp_ar not in result:
                    result.append(temp_ar)
                return
            #include
            backtrack(start+1,temp_ar + [nums[start]])

            # exclude
            next_start = start + 1
            while next_start < len(nums) and nums[next_start] == nums[start]:
                next_start = next_start + 1
            backtrack(start+1,temp_ar)
        
        backtrack(0,[])
        return result
print(Solution().subsetsWithDup([1,2,2]))
print()
a = [4,1,4,4,4]
Solution().subsetsWithDup(a)
````

````python
# another for loop way which handle for duplicate and non duplicate cases
class Solution:
    def subsetsWithDup(self, nums):
        nums.sort()
        result = []
        def backtrack(start,temp_ar):
            result.append(temp_ar[:])
            for i in range(start,len(nums)):
                if(i>start and nums[i] == nums[i-1]):
                    continue
                
                #inclue
                temp_ar.append(nums[i])
                backtrack(i+1,temp_ar)
                temp_ar.pop()
        
        backtrack(0,[])
        return result
print(Solution().subsetsWithDup([1,2,2]))
print()
a = [4,1,4,4,4]
Solution().subsetsWithDup(a)
````

3. **https://leetcode.com/problems/partition-to-k-equal-sum-subsets/description/**

````python
class Solution:
    def canPartitionKSubsets(self, nums, k):
        sum_nums = sum(nums)
        if(sum_nums%k != 0):
            return False

        # sort in descending order to return early in case of big numbers which is > target
        nums.sort(reverse=True)
        target = sum_nums//k
        visited = [False] * len(nums)

        def dfs(start,ans_sum,k):
            # base case
            if(k==1):   # matematically the 1 bucket equal to target will definiterly exist so we can confidently return from 1 no need to check for 0
                return True
            
            # reset case
            if(ans_sum == target):
                return dfs(0,0,k-1)
            
            for i in range(start,len(nums)):
                if not visited[i] and ans_sum + nums[i] <= target:
                    visited[i] = True
                    if(dfs(i+1,ans_sum+nums[i],k)):
                        return True
                    visited[i] = False
            return False
        return dfs(0,0,k)

a = [1,2,2,3,4]
print(Solution().canPartitionKSubsets([1,2,2,3,7],3))
print()
Solution().canPartitionKSubsets(a,3)
````

## 6 Core Backtracking Pattern Types

### 1. Subsets / Power Set
Pick or skip each element. No ordering matters.

| # | Problem | Difficulty |
|---|---|---|
| LC 78 | Subsets | Easy |
| LC 90 | Subsets II (with duplicates) | Medium |
| LC 698 | Partition to K Equal Sum Subsets | Medium |

**Tree shape:** binary — at each step you either **include** or **exclude** an element.

***

### 2. Combinations
Pick `k` elements from `n`, order doesn't matter, no reuse (or with reuse).

| # | Problem | Difficulty |
|---|---|---|
| LC 39 | Combination Sum (reuse allowed) | Medium |
| LC 40 | Combination Sum II (no reuse, duplicates) | Medium |
| LC 77 | Combinations | Medium |
| LC 216 | Combination Sum III | Medium |

**Tree shape:** at each step, pick a number and move forward (never go back to previous index).

***

### 3. Permutations
Every ordering matters. All elements must be used.

| # | Problem | Difficulty |
|---|---|---|
| LC 46 | Permutations | Medium |
| LC 47 | Permutations II (with duplicates) | Medium |

**Tree shape:** at each step, pick any **unused** element from the remaining pool.

***

### 4. String Partitioning / Splitting
Cut the string into pieces, each satisfying a condition.

| # | Problem | Difficulty |
|---|---|---|
| LC 131 | Palindrome Partitioning ← you just did this | Medium |
| LC 93 | Restore IP Addresses | Medium |
| LC 140 | Word Break II | Hard |

**Tree shape:** at each step, try every possible **cut point** from current index.

***

### 5. Grid / Path Problems
Move in a grid (up/down/left/right), find all paths or valid arrangements.

| # | Problem | Difficulty |
|---|---|---|
| LC 79 | Word Search | Medium |
| LC 212 | Word Search II | Hard |
| Rat in a Maze | Classic | Medium |
| LC 980 | Unique Paths III | Hard |

**Tree shape:** at each cell, branch into 4 directions, mark visited, unmark on backtrack.

***

### 6. Constraint Satisfaction (Hard — Asked in Top Companies)
Place items satisfying strict rules. These are the "wow" problems in interviews.

| # | Problem | Difficulty |
|---|---|---|
| LC 51 | N-Queens | Hard |
| LC 37 | Sudoku Solver | Hard |
| LC 22 | Generate Parentheses | Medium |
| LC 1239 | Max Length of Concatenated String with Unique Chars | Medium |

**Tree shape:** at each step, try placing something valid, skip invalid placements entirely.

***

## The Order to Practice In

```
Subsets → Combinations → Permutations → String Splitting → Grid → Constraint Satisfaction
```

Each type builds on the previous. Subsets and combinations share the same "start index moves forward" idea. Permutations introduce a "visited" array. Grid introduces 2D visited marking. Constraint satisfaction combines all of it. [geeksforgeeks](https://www.geeksforgeeks.org/dsa/top-20-backtracking-algorithm-interview-questions/)

***

## One Key Difference Between Types

| Type | Do I reuse index? | Do I need visited[]? | Loop starts at |
|---|---|---|---|
| Subsets | No | No | `i` (current) |
| Combinations | No | No | `i` (current) |
| Permutations | No | Yes | `0` always |
| String Split | No | No | `end_index` from `staimpacteers.com/blogs/backtracking-problems-for-product-company-interviews-the-5-patterns-you-cannot-ignore)
