1. **https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/description/**

```python
def remove_all_adjacent_brute(st):
    run_loop_again = True
    while run_loop_again:# in worst case it will cost O(n)
        run_loop_again = False
        new_st = "" # space complexity = O(n)
        i = 0
        while i < len(st):  # everytime it will run O(n)  , so worst case time complexity = O(n2)
            if i < len(st) - 1 and st[i] == st[i+1]:
                run_loop_again = True
                i += 2
            else:
                new_st += st[i]
                i += 1
        st = new_st
    return st
s = "abbaca"
remove_all_adjacent_brute(s)
```

```python
# time complexity = O(n2)
# space complexity = O(n)
def remove_all_adjacent_duplicates(st):
    stack = []  # O(n) space
    for i in range(len(st)): # O(n)
        if len(stack) != 0 and stack[-1] == st[i]:
            stack.pop()
        else:
            stack.append(st[i])
    res = "" # O(n) space
    while len(stack) != 0:    # in worst case it can be O(n),,,, so this while loop in string concatenation will cose O(n2) in worst case
        res += str(stack.pop())  # String concatenation in Python is NOT O(1).
# Strings are immutable. Each concatenation copies the whole string. So this part costs: 1 + 2 + 3 + ... + n
    return res[::-1]   # this reversal also cost O(n)

s = "abbaca"
remove_all_adjacent_duplicates(s)
```

```python
# time complexity = O(n)
def remove_all_adjacent(st):
    stack = []
    for i in range(len(st)):
        if len(stack) != 0 and stack[-1] == st[i]:
            stack.pop()
        else:
            stack.append(st[i])
    return "".join(stack)

# How join Is Faster
# Calculates total required size
# Allocates memory once
# Copies all elements in one pass

# So:
# O(n)
# Much faster.

s = "abbaca"
remove_all_adjacent(s)
```

2. **https://leetcode.com/problems/valid-parentheses/description/**

```python
def isValid(s):
        st = []
        for i in range(len(s)):
            if s[i] == "(" or s[i] == "[" or s[i] == "{":
                st.append(s[i])
            elif s[i] == ")":
                if len(st) == 0 or st[-1] != "(":
                    return False
                st.pop()

            elif s[i] == "}":
                if len(st) == 0 or st[-1] != "{":
                    return False
                st.pop()
            elif s[i] == "]":
                if len(st) == 0 or st[-1] != "[":
                    return False
                st.pop()
        return len(st) == 0

s = "()()()"
isValid(s)
```
