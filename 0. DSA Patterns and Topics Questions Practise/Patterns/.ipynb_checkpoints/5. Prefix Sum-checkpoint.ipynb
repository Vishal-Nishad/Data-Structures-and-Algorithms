1. **https://leetcode.com/problems/find-pivot-index/submissions/1912395158/**

````python
# brute force 
# time complexity = O(n2)
# space complexity = O(1)
def find_pivot_index_brute(nums):
    for i in range(len(nums)):
        left_sum = 0
        right_sum = 0
        for l in range(i):
            left_sum += nums[l]
        for r in range(i+1,len(nums)):
            right_sum += nums[r]
        if left_sum == right_sum:
            return i
    return -1
nums = [2,1,-1]
find_pivot_index_brute(nums)
````

````python
# optimal
# time complexity = O(n)
# space complexity = O(1)
def find_pivot_index(nums):
    left = 0
    sum_nums = sum(nums) 
    for i in range(len(nums)):
        right = sum_nums - nums[i] - left
        if left == right:
            return i
        left += nums[i]
    return -1

nums = [2,1,-1]
find_pivot_index(nums)
````

2. **https://leetcode.com/problems/subarray-sum-equals-k/description/**

````python
# brute force 
# time complexity = O(n2)
# space complexity = O(1)
def sub_sum_k(nums,k):
    res = 0
    for i in range(len(nums)):
        sub_sum = 0
        for j in range(i,len(nums)):
            sub_sum += nums[j]
            if sub_sum == k:
                res += 1
    return res

nums = [1,2,3]
sub_sum_k(nums,3)
````

````python
# optimal
# time complexity = O(n2)
# space complexity = O(1)
def sub_sum_k_optimal(nums,k):
    prefix_hash = {0:1}  #0 exists so that subarrays starting at index 0 satisfy the same equation as all other subarrays, i.e sum_sub(l,r) = r - (l-1)
    prefix_sum = 0
    count = 0
    for i in range(len(nums)):
        prefix_sum += nums[i]
        if prefix_sum - k in prefix_hash:
            count += prefix_hash[prefix_sum - k]
        prefix_hash[prefix_sum] = prefix_hash.get(prefix_sum,0) + 1
    return count

nums = [4,5,0,-2,-3,1]
sub_sum_k_optimal(nums,5)
````

# 1️⃣ Formal Logical Derivation (Equations Only)

### Step A — Define prefix sum

Let:

[
P(i) = \sum_{t=0}^{i} nums[t]
]

This is the sum of the array from index `0` to index `i`.

---

### Step B — Express subarray sum using prefix sums

For any subarray from index `l` to `r`:

[
sum(l, r) = P(r) - P(l - 1)
]

This is always true because all terms before `l` cancel out.

---

### Step C — Impose the condition “sum = k”

We want:

[
P(r) - P(l - 1) = k
]

Rearrange:

[
P(l - 1) = P(r) - k
]

---

### Step D — Logical conclusion

For a **fixed right index `r`**:

> The number of subarrays ending at `r` with sum `k`
> equals the number of indices `(l-1)` such that
> `P(l-1) = P(r) - k`.

So the problem reduces to:

> **For each prefix sum `P(r)`, count how many times `P(r) - k` has appeared before.**

---

# 2️⃣ What You Say to the Interviewer (Exact Words)

> “I use prefix sums.
> Any subarray sum from `l` to `r` can be written as
> `prefix[r] - prefix[l-1]`.
>
> To get sum `k`, we need
> `prefix[l-1] = prefix[r] - k`.
>
> So while iterating, for each `r`, I count how many times
> `prefix[r] - k` has appeared earlier.
>
> I store frequencies of prefix sums in a hashmap,
> initialize it with `{0:1}` to handle subarrays starting at index `0`,
> and accumulate the count in one pass.”

That explanation is **complete and correct**.

# 3️⃣ Code (Optimal O(n))

```python
def subarraySum(nums, k):
    prefix_count = {0: 1}
    prefix_sum = 0
    count = 0

    for num in nums:
        prefix_sum += num

        if prefix_sum - k in prefix_count:
            count += prefix_count[prefix_sum - k]

        prefix_count[prefix_sum] = prefix_count.get(prefix_sum, 0) + 1

    return count
```

3. **https://leetcode.com/problems/subarray-sums-divisible-by-k/description/**

````python
# brute force
# space complexity = O(1)
# time complexity = O(n2)
def sub_sum_div_k(nums,k):
    count = 0
    for i in range(len(nums)):
        sub_sum = 0
        for j in range(i, len(nums)):
            sub_sum += nums[j]
            if sub_sum % k == 0:
                count += 1
    return count

nums = [4,5,0,-2,-3,1]
sub_sum_div_k(nums,5)
````

````python
# optimal
# space complexity 
# time complexity
def sub_sum_div_k_optimal(nums,k):
    prefix_sum = 0
    prefix_count = {0:1} # 0 is remainder before start of array
    result = 0
    for i in range(len(nums)):
        prefix_sum += nums[i]
        rem = prefix_sum % k
        if rem < 0:
            rem += k
        if rem in prefix_count:
            result += prefix_count[rem]
        prefix_count[rem] = prefix_count.get(rem,0) + 1
    return result

nums = [4,5,0,-2,-3,1]
sub_sum_div_k_optimal(nums,5)
````

# 1️⃣ What Does “Modulo” Actually Mean?

When we say:

[
a \bmod k = r
]

It means:

[
a = qk + r
]

Where:

* `q` is some integer
* `r` is the remainder
* `0 ≤ r < k`

Example with `k = 5`:

```
9 = 1×5 + 4  → remainder 4
14 = 2×5 + 4 → remainder 4
```

Both have remainder **4**.

---

# 2️⃣ Now Let’s Take Two Numbers With SAME Remainder

Let:

[
A = q_1 k + r
]
[
B = q_2 k + r
]

Both leave remainder `r` when divided by `k`.

Now subtract them:

[
A - B
]

Substitute:

[
(q_1 k + r) - (q_2 k + r)
]

Expand:

[
q_1 k + r - q_2 k - r
]

The `r` cancels:

[
(q_1 - q_2) k
]

That is:

[
k × (some\ integer)
]

Which means:

[
A - B \text{ is divisible by } k
]

🚨 THIS is what “equal remainders cancel out” means.

Not magic. Just algebra.

---

# 3️⃣ Now Connect This to Prefix Sums

Let:

[
P(r) = sum\ up\ to\ index\ r
]

If:

[
P(r) \bmod k = P(l-1) \bmod k
]

Then:

[
P(r) - P(l-1)
]

is divisible by `k`.

And:

[
P(r) - P(l-1)
]

is exactly the subarray:

[
nums[l] + \dots + nums[r]
]

So the subarray is divisible by `k`.

---

# 4️⃣ Let’s Use Small Concrete Numbers

Take:

```
k = 5
```

Suppose:

```
Earlier prefix sum = 9
Current prefix sum = 14
```

Compute remainders:

```
9 % 5 = 4
14 % 5 = 4
```

Same remainder.

Subtract:

```
14 - 9 = 5
```

5 is divisible by 5.

That difference corresponds to the subarray between those two prefix points.

# 6️⃣ Why We Count Combinations

If remainder `4` appears 4 times:

Call them:

```
R1, R2, R3, R4
```

Every pair:

```
(R1,R2)
(R1,R3)
(R1,R4)
(R2,R3)
(R2,R4)
(R3,R4)
```

forms a valid subarray.

That is:

[
\binom{4}{2} = 6
]

Because any two equal remainders subtract to a multiple of `k`.

---

# 7️⃣ The Deep Intuition

Modulo arithmetic only keeps the “leftover part”.

If two numbers leave the **same leftover**,
their difference has **no leftover**.

“No leftover” means remainder = 0.
Which means divisible by `k`.

That’s it.

# 8️⃣ One Final Compact Summary

If:

[
a \equiv b \pmod{k}
]

Then:

[
a - b \equiv 0 \pmod{k}
]

This is a basic property of modular arithmetic.

We are just applying it to prefix sums.
