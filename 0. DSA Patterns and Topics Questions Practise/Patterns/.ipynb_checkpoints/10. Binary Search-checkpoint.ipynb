1. **https://leetcode.com/problems/binary-search/description/**

```python
def bin_search(nums,target):
    left = 0
    right = len(nums)-1
    while left <= right:
        mid = (left+right)//2
        if nums[mid] == target:
            return mid
        elif target > nums[mid]:
            left = mid+1
        else:
            right = mid - 1
    return -1

nums = [-1,0,3,5,9,12]
target = 9
bin_search(nums,target)
```

3. **https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/**

```python
# try brute force here
```

```python
# optimal logn
def searchRange(nums, target):
        def left(nums,target):
            res = -1
            l = 0
            r = len(nums)-1
            while l <= r:
                mid = (l+r)//2
                if nums[mid] == target:
                    res = mid
                    r = mid - 1
                elif target > nums[mid]:
                    l = mid+1
                else:
                    r = mid - 1
            return res
        
        def right(nums,target):
            res = -1
            l = 0
            r = len(nums)-1
            while l <= r:
                mid = (l+r)//2
                if nums[mid] == target:
                    res = mid
                    l = mid + 1
                elif target > nums[mid]:
                    l = mid+1
                else:
                    r = mid - 1
            return res
        
        return [left(nums,target),right(nums,target)]

nums = [5,7,7,8,8,10]
target = 8
print(searchRange(nums,target))

searchRange(nums,9)
```

```python
# single pass more optimized solution
def searchRangeOptimized(nums, target):
        def binary_search(nums, target, is_searching_left):
            left = 0
            right = len(nums) - 1
            idx = -1
            
            while left <= right:
                mid = (left + right) // 2
                
                if nums[mid] > target:
                    right = mid - 1
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    idx = mid
                    if is_searching_left:
                        right = mid - 1
                    else:
                        left = mid + 1
            
            return idx
        
        left = binary_search(nums, target, True)
        right = binary_search(nums, target, False)
        
        return [left, right]

nums = [5,7,7,8,8,10]
target = 8
print(searchRangeOptimized(nums,target))

searchRangeOptimized(nums,9)
```

4. **https://www.geeksforgeeks.org/problems/number-of-occurrence2259/1**

```python
# try brute force here
```

```python
#logn
def countFreq(arr, target):
        # code here
        def first(arr,target):
            l = 0
            r = len(arr)-1
            res = -1
            while l<=r:
                mid = (l+r)//2
                if arr[mid] == target:
                    res = mid
                    r = mid - 1
                elif target > arr[mid]:
                    l = mid + 1
                else:
                    r = mid - 1
            return res
        
        start = first(arr,target)
        if start == -1:
            return 0
        
        def last(arr,target):
            l = 0
            r = len(arr)-1
            res = -1
            while l<=r:
                mid = (l+r)//2
                if arr[mid] == target:
                    res = mid
                    l = mid + 1
                elif target > arr[mid]:
                    l = mid + 1
                else:
                    r = mid - 1
            return res
        
        end = last(arr,target)
        return end - start + 1

arr = [1, 1, 2, 2, 2, 2, 3]
target = 2
countFreq(arr,target)
```

5. **https://leetcode.com/problems/peak-index-in-a-mountain-array/description/**

```python
# try brute force here
```

```python
# optimal logn
def peakIndexInMountainArray(arr):
        l = 0
        r = len(arr)-1
        res = -1
        while l<= r:
            mid = (l+r)//2
            if arr[mid] < arr[mid+1]:
                l = mid+1
            else:
                res = mid
                r = mid-1
        return res

arr = [0,10,5,2]
peakIndexInMountainArray(arr)
```

6. **https://leetcode.com/problems/find-peak-element/**

```python
# logn
def findPeakElement(nums):
        st = 0
        en = len(nums)-1
        res = -1
        while st<=en:
            mid = (st+en)//2
            if (len(nums)> mid+1 and nums[mid]<nums[mid+1]):
                st = mid+1
            else:
                res = mid
                en = mid-1
        return res

nums = [1,2,1,3,5,6,4]
findPeakElement(nums)
```

7. **https://leetcode.com/problems/find-minimum-in-rotated-sorted-array/description/**

```python
#optimal logn
def findMin(nums):
        res = 0
        if nums[0]<nums[-1]:
            return nums[0]
        st = 0
        en = len(nums)-1
        while st<=en:
            mid = (st+en)//2
            if nums[mid]>nums[-1]:
                st = mid+1
            else:
                res = nums[mid]
                en = mid-1
        return res

nums = [4,5,6,7,0,1,2]
findMin(nums)
```

8. **https://www.geeksforgeeks.org/problems/rotation4723/1**

```python
# logn
def findKRotation(arr):
        # code here
        res = 0
        if arr[0]<arr[-1]:
            return 0
        st = 0
        en = len(arr)-1
        while st<=en:
            mid = (st+en)//2
            if arr[mid]>arr[-1]:
                st = mid+1
            else:
                res = mid
                en = mid-1
        return res

nums = [4,5,6,7,0,1,2]
findKRotation(nums)
```
