1. **https://leetcode.com/problems/maximum-subarray/description/?utm_source=chatgpt.com**

````python
# brute force
# space complexity = O(1)
# time complexity = O(n2)
def maxsubarray_brute(nums):
    result = float("-inf")
    n = len(nums)
    for i in range(n):
        max_sum = 0
        for j in range(i,n):
            max_sum += nums[j]
            if max_sum > result :
                result = max_sum
    return result

nums = [5,4,-1,7,8]
maxsubarray_brute(nums)
````

````python
# optimal kadane algo
# time complexity = O(n)
# space complexity = O(1)
def maxsubarray_optimal(nums):
    result = nums[0]
    best = nums[0]
    for i in range(1,len(nums)):
        if best + nums[i] > nums[i]:
            best += nums[i]
        else:
            best = nums[i]
        if best > result:
            result = best
    return result

nums = [5,4,-1,7,8]
maxsubarray_optimal(nums)
````

3. **https://leetcode.com/problems/maximum-product-subarray/?utm_source=chatgpt.com**

````python
# space complexity = O(1)
# time complexity = O(n2)
def maxproduct_brute(nums):
    result = float("-inf")
    n = len(nums)
    for i in range(n):
        max_sum = 1
        for j in range(i,n):
            max_sum *= nums[j]
            if max_sum > result :
                result = max_sum
    return result

nums = [2,3,-2,4]
maxproduct_brute(nums)
````

````python
# optimal
# space complexity = O(1)
# time complexity = O(n)

def maxprod_optimal(nums):
    max_prod = nums[0]
    min_prod = nums[0]
    result = nums[0]
    for i in range(1,len(nums)):
        temp_max = max(nums[i],nums[i]*max_prod,nums[i]*min_prod)
        min_prod = min(nums[i],nums[i]*max_prod,nums[i]*min_prod)
        max_prod = temp_max
        if max_prod > result:
            result = max_prod
    return result

nums = [2,3,-2,4]
maxprod_optimal(nums)
````

4. **https://leetcode.com/problems/maximum-subarray-sum-with-one-deletion/description/**

````python
# brute force
# space complexity = O(n)
# time complexity = O(n4)
def mx_sum_one_del(arr):
    max_sum = arr[0]
    for i in range(len(arr)):  # n
        for j in range(i,len(arr)): # n*n
            sub_arr = arr[i:j+1] # space n
            # case 1: no deletion
            max_sum = max(max_sum,sum(sub_arr))  # n*n*n

            # case 2: one deletion
            if len(sub_arr) > 1:
                for k in range(len(sub_arr)):      # n*n*n
                    new_sub_sum = sum(sub_arr) - sub_arr[k] # n*n*n*n hence n4
                    max_sum = max(max_sum,new_sub_sum)
    return max_sum

arr = [1,-2,0,3]
mx_sum_one_del(arr)
````

````python
# one deletion + Kadane = Kadane with TWO lanes 

# two states in this question
# One delete → either extend or delete current.
# No delete → normal Kadane.

# time complexity O(n)
# space complexity O(1)
def mx_sum_one_del_optimal(arr):
    no_del = arr[0]
    one_del = float("-inf")
    result = arr[0]  # suppose if there is a single element, then just return that,bcz loop will not run for single element

    for i in range(1,len(arr)):
        one_del = max(one_del + arr[i], no_del)
        no_del = max(no_del + arr[i],arr[i])
        result = max(result,one_del,no_del)
    return result
arr = [1,-2,0,3]
mx_sum_one_del_optimal(arr)
````
