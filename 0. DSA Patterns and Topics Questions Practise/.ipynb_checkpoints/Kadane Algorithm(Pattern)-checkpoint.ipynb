```python
# brute force
# space complexity = O(1)
# time complexity = O(n2)
def maxsubarray_brute(nums):
    result = float("-inf")
    n = len(nums)
    for i in range(n):
        max_sum = 0
        for j in range(i,n):
            max_sum += nums[j]
            if max_sum > result :
                result = max_sum
    return result

nums = [5,4,-1,7,8]
maxsubarray_brute(nums)
```

```python
# optimal kadane algo
# time complexity = O(n)
# space complexity = O(1)
def maxsubarray_optimal(nums):
    result = nums[0]
    best = nums[0]
    for i in range(1,len(nums)):
        if best + nums[i] > nums[i]:
            best += nums[i]
        else:
            best = nums[i]
        if best > result:
            result = best
    return result

nums = [5,4,-1,7,8]
maxsubarray_optimal(nums)
```
