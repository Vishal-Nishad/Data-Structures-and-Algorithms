## Two Pointers:
1.**https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/submissions/1876732526/**

```python
# 167. Two Sum II - Input Array Is Sorted
# brute force approach, time complexity O(n2)
def two_sum(arr,target):
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] + arr[j] == target:
                return [i+1,j+1]

ar = [2,7,11,15]
two_sum(ar,9)
```

```python
# Hashmap approach, time complexity = O(n), space complexity = O(n)
def two_sum_hash(ar, target):
    seen = {}
    n = len(ar)
    for i in range(n):
        needed = target - ar[i]
        if needed in dict:
            return [seen[needed]+1, i+1]
        else:
            dict[ar[i]] = i

ar = [2,7,11,15]
two_sum_hash(ar, 26)
```

```python
# Optimal two pointer approach, time complexity = O(n) if array is already sorted, otherwise if sorting needed then O(nlogn)
# space complexity O(1) constant we are not using any etc space 
def two_sum_optimal(ar, target):
    i,j = 0, len(ar) - 1
    while i < j:
        s = ar[i] + ar[j]
        if s == target:
            return [i+1, j+1]
        # if sum is greater than target then we will move j to the left to decrease the sum
        elif s > target:
            j -= 1
        # else sum is less than target then we will move i to the right to increase the sum
        else:
            i += 1
    return -1
ar = [2,7,11,15]
two_sum_optimal(ar, 26)
```

2. **https://leetcode.com/problems/remove-duplicates-from-sorted-list/description/**

```python
# Remove duplicates from sorted array
# Brute force, space complexity = O(n), time complexity = O(n)
def rem_duplicated(arr):
    res = [arr[0]]
    for j in range(1,len(arr)):
        if res[-1] != arr[j]:
            res.append(arr[j])
    return len(res)
ar = [2,3,3,4,4,5,6]
rem_duplicated(ar)
```

```python
# Optimal two pointer
def rem_duplicate_optimal(arr):
    i,j,result= 0,1,1
    while j<len(arr):
        if arr[j] != arr[i]:  # we can compare j with j-1 means with previos element as well,
            arr[i+1] = arr[j]
            i += 1
            j += 1
            result += 1
        else:
            j +=1
    return i+1, result
ar = [2,3,3,4,4,5,6]
print(rem_duplicate_optimal(ar))

def rem_duplicate_optimal_easier(arr):
    i = 0
    for j in range(1,len(arr)):
        if arr[i] != arr[j]: 
            arr[i+1] = arr[j]
            i +=1
    return i+1

ar = [2,3,3,4,4,5,6]
rem_duplicate_optimal_easier(ar)
```

3. **https://leetcode.com/problems/squares-of-a-sorted-array/description/**

```python
# return squares of sorted array, array can contain negative positive elements
# Brute force, just do the square and sort and return, time complexity = O(nlogn), because of doing sorting
def sq_sorted_brute(arr):
    sq_sorted = [i**2 for i in arr]
    arr = sq_sorted
    copy = sq_sorted
    # doing manually, O(n2) time complexity
    for i in range(len(sq_sorted)):
        min_value = copy[i]
        min_ind = i     
        for j in range(i+1,len(sq_sorted)):
            if min_value > copy[j]:
                min_value = copy[j]
                min_ind = j
        copy[min_ind], copy[i] = copy[i], copy[min_ind]
    return sq_sorted, arr,copy
ar = [-4, -1, 0, 4,3,5]
sq_sorted_brute(ar)
```

```python
# optimal approach using two pointer
def sq_sorted_optimal(arr):
    neg = []
    pos = []
    for i in range(len(arr)):
        if arr[i]<0:
            neg.append(arr[i]**2)
        else:
            pos.append(arr[i]**2)
    if len(neg) == 0:
        return pos
    neg = neg[::-1]
    
    res_sorted = []
    i,j = 0, 0
    while i<len(neg) and j < len(pos):
        if neg[i] <= pos[j]:
            res_sorted.append(neg[i])
            i +=1
        else:
            res_sorted.append(pos[j])
            j +=1
    # if neg element still remainin
    while i < len(neg):
        res_sorted.append(neg[i])
        i += 1
    while j < len(pos):
        res_sorted.append(pos[j])
        j +=1
    return res_sorted
ar = [-4, -1, 0, 3,4,5]
sq_sorted_optimal(ar)
```

4.**https://leetcode.com/problems/3sum/description/**

```python
# Triplet sum to zero
# brute force, time complexity = O(n3), space complexity = O(k) i.e total number of triplets
def sum_to_zero(arr):
    n = len(arr)
    result = set()
    for i in range(n):
        for j in range(i+1, n):
            for k in range(j+1,n):
                if arr[i] + arr[j] + arr[k] == 0:
                    element = tuple(sorted([arr[i],arr[j],arr[k]]))
                    result.add(element)
    return list(result)

ar = [-1,0,1,2,-1,-4]
sum_to_zero(ar)
```

```python
# Optimal using tow pointers
# Time complexity = O(n2)
# space complexity
# Auxiliary space (working memory)
# What extra memory does the algorithm use to compute?
# Loop variables: i, j, k
# Temporary list [arr[i], arr[j], arr[k]]
# Temporary tuple
# All of these are constant size.
# 𝑂(1)
# auxiliary space
# O(1)
# 	​
# 🔹 Output space (result storage)
# result = set()


# This stores unique triplets.

# Let:𝑘
# k = number of valid unique triplets in the output
# Then:
# Each triplet uses constant space
# Total space = proportional to number of triplets
# 𝑂(𝑘)
# output space
# O(k)
# 	​
# This is unavoidable, because the problem asks you to return all triplets.
# ✅ Final Space Complexity (properly stated)
# Auxiliary Space: O(1)
# Output Space: O(k), where k is the number of unique triplets
def sum_to_zero_optimal(arr):
    arr = sorted(arr)
    result = []
    for i in range(len(arr) - 2):
        if (i>0 and arr[i] == arr[i-1]):
            continue
        target = -arr[i]
        j,k = i+1, len(arr)-1
        while (j<k):
            if arr[j] + arr[k] == target:
                result.append([arr[i],arr[j],arr[k]])
                # print(result)
                j += 1
                k -= 1
                while (j<k and arr[j] == arr[j-1]):
                    j += 1
                while (j<k and arr[k] == arr[k+1]):
                    k -=1
            elif arr[j] + arr[k] < target:
                j +=1
            else:
                k -= 1
    return result
ar = [-1,0,1,2,-1,-4]
sum_to_zero_optimal(ar)
```

5.**https://leetcode.com/problems/3sum-closest/description/**

```python
# time complexity = O(n3), space complexity = O(1)
def t_sum_to_closest(arr, target):
    n = len(arr)
    min_diff = float("inf")
    result = 0
    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                sum = arr[i] + arr[j] + arr[k]
                if sum == target:
                    return sum
                diff = abs(target - sum)
                if (diff<min_diff):
                    min_diff = diff
                    result = sum
                    
    return min_diff, result
ar = [-1,2,1,-4]
t_sum_to_closest(ar,1)
```

```python
# time complexity = O(n2), space complexity = O(1)
def t_sum_to_closest_optimal(arr,target):
    # n = len(arr)
    arr.sort()
    # min_diff = float("inf")
    result= arr[0] + arr[1] + arr[2]
    for i in range(len(arr)-2):
        left,right = i+1, len(arr)-1
        while(left<right):
            total = arr[i] + arr[left] + arr[right]
            if abs(target - total)<abs(target - result):
                result = total
            if result == target:
                return result
            elif total < target:
                left +=1
            else:
                right -=1
    return result
ar = [-1,2,1,-4]
t_sum_to_closest_optimal(ar,1)
```

6. **https://www.geeksforgeeks.org/problems/count-triplets-with-sum-smaller-than-x5549/1**

```python
# brute force approach 
# time complexity = O(n3) and space complexity = O(1)
def count_smaller_triplet_brute(arr,sum):
    n = len(arr)
    ans = 0
    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                if arr[i]+arr[j]+arr[k] < sum:
                    ans += 1
    return ans

ar = [30,8,23,6,10,9,31,7,19,20,1,33,21,27,28,3,25,26]
print(len(ar))
count_smaller_triplet_brute(ar, 86)
```

```python
# two pointer optimal approach
# time complexity = O(n2) and space complexity = O(1)
def count_smaller_triplet_brute_optimal(arr,sum):
    arr.sort()
    n = len(arr)
    ans = 0
    for i in range(n-2):
        left, right = i+1, n-1
        while (left < right):
            total = arr[i] + arr[left] + arr[right]
            if total >= sum:
                right -= 1
            else:
                # for a left we have to consider all valid triplets, means here in else all left numbers from right, together with i+left+..... must be 
                # less then sum so we are including all count in ans, i.e indices (High - low)
                ans = ans + (right - left)
                left += 1
    return ans
ar = [30,8,23,6,10,9,31,7,19,20,1,33,21,27,28,3,25,26]
print(len(ar))
count_smaller_triplet_brute_optimal(ar, 86)
```

*7.**https://leetcode.com/problems/sort-colors/submissions/1881682128/**  Dutch National Flag Problem

```python
# Brute Force, Just do Sorting
# time complexity = O(n2), space complexity = O(1)
def ducth_national_flag_sorting(arr):
    for i in range(len(arr)):
        min_value = arr[i]
        min_ind = i
        for j in range(i+1,len(arr)):
            if arr[j] < min_value:
                min_value = arr[j]
                min_ind = j
        arr[i], arr[min_ind] = arr[min_ind], arr[i]
    return arr
nums = [2,0,2,1,1,0]
ducth_national_flag_sorting(nums)
```

```python
# using 3 different array to store 0,1,2, and then modifying the original array in inplace
# time complexity = O(n), space complexity = O(n)
def dutch_flag_creating_three_arr(arr):
    zero = []
    one = []
    two = []
    for i in range(len(arr)):
        if arr[i] == 0:
            zero.append(arr[i])
        elif arr[i] == 1:
            one.append(arr[i])
        else:
            two.append(arr[i])
    for z in range(len(zero)):
        arr[z] = zero[z]
    for o in range(len(one)):
        arr[(len(zero)+o)] = one[o]
    for t in range(len(two)):
        arr[(len(zero) + len(one) + t)] = two[t]
    return arr

nums = [2,0,2,1,1,0]
dutch_flag_creating_three_arr(nums)
```

```python
# now rather than using three list of 0,1,2, just use variable names with count 
# time complexity = O(n), space complexity = O(1)
def dutch_flag_three_variable_count(arr):
    zero = 0
    one = 0
    two = 0
    for i in range(len(arr)):
        if arr[i] == 0:
            zero += 1
        elif arr[i] == 1:
            one += 1
        else:
            two +=1
    print(zero,one,two)
    for z in range(zero):
        arr[z] = 0
    for o in range(zero, zero+one):
        arr[o] = 1
    for t in range(zero+one, zero+one+two):
        arr[t] = 2
    return arr
nums = [2,0,2,1,1,0]
dutch_flag_three_variable_count(nums)
```

```python
# What does “optimal” REALLY mean?
# When we say “optimal”, we must specify what dimension:
# - Time complexity
# - Space complexity
# - Number of passes (constant-factor optimization)

# Dutch Flag, Passes -> 1
def dutch_flag_sorting(arr):
    low, mid, high = 0, 0, len(arr) - 1
    while (mid<=high):
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        else:
            arr[high], arr[mid] = arr[mid], arr[high]
            # we are not moving mid because we have not check what is coming from high 
            # mid is checked means it is 2 so the region end is now sure so we will move high to left
            high -= 1
    return arr
nums = [2,0,2,1,1,0]
dutch_flag_sorting(nums)
```

## Sliding Window
1. **https://www.geeksforgeeks.org/problems/max-sum-subarray-of-size-k5313/1**

```python
# brute force approach
# Outer loop → n-k+1 ≈ n
# Inner loop → k
# O(n*k)
# space complexity = O(1)
def max_sub_array_brute(arr,k):
    n = len(arr)
    max_sum = float("-inf")
    for i in range(n-k+1):
        sum = 0
        for j in range(i,k+i):
            sum += arr[j]
        max_sum = max(max_sum,sum)
    return max_sum
ar = [100, 200, 300, 500]
max_sub_array_brute(ar,2)
```

```python
# Sliding Window Optimal way
# time complexity = O(n)
# space complexity = O(1)
def max_sub_array_of_k(arr,k):
    n = len(arr)
    sum = 0
    for i in range(k):
        sum += arr[i]
    max = sum
    low, high = 1, k
    while high < n:
        sum = sum - arr[low - 1] + arr[high]
        if sum > max:
            max = sum
        low += 1
        high += 1
    return max
ar = [100, 200, 300, 400]
max_sub_array_of_k(ar,2)
```

2. **https://leetcode.com/problems/minimum-size-subarray-sum/**

```python
# time complexity = O(n2)
# Space complexity = O(1)
def minimum_size_subarray_sum_brute(arr, target):
    n = len(arr)
    min_len = float("inf")
    # sum = 0
    for i in range(n):
        sum = arr[i]
        if sum >=target:
            return 1
        for j in range(i+1,n):
            sum += arr[j]
            if sum>=target:
                current_len = j-i+1
                if current_len<min_len:
                    min_len = current_len
    if min_len != float("inf"):
        return min_len
    else:
        return 0

ar = [2,3,1,2,4,3]
minimum_size_subarray_sum_brute(ar, 7)
```

```python
# time complexity = O(n)
# Space complexity = O(1)
def minimum_size_subarray_sum_optimal(arr,target):
    min_len = float("inf")
    sum = 0
    low, high = 0, 0
    while high < len(arr):
        sum = sum + arr[high]
        while sum>=target:
            current_len = high - low + 1
            if current_len < min_len:
                min_len = current_len
            sum = sum -arr[low]
            low += 1
        high += 1
    if low > 0:
        return min_len
    else:
        return 0

ar = [2,3,1,2,4,3]
minimum_size_subarray_sum_optimal(ar, 7)
```

*3. **https://www.geeksforgeeks.org/problems/longest-k-unique-characters-substring0853/1**

```python
# Memory contains:
# substring  → O(n)
# set        → O(n)
# Total Space complexity = O(n

# Time complexity = O(n2)
def longest_k_unique_substring(st,k):
    max_len=float("-inf")
    n = len(st)
    for i in range(n):
        string = ""
        for j in range(i,n):
            string += st[j]
            if len(set(string)) == k:
                if len(string) > max_len:
                    max_len = len(string)
    return (max_len if max_len != float("-inf") else -1) 
st= "aabacbebebe"
longest_k_unique_substring(st,3)
```

```python
# space complexity = O(k), using hashmap for storing all unique character count
# time complexity = O(n)

def longest_k_unique_substring_optimal(st,k):
    left = 0
    freq = {}
    distinct_count = 0 
    max_len = -1
    n = len(st)
    for right in range(n):
        if st[right] not in freq or freq[st[right]] == 0:
            distinct_count += 1
            freq[st[right]] = 0
        freq[st[right]] += 1
        while distinct_count > k:
            freq[st[left]] -= 1
            if freq[st[left]] ==0:
                distinct_count -= 1
            left += 1
        if distinct_count ==k:
            if right - left + 1 > max_len:
                max_len = right -left + 1
    return max_len
st= "aabacbebebe"
longest_k_unique_substring_optimal(st,3)
```

4. **https://leetcode.com/problems/fruit-into-baskets/**

```python
# Space complexity = O(n)
# Time complexity = O(n2)
def fruit_into_basket_brute(arr,k):
    n = len(arr)
    max_len = -1
    for left in range(n):
        subarray = []
        for right in range(left, n):
            subarray.append(arr[right])
            if len(set(subarray))<= k:
                if len(subarray)>max_len:
                    max_len = len(subarray)
    return max_len

arr = [0,1,2,2]
fruit_into_basket_brute(arr, 2)
```

```python
# time complexity = O(n)
# Space complexity = O(k)
def fruit_into_basket_optimal(arr):
    max_fruit = -1
    low = 0
    freq ={}
    distinct = 0
    for right in range(len(arr)):
        if arr[right] not in freq or freq[arr[right]] == 0:
            distinct += 1
            freq[arr[right]] = 0
        freq[arr[right]] += 1
        while distinct > 2:
            freq[arr[low]] -= 1
            if freq[arr[low]] == 0:
                distinct -= 1
            low += 1
        if distinct <= 2:
            if right - low + 1 > max_fruit:
                max_fruit = right - low + 1
    return max_fruit
arr = [0,1,2,2]
fruit_into_basket_optimal(arr)
```

5. **https://leetcode.com/problems/longest-substring-without-repeating-characters/description/**

```python
# brute force
# time complexity = O(n3)
# space complexity = O(n), we are creating new n length of substring, and also set inbetween
# Combined cost per inner iteration
# Inside the inner loop:
# O(n)  (string concat) + O(n) (set creation) = O(n)
# 🔥 Total time complexity
# Outer loop  → n
# Inner loop  → n
# Inner work  → n
# 𝑂(𝑛3)
def longest_substring_without_dup_brute(st):
    max_st_len = 0
    n = len(st)
    for left in range(n):
        substring = ""
        for right in range(left,n):
            substring += st[right]
            if len(substring) == len(set(substring)):
                if len(substring) > max_st_len:
                    max_st_len = len(substring)
    return max_st_len
st = "abcabcbb"
longest_substring_without_dup_brute(st)
```

```python
# Time complexity = O(n)
# Space complexity = O(n) in general, O(1) if character are fixed means small upper , as alphabets has only 26 characters 
def longest_substring_without_dup_optimal(st):
    low = 0
    long_st_len = 0
    freq = {}
    n = len(st)
    distinct = 0
    for right in range(n):
        if st[right] not in freq or freq[st[right]] == 0:
            distinct += 1
            freq[st[right]] = 0
        freq[st[right]] += 1
        while distinct < right - low + 1:
            freq[st[low]] -= 1
            if freq[st[low]] == 0:
                distinct -= 1
            low += 1
        if right - low + 1 == distinct:
            if long_st_len < right - low + 1:
                long_st_len = right - low + 1
    return long_st_len
st = "abcabcbb"
longest_substring_without_dup_optimal(st)
```

6. **https://leetcode.com/problems/longest-repeating-character-replacement/**

```python
# brute force
# Key - To make a substring uniform, convert all characters except the most frequent one.
# So replacements needed:
# replacements = length_of_substring - max_frequency

# time complexity = O(n2)
# space complexity = O(1), yes we creating freq = {}, but since it is for only uppercase letter so it is limited to 26 hence it is constant
def longest_rep_char_repl_brute(st, k):
    n =len(st)
    max_len = 0
    for left in range(n):
        freq = {}                  # Because each new left starts a NEW substring/window.
                                   # So its frequency counts and max frequency must be RESET.
        max_freq = 0
        for right in range(left,n):
            if st[right] not in freq:
                freq[st[right]] = 0
            freq[st[right]] += 1
            if max_freq < freq[st[right]]:
                max_freq = freq[st[right]]
            win_len = right - left + 1
            if win_len - max_freq <= k:
                if win_len > max_len:
                    max_len = win_len
    return max_len
st = "ABAB"
longest_rep_char_repl_brute(st,2)
```

```python
# space complexity = O(1), fixed/constant for alphabets
# time complexity = O(n)
def longest_rep_char_replace_optimal(st,k):
    max_len = 0
    left = 0
    freq = {}
    max_freq = 0
    for right in range(len(st)):
        freq[st[right]] = freq.get(st[right],0) + 1
        if freq[st[right]] > max_freq:
            max_freq = freq[st[right]]
        while (right - left + 1) - max_freq > k:
            freq[st[left]] -= 1
            left += 1
        if right - left + 1 > max_len:
            max_len = right - left + 1
    return max_len

st = "ABAB"
longest_rep_char_replace_optimal(st,2)
```
