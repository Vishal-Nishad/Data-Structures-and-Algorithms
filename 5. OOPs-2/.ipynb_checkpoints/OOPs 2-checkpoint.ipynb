```python
class Vehicle:
    def __init__(self,color,max_speed):
        self.color=color
        self.max_speed=max_speed
class Car(Vehicle):
    def __init__(self,color,max_speed,num_gears,is_convertible):
        super().__init__(color,max_speed)
        self.num_gears=num_gears
        self.is_convertible=is_convertible
    def print_car(self):
        print(f"Color is : {self.color}")
        print(f"Max Speed : {self.max_speed}")
        print(f"Number of gears are : {self.num_gears}")
        print(f"Is convertible : {self.is_convertible}")

c=Car("White",200,6,False)
c.print_car()
```
