# Heap(its 2 property)
## Complete Binary Tree
- All level except last level should be filled
- elements in level should be inserted from left to right

## Heap Order Property
- can be Min Heap -> every parent should be less than its child
- can be Max Heap -> every parent should be more than its child

# How to store CBT

- using array, will cost O(logn)
