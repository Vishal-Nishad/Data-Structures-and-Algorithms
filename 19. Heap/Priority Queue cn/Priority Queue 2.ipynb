Heap sort **https://leetcode.com/problems/sort-an-array/description/**

```python
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        def max_down_heapify(nums,parent_index,n):
            while True:
                largest = parent_index
                left_indx = 2*parent_index +1
                right_indx = 2*parent_index +2

                if(left_indx < n and nums[left_indx] > nums[largest]):
                    largest = left_indx
                if(right_indx < n and nums[right_indx] > nums[largest]):
                    largest = right_indx
                
                if(parent_index == largest):
                    break
                nums[largest],nums[parent_index] = nums[parent_index],nums[largest]
                parent_index = largest
        n = len(nums)
        for i in range((n//2)-1,-1,-1):
            max_down_heapify(nums,i,n)
        
        for end in range(n-1,0,-1):
            nums[0],nums[end] = nums[end],nums[0]
            max_down_heapify(nums,0,end)
        return nums
```

**https://leetcode.com/problems/kth-largest-element-in-an-array/description/**

```python
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        def min_down_heapify(nums, parent_index, n):
            while True:
                smallest = parent_index
                left_indx = 2 * parent_index + 1
                right_indx = 2 * parent_index + 2
                if left_indx < n and nums[left_indx] < nums[smallest]:
                    smallest = left_indx
                if right_indx < n and nums[right_indx] < nums[smallest]:
                    smallest = right_indx
                if smallest == parent_index:
                    break
                nums[smallest], nums[parent_index] = (
                    nums[parent_index],
                    nums[smallest]
                )
                parent_index = smallest

        # Build min heap using first k elements
        for i in range((k // 2) - 1, -1, -1):
            min_down_heapify(nums, i, k)

        # Process remaining elements
        for i in range(k, len(nums)):
            if nums[i] > nums[0]:
                nums[i], nums[0] = nums[0], nums[i]
                min_down_heapify(nums, 0, k)

        # Root = kth largest
        return nums[0]
```

**https://www.geeksforgeeks.org/problems/minimum-cost-of-ropes-1587115620/1**

```python
class Solution:
    def minCost(self, arr):
        # code here
        def min_down_heapify(parent_index):
            n = len(arr)
            while True:
                smallest = parent_index
                left = 2 * parent_index + 1
                right = left + 1
                if left < n and arr[left] < arr[smallest]:
                    smallest = left
                if right < n and arr[right] < arr[smallest]:
                    smallest = right
                if smallest == parent_index:
                    break
                arr[parent_index], arr[smallest] = (
                    arr[smallest],
                    arr[parent_index]
                )
                parent_index = smallest

        # Build min heap: O(n)
        for i in range(len(arr) // 2 - 1, -1, -1):
            min_down_heapify(i)

        def remove_minimum():
            minimum = arr[0]
            last = arr.pop()
            if arr:
                arr[0] = last
                min_down_heapify(0)
            return minimum

        def insert_value(value):  # up heapify
            arr.append(value)
            child = len(arr) - 1
            while child > 0:
                parent = (child - 1) // 2
                if arr[child] >= arr[parent]:
                    break
                arr[child], arr[parent] = arr[parent], arr[child]
                child = parent

        total = 0

        while len(arr) > 1:
            first = remove_minimum()
            second = remove_minimum()
            cost = first + second
            total += cost
            insert_value(cost)
        return total
```

```python
# same algo but by using heapq built in class
import heapq
class Solution:
    def minCost(self, arr):
        # code here
        heapq.heapify(arr)

        total = 0

        while len(arr) > 1:
            first = heapq.heappop(arr)
            second = heapq.heappop(arr)

            cost = first + second
            total += cost

            heapq.heappush(arr, cost)

        return total
```

**https://leetcode.com/problems/the-k-weakest-rows-in-a-matrix/description/**

```python
# array and priority queue

class Solution:
    def kWeakestRows(self, mat: List[List[int]], k: int) -> List[int]:
        arr = []
        for row in range(len(mat)):
            ones = 0
            for col in range(len(mat[0])):
                if(mat[row][col] == 1):
                    ones += 1
            arr.append([ones,row])
        

        
        def max_down_heapify(arr,parent_index,n):
            while True:
                largest = parent_index
                left_indx = 2 * parent_index + 1
                right_indx = 2 * parent_index + 2

                if left_indx < n and arr[left_indx] > arr[largest]:
                    largest = left_indx
                if right_indx < n and arr[right_indx] > arr[largest]:
                    largest = right_indx
                if(largest == parent_index):
                    break
                arr[largest], arr[parent_index] = arr[parent_index],arr[largest]
                parent_index = largest
        
        for i in range((k//2)-1,-1,-1):
            max_down_heapify(arr,i,k)
        
        for i in range(k,len(arr)):
            if arr[i] < arr[0]:
                arr[i], arr[0] = arr[0], arr[i]
                max_down_heapify(arr, 0, k)

        
        ans = []
        for i in range(k-1,0,-1):
            arr[0],arr[i] = arr[i],arr[0]
            max_down_heapify(arr,0,i)
        
        for i in range(k):
            ans.append(arr[i][1])
        return ans
```

```python
# binary  and priority queue
import heapq

class Solution:
    def kWeakestRows(self, mat: List[List[int]], k: int) -> List[int]:
        def count_soldiers(row):
            left = 0
            right = len(row) - 1
            while left <= right:
                mid = (left + right) // 2
                if row[mid] == 1:
                    left = mid + 1
                else:
                    right = mid - 1
            return left
        heap = []
        for index in range(len(mat)):
            soldiers = count_soldiers(mat[index])
            heapq.heappush(heap, (soldiers, index))
            
        ans = []

        for _ in range(k):
            soldiers, index = heapq.heappop(heap)
            ans.append(index)
        return ans
```
