Heap sort **https://leetcode.com/problems/sort-an-array/description/**

```python
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        def max_down_heapify(nums,parent_index,n):
            while True:
                largest = parent_index
                left_indx = 2*parent_index +1
                right_indx = 2*parent_index +2

                if(left_indx < n and nums[left_indx] > nums[largest]):
                    largest = left_indx
                if(right_indx < n and nums[right_indx] > nums[largest]):
                    largest = right_indx
                
                if(parent_index == largest):
                    break
                nums[largest],nums[parent_index] = nums[parent_index],nums[largest]
                parent_index = largest
        n = len(nums)
        for i in range((n//2)-1,-1,-1):
            max_down_heapify(nums,i,n)
        
        for end in range(n-1,0,-1):
            nums[0],nums[end] = nums[end],nums[0]
            max_down_heapify(nums,0,end)
        return nums
```

```python
def max_down_heapify(arr,parent_index,n):
    while True:
        largest = parent_index
        left_indx = 2 * parent_index + 1
        right_indx = 2 * parent_index + 2

        if left_indx < n and arr[left_indx] > arr[largest]:
            largest = left_indx
        if right_indx < n and arr[right_indx] > arr[largest]:
            largest = right_indx
        if(largest == parent_index):
            break
        arr[largest], arr[parent_index] = arr[parent_index],arr[largest]
        parent_index = largest
def find_k_smallest(arr,k):
    for i in range((k//2)-1,-1,-1):
        max_down_heapify(arr,i,k)
    for i in range(k,len(arr)):
        if(arr[i] < arr[0]):
            arr[i],arr[0] = arr[0],arr[i]
            max_down_heapify(arr,0,k)
    return arr[:k]

# arr = [1, 23, 12, 9, 30, 2, 50]
# k = 3
# Output =  [1, 2, 9]

arr = [11, 5, 12, 9, 44, 17, 2]
k = 2
Output = [2, 5]

find_k_smallest(arr,k)
```
