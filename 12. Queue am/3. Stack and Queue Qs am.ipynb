# Q1. Three in One
## Describe how you could use a single python list to implement three stacks.

```python
class ThreeStacks:
    def __init__(self,stack_size):
        self.stack_size = stack_size
        self.stack = [None]*(self.stack_size * 3)
        self.top = [-1,-1,-1]
    def top_index(self,stacknum):
        return stacknum*self.stack_size + (self.top[stacknum])

    def push(self,stacknum,value):
        if self.is_full(stacknum):
            return "Stack is full"
        self.top[stacknum] += 1
        index = self.top_index(stacknum)
        self.stack[index] = value

    def pop(self,stacknum):
        if self.is_empty(stacknum):
            return "The stack is empty"
        index = self.top_index(stacknum)
        value = self.stack[index]
        self.stack[index] = None
        self.top[stacknum] -= 1
        return value
        

    def is_full(self,stacknum):
        if self.top[stacknum] + 1 == self.stack_size:
            return True
        else:
            return False

    def is_empty(self,stacknum):
        if self.top[stacknum] == -1:
            return True
        else:
            return False
    def peek(self, stacknum):
        if self.is_empty(stacknum):
            return "Empty"
        return self.stack[self.top_index(stacknum)]

multistack = ThreeStacks(3)
multistack.push(1,34)
multistack.push(1,30)
multistack.push(0,10)
multistack.pop(2)
            
        
```

# Q2. Stack Min
## How would you design a stack which, in addition to push and pop, has a function min which returns the minimum element? Push, pop and min should all operate in O(1)

```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, value):
        self.stack.append(value)

        if not self.min_stack or value <= self.min_stack[-1]:
            self.min_stack.append(value)

    def pop(self):
        if not self.stack:
            return "Stack is empty"

        value = self.stack.pop()

        if value == self.min_stack[-1]:
            self.min_stack.pop()

        return value

    def min(self):
        if not self.min_stack:
            return "Stack is empty"

        return self.min_stack[-1]

    def peek(self):
        if not self.stack:
            return "Stack is empty"

        return self.stack[-1]


ms = MinStack()

ms.push(5)
ms.push(3)
ms.push(7)
ms.push(2)

print(ms.min())   # 2
print(ms.pop())   # 2
print(ms.min())   # 3
```

# implementation using linked list

```python
class Node:
    def __init__(self, value, current_min):
        self.value = value
        self.current_min = current_min
        self.next = None


class MinStack:
    def __init__(self):
        self.head = None

    def push(self, value):
        if self.head is None:
            new_node = Node(value, value)
        else:
            new_node = Node(value, min(value, self.head.current_min))
            new_node.next = self.head
    
        self.head = new_node

    def pop(self):
        if self.head is None:
            return "Stack is empty"

        value = self.head.value
        self.head = self.head.next
        return value

    def peek(self):
        if self.head is None:
            return "Stack is empty"

        return self.head.value

    def min(self):
        if self.head is None:
            return "Stack is empty"

        return self.head.current_min

    def is_empty(self):
        return self.head is None

    def display(self):
        curr = self.head

        while curr:
            print(f"[{curr.value}|min={curr.current_min}]", end=" -> ")
            curr = curr.next

        print("None")

ms = MinStack()

ms.push(5)
ms.push(3)
ms.push(7)
ms.push(2)

ms.display()
print("Current Min:", ms.min())
```
