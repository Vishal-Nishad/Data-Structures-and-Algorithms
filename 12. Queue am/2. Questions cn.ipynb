````python
# implementing queue using two stacks
class QueueUsingTwoStacks:
    def __init__(self):
        self.__s1 = []
        self.__s2 = []
    def enqueue(self,data):  # O(n)
        while len(self.__s1) != 0:
            self.__s2.append(self.__s1.pop())
        self.__s1.append(data)
        while(len(self.__s2) != 0):
            self.__s1.append(self.__s2.pop())
        return

    def dequeue(self): # O(1)
        if (len(self.__s1) == 0):
            return "Queue is already empty"
        return self.__s1.pop()
        
    def front(self):
        if (len(self.__s1) == 0):
            return "Queue is already empty"
        return self.__s1[-1]

    def size(self):
            return len(self.__s1)
    def isEmpty(self):
        if (len(self.__s1) == 0):
            return True
        return False

q = QueueUsingTwoStacks()
q.enqueue(3)
q.enqueue(5)
q.enqueue(7)

while(not q.isEmpty()):
    print(q.front())
    q.dequeue()

q.size()
````

````python
from collections import deque

def reverse_first_k(dq, k):
    in_order = []
    to_reverse = []

    # remove first k elements
    for _ in range(k):
        to_reverse.append(dq.popleft())

    # remove remaining elements
    while dq:
        in_order.append(dq.popleft())

    # add reversed first k
    for i in range(len(to_reverse) - 1, -1, -1):
        dq.append(to_reverse[i])

    # add remaining in same order
    for i in range(len(in_order)):
        dq.append(in_order[i])

    return dq


cus_q = deque()

cus_q.append(1)
cus_q.append(2)
cus_q.append(3)
cus_q.append(4)
cus_q.append(5)

print(reverse_first_k(cus_q, 3))
print(list(cus_q))
````

# 1. What is `deque`?

`deque` stands for:

```text
double-ended queue
```

Meaning:

* insert/remove from left side
* insert/remove from right side

efficiently.

It comes from Python’s standard library:

```python id="u1"
from collections import deque
```

So `deque` is a **class** provided by the `collections` module.

You create objects from it:

```python id="u2"
dq = deque()
```

That is similar to:

```python id="u3"
lst = list()
d = dict()
s = set()
```

---

# Example

```python id="u4"
from collections import deque

dq = deque([1,2,3])

dq.append(4)        # right side
dq.appendleft(0)    # left side

print(dq)
```

Result:

```text
deque([0,1,2,3,4])
```

---

# Why Use `deque` Instead of List for Queue?

If you do queue operations on list:

```python id="u5"
arr.pop(0)
```

Removing from front is slow: `O(n)`

But deque:

```python id="u6"
dq.popleft()
```

is fast: `O(1)`

So `deque` is ideal for:

* queue
* stack
* sliding window
* BFS

# 2. What is `import queue`?

import queue

A module is just a Python file/library containing code.

Inside that module, Python provides classes like:

* `Queue`
* `LifoQueue`

# Difference: `queue.Queue` vs `collections.deque`

## `queue.Queue`

Designed for:

* multithreading
* thread-safe producer/consumer systems

Methods:

```python id="u11"
put()
get()
empty()
qsize()
```

It uses locks internally.

## `deque`

Designed for fast general-purpose data structure operations.

Methods:

```python id="u12"
append()
appendleft()
pop()
popleft()
```

Usually preferred in DSA.

# 4. If `deque` is a Class, Why Not `Deque`?

Very good observation.

Normally Python classes use **CapWords / PascalCase**:

```python id="u13"
MyClass
Queue
Thread
Person
```

So why lowercase `deque`?

---

# Short Answer

Because Python built-in/container types are often lowercase.

Examples:

```python id="u14"
list
dict
set
tuple
str
int
float
deque
defaultdict
namedtuple
```

These are classes/types too, but named like data structures or factory types.

---

# Why Python Does This

PEP 8 says:

* Regular classes → `CapWords`
* Functions/variables → `lowercase_with_underscores`

But some standard library types intentionally use lowercase because they behave like built-in containers.

`deque` feels like:

```python id="u15"
list()
dict()
set()
```

not like a user-defined class such as `Employee()`.

So lowercase naming was chosen.
