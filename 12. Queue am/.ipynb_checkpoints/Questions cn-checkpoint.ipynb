```python
# implementing queue using two stacks
class QueueUsingTwoStacks:
    def __init__(self):
        self.__s1 = []
        self.__s2 = []
    def enqueue(self,data):  # O(n)
        while len(self.__s1) != 0:
            self.__s2.append(self.__s1.pop())
        self.__s1.append(data)
        while(len(self.__s2) != 0):
            self.__s1.append(self.__s2.pop())
        return

    def dequeue(self): # O(1)
        if (len(self.__s1) == 0):
            return "Queue is already empty"
        return self.__s1.pop()
        
    def front(self):
        if (len(self.__s1) == 0):
            return "Queue is already empty"
        return self.__s1[-1]

    def size(self):
            return len(self.__s1)
    def isEmpty(self):
        if (len(self.__s1) == 0):
            return True
        return False

q = QueueUsingTwoStacks()
q.enqueue(3)
q.enqueue(5)
q.enqueue(7)

while(not q.isEmpty()):
    print(q.front())
    q.dequeue()

q.size()
```

# Reverse the first k elements in the Queue

```python
import queue as q
def reverse_first_k(q,k):
    in_order = []
    to_reverse = []
    for i in range(k):
        to_reverse.append(q.get())
    while not q.empty():
        in_order.append(q.get())

    for i in range(len(to_reverse)-1,-1,-1):
        q.put(to_reverse[i])
    for i in range(len(in_order)):
        q.put(in_order[i])
    return q

cus_q = q.Queue()
cus_q.put(1)
cus_q.put(2)
cus_q.put(3)
cus_q.put(4)
cus_q.put(5)
print(reverse_first_k(cus_q,3))
print(list(cus_q.queue))
```

```python
from collections import deque

def reverse_first_k(dq, k):
    in_order = []
    to_reverse = []

    # remove first k elements
    for _ in range(k):
        to_reverse.append(dq.popleft())

    # remove remaining elements
    while dq:
        in_order.append(dq.popleft())

    # add reversed first k
    for i in range(len(to_reverse) - 1, -1, -1):
        dq.append(to_reverse[i])

    # add remaining in same order
    for i in range(len(in_order)):
        dq.append(in_order[i])

    return dq


cus_q = deque()

cus_q.append(1)
cus_q.append(2)
cus_q.append(3)
cus_q.append(4)
cus_q.append(5)

print(reverse_first_k(cus_q, 3))
print(list(cus_q))
```
