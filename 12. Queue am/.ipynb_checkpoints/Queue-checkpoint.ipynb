```python
# Queue is a data structure that stores items in a First-in/First-Out manner.
# - A new addition to the queue happens at the end of the queue
# - First element in the queue will be served First(Means utitlize first coming data first, while others wait for their turn)
# - FIFO METHOD - first in first out

# Applications of Queue
# - Point sale system of a restaurant
# - Printer Queue
# - Call Center phone systems
```

## Queue Operations
- Create Queue
- Enqueue(insert element at last position)
- Dequeue(delete element from first position)
- Peek(return element at first position)
- isEmpty
- isFull(in queue with capacity, if queue is full then to check that we have isFull)-> (in queue without capacity we don't have this method)
- deleteQueue(delete the entire queue)

# Implementation (Ways of Creating Queues)
- 1. Python List/Arrays
     - Queue without capacity
     - queue with capacity(Circular Queue)
- 2. Linked List

```python
# Queue without capacity
class Queue:
    def __init__(self):
        self.items = []

    def __str__(self):
        values = [str(x) for x in self.items]
        return " ".join(values)
    def isEmpty(self):
        if self.items == []:
            return True
        else:
            return False
    def enque(self,value):
        self.items.append(value) # time complexity O(n), adding element at the last take O(n)
        return "The element is inserted at the end of Queue"

    def deque(self):
        if self.isEmpty():
            return "There is no element in queue"
        else:
            return self.items.pop(0) # time complexity O(n)
    def peek(self):
        if self.isEmpty():
            return "There is no element in queue"
        else:
            return self.items[0] # time complexity O(1)
    def delete(self):
        self.items = None

custom_queue = Queue()
custom_queue.enque(3)
custom_queue.enque(5)
custom_queue.enque(0)
print(custom_queue.isEmpty())
print(custom_queue.items)
print(custom_queue)
custom_queue.deque()
print(custom_queue)
custom_queue.peek()
```

```python
#Circular queue, Queue with capacity

class Queue:
    def __init__(self,maxSize):
        self.items = maxSize * [None]
        self.maxSize = maxSize
        self.start = -1
        self.top = -1

    def __str__(self):
        values = [str(x) for x in self.items]
        return " ".join(values)
    def isFull(self):
        if self.top +1 == self.start:
            return True
        elif self.start ==0 and self.top +1 == self.maxSize:
            return True
        else:
            return False
    def isEmpty(self):
        if self.top == -1:
            return True
        else:
            return False
    def enque(self,value):
        if self.isFull():
            return "The queue is full"
        else:
            if self.top +1==self.maxSize:
                self.top = 0
            else:
                self.top += 1
                if self.start == -1:
                    self.start = 0
            self.items[self.top]= value
            return "the element is inserted at the end of queue"

    def deque(self):
        if self.isEmpty():
            return "There is no element in queue"
        else:
            firstelement = self.items[self.start]
            start = self.start
            if self.start == self.top:
                self.start = -1
                self.top = -1
            elif self.start + 1 == self.maxSize:
                self.start = 0
            else:
                self.start += 1
            self.items[start] = None
            return firstelement
    def peek(self):
        if self.isEmpty():
            return "There is no element in queue"
        else:
            return self.items[self.start] # time complexity O(1)
    def delete(self):
        self.items = self.maxSize *[None]
        self.top = -1
        self.start = -1

customqueue = Queue(5)
customqueue.enque(3)
customqueue.enque(4)
customqueue.enque(8)
customqueue.enque(9)
print(customqueue)
print(customqueue.deque())
print(customqueue)
print(customqueue.peek())
```
