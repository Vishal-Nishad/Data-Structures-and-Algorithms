**Creation of singly Linked List**

```python
class Node:
    def __init__(self, value=None):
        self.value = value
        self.next = None

class SLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

singlyLinkedList = SLinkedList()
node1 = Node(1)
node2 = Node(2)
node3 = Node(1)
node4 = Node(2)

singlyLinkedList.head = node1
node1.next = node2
node2.next = node3
node3.next = node4
singlyLinkedList.tail = node4

cur = singlyLinkedList.head
while cur:
    print(cur.value)
    cur = cur.next
```
