```python
a = {1:2, 3:4, "list": [1,23], "dict":{1:2}}
```

```python
a
```

```python
# if the key is not present then it will raise an error
a["li"]
```

```python
a["list"]
```

```python
# get function does not give error if the key does not present in the dict, infact it take parameter which we want if key 
# does not present in the dict
print(a.get("li"))
```

```python
# if the key is present then here it will return its associated value, if key is not present then it will return the default value
# which we will pass in get function
a.get("li",0)
```
