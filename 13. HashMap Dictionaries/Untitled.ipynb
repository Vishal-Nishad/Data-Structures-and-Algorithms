```python
st = "this is a word string having many many word"
for i in st:
    print(i)
```

```python
li = st.split(" ")
li
```

```python
freq = {}
for i in range(len(li)):
    freq[li[i]] = freq.get(li[i],0) + 1

freq
```

```python
for key in freq:
    if freq[key] > 1:
        print(key)
```

```python
def print_k_freq_words(s,k):
    li_words = s.split(" ")
    freq = {}
    for i in range(len(li_words)):
        freq[li_words[i]] = freq.get(li_words[i],0) + 1
    for key in freq:
        if freq[key] == k:
            print(key)
print_k_freq_words(st,2)
```

# maximum frequency

```python
a = [4,4,2,3,3,1,2,5,6,8,8]
def max_freq(ar):
    freq = {}

    for i in range(len(ar)):
        freq[ar[i]] = freq.get(ar[i],0) + 1
    print(freq)
    max = 0
    ans = 0 
    for key, value in freq.items():
        if value > max:
            max = value
            ans = key
    return ans
max_freq(a)
```
