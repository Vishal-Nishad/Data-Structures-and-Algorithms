# majority element

```python
# brutte
def majority_element_brute(ar):
    ans = set()
    n = len(ar)
    for i in range(len(ar)):
        count = 1
        for j in range(i+1,len(ar)):
            if ar[i] == ar[j]:
                count += 1
                if count > n/3:
                    if ar[i] not in ans:
                        ans.add(ar[i])
    return ans
a = [1,2,3,42,2,2,3,4,5,4,4,1,1,1,1,1]

majority_element_brute(a)
```

```python
#optimal
def majority_element_optimal(ar):
    freq = {}
    n = len(ar)
    for i in range(len(ar)):
        freq[ar[i]] = freq.get(ar[i],0) + 1
    for j in range(len(ar)):
        if freq[ar[j]] > n/3:
            return ar[j]

a = [1,2,3,42,2,2,3,4,5,4,4,1,1,1,1,1]
majority_element_optimal(a)
```

# valid anagram

```python
def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            False
        freq={}
        for i in range(len(s)):
            freq[s[i]] = freq.get(s[i],0) + 1
        
        for j in range(len(t)):
            freq[t[j]] = freq.get(t[j],0) -1
        
        for key,value in freq.items():
            if value != 0:
                return False
        return True
```

```python
# one pass
def isanagram_onepass(s,t):
    if len(s) != len(t):
        False
    s_count = {}
    t_count = {}
    for i in range(len(s)):
        s_count[s[i]] = s_count.get(s[i],0) + 1
        t_count[t[i]] = t_count.get(t[i],0) + 1

    return t_count == s_count

s = "aacc"
t = "caac"
isanagram_onepass(s,t)
```
