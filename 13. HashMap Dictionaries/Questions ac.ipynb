# majority element

```python
def majority_element_brute(ar,n):
    ans = set()
    for i in range(len(ar)):
        count = 1
        for j in range(i+1,len(ar)):
            if ar[i] == ar[j]:
                count += 1
                if count > n//3:
                    if ar[i] not in ans:
                        ans.add(ar[i])
    return ans
a = [1,2,3,42,2,2,3,4,5,4,4,1,1]

majority_element_brute(a,3)
```
