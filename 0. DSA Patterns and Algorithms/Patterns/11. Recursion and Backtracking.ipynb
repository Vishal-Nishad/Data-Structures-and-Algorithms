````python
# print sum of n natural numbers by recursion
def rec_sum(n):
    if n == 1:
        return 1
    ans = rec_sum(n-1)
    return ans + n
rec_sum(5)
````

Your current way (mentally “following calls”) will break on harder recursion. You need a **structured tracing method**, not intuition.

Let’s build that skill properly.

---

# 1. First understand what recursion *really is*

Think of recursion as:

> “I don’t solve everything. I solve **one step**, and delegate the rest.”

For your function:

```python
def rec_sum(n):
    if n == 1:
        return 1
    ans = rec_sum(n-1)
    return ans + n
```

Each call says:

> “Give me sum till (n-1), I’ll just add `n`.”

---

# 2. The SMART way to draw recursion (this is the key technique)

Instead of messy trees, always use this **2-column method**:

| Call | Returns |
| ---- | ------- |

---

## Step-by-step for `rec_sum(5)`

### Step 1: Write all calls first (top → bottom)

```
rec_sum(5)
rec_sum(4)
rec_sum(3)
rec_sum(2)
rec_sum(1)
```

👉 Just keep expanding until base case.

---

### Step 2: Now resolve from bottom (VERY IMPORTANT)

Start filling returns **backwards**:

| Call       | Returns     |
| ---------- | ----------- |
| rec_sum(1) | 1           |
| rec_sum(2) | 1 + 2 = 3   |
| rec_sum(3) | 3 + 3 = 6   |
| rec_sum(4) | 6 + 4 = 10  |
| rec_sum(5) | 10 + 5 = 15 |

---

# 3. Visualizing call stack (how interviewer expects you to think)

Think in **stack behavior (LIFO)**:

```
rec_sum(5)
  → rec_sum(4)
      → rec_sum(3)
          → rec_sum(2)
              → rec_sum(1)  ✅ base case

Now returning:

rec_sum(1) = 1
rec_sum(2) = 1 + 2 = 3
rec_sum(3) = 3 + 3 = 6
rec_sum(4) = 6 + 4 = 10
rec_sum(5) = 10 + 5 = 15
```

---

# 4. The REAL trick for complex problems (this is what you're asking)

For hard recursion problems, follow this **4-step system**:

---

## ✅ Step 1: Define “what does function return”

Ask:

> “What exactly does `rec_sum(n)` return?”

👉 Answer:

> Sum of numbers from `1 → n`

This avoids confusion later.

---

## ✅ Step 2: Assume smaller problem works

Never expand everything.

Instead say:

```
Assume rec_sum(4) = 10
```

Then:

```
rec_sum(5) = 10 + 5 = 15
```

👉 This is called **faith in recursion**

---

## ✅ Step 3: Only expand when stuck

If confused:

* Expand only 2–3 levels
* Not the entire tree

---

## ✅ Step 4: Track variables per call

For complex problems (like trees, DP):

Write like this:

```
rec(n=5)
  ans = rec(4) → returns 10
  return 10 + 5
```

---

# 5. Why your current approach may fail later

In problems like:

* Backtracking
* Trees
* DP recursion

If you:
❌ draw full tree → too big
❌ don’t track return values → confusion
❌ don’t define function meaning → lost

---

# 6. Advanced mindset (this is interview-level thinking)

Stop thinking:

> “What is happening step by step?”

Start thinking:

> “What contract does this function guarantee?”

For example:

```
rehen:

```
rec_sum(n) = rec_sum(n-1) + n
```

No confusion. No large diagrams needed.

# find maximum number in array recursively

````python
def max_rec(arr,n):
    if n == 1:
        return arr[0]
    last = arr[n-1]
    max_no = max_rec(arr,n-1)
    return max(last,max_no)

a = [3,4,1]
max_rec(a,3)
````

# Check Palindrome recursively

````python
# time complexity = O(n)
# space = O(n) i.e because of recursive call stack,
def check_palindrome(str,st,end):
    if(st>=end):
        return True
    if(str[st] != str[end]):
        return False
    return check_palindrome(str,st+1,end -1)

st = "namal"
check_palindrome(st,0,len(st)-1)
````

# Q. recursively print sum of digits


````python
# iterative
def iter_digit_sum(n):
    ans = 0
    while n>0:
        rem = n%10
        ans += rem
        n = n//10
    return ans

iter_digit_sum(123)
````

````python
# recursive
def rec_digit_sum(n,ans):
    if(n==0):
        return ans
    ans += n%10
    return rec_digit_sum(n//10,ans)

rec_digit_sum(123,0)
````

# Q. remove char from given string

````python
def rec_remove_ch(str,ans_str,ch,ind):
    if(ind == len(str)):
        return ans_str
    if(str[ind] != ch):
        ans_str += str[ind]
    return rec_remove_ch(str,ans_str,ch,ind+1)

st = "apple"
rec_remove_ch(st,"","e",0)
````

🔴 1. **https://leetcode.com/problems/combination-sum/description/**

````python
# time complexity = O(N^target)
'''Branching Factor -> Maximum recursive calls created by ONE call
Depth -> Maximum recursion chain length
Total Work -> branching_factor^depth

because every level multiplies possibilities.'''

# Space Complexity:
'''O(target) auxiliary space due to recursion stack
and current path storage.

If output storage is included,
additional space is required for storing combinations.'''

def combination_sum(candidates,target):
    result = []
    def backtrack(start,res_sum,temp_arr):
        # base case
        if(res_sum == target):
            result.append(temp_arr[:])
            return

        # invalid case
        if(res_sum > target):
            return

        # recurse, all choices
        for i in range(start,len(candidates)):
            num = candidates[i]
            # work, do
            temp_arr.append(num)
            backtrack(i,res_sum+num,temp_arr)

            # undo
            temp_arr.pop()
    backtrack(0,0,[])
    return result

candidates = [2,3,6,7]
target = 7
combination_sum(candidates,target)
````

# 1. `candidates.sort()`

```python
candidates.sort()
```

Purpose:

```text id="4ylmzz"
Bring duplicate numbers together.
```

Why needed?

Because duplicate skipping condition works only when equal elements are adjacent.

Example:

Without sorting:

```python id="gzaybq"
[1,2,1]
```

duplicates are separated.

After sorting:

```python id="0b7k4f"
[1,1,2]
```

now duplicates are easy to detect.

---

# 2. `backtrack(i+1, ...)`

```python
backtrack(i+1, ...)
```

Purpose:

```text id="9qj0jg"
Move forward to next index.
```

Why needed?

Because in Combination Sum II:

```text id="hkwy1u"
Each element can be used only ONCE.
```

So after choosing current element at index `i`,
next recursion starts from:

```python id="dwng1d"
i+1
```

which prevents reusing same index again.

---

# 3. Duplicate Skip Condition

```python
if i > start and candidates[i] == candidates[i-1]:
    continue
```

Purpose:

```text id="l5v1v8"
Skip duplicate choices at SAME recursion level.
```

Why needed?

To avoid duplicate combinations like:

```python id="0dgz0s"
[1,7]
[1,7]
```

coming from different duplicate `1`s.

---

# Meaning of `i > start`

```text id="1u85m4"
Skip duplicates only among siblings,
NOT parent-child recursion.
```

This allows:

```python id="bh6ik2"
[1,1,6]
```

but prevents duplicate branches starting from same level.

---

# Ultra-Short Interview Version

---

## `sort()`

```text id="q4os40"
Sort to place duplicates together.
```

---

## `i+1`

```text id="wc6if7"
Move forward because each element can be used once.
```

---

## `i > start and nums[i] == nums[i-1]`

```text id="nhzv0f"
Skip duplicate values at same recursion level
to avoid duplicate combinations.

````python
def exist(board, word):
        def backtrack(col,row,index):
            # base case
            if(index == len(word)):
                return True
                
            #invalid case
            if(row<0 or col < 0 or row>=len(board) or col>=len(board[0]) or
            board[row][col] != word[index]):
                return False

            # mark visited/ do
            temp = board[row][col]
            board[row][col] = "#"

            # down
            if(backtrack(col,row+1,index+1)):
                return True

            # up
            if(backtrack(col,row-1,index+1)):
                return True
            
            # right
            if(backtrack(col+1,row,index+1)):
                return True

            # left
            if(backtrack(col-1,row,index+1)):
                return True

            # undo
            board[row][col] = temp
            return False
        
        for r in range(len(board)):
            for c in range(len(board[0])):
                if(backtrack(c,r,0)):
                    return True
        return False

board = [["A","B","C","E"],["S","F","C","S"],["A","D","E","E"]]
word = "ABCCED"
exist(board,word)
````

5. **https://leetcode.com/problems/sudoku-solver/description/**

````python
def solveSudoku(board):
        """
        Do not return anything, modify board in-place instead.
        """
        def is_safe(row,col,digit):
            #vertically
            r = 0
            while r<=8:
                if(board[r][col] == digit):
                    return False
                r += 1
            
            # horizontally
            c = 0
            while(c<=8):
                if(board[row][c]== digit):
                    return False
                c += 1
            
            # grid check
            r = (row//3)*3
            c = (col//3)*3
            for i in range(r,r+3):
                for j in range(c,c+3):
                    if(board[i][j]==digit):
                        return False
            return True

        def backtrack(row,col):
            # base case
            if(row == 9):
                return True
            
            # call for next cell
            next_row = row
            next_col = col + 1
            if(col+1 == 9):
                next_col = 0
                next_row = row + 1
            
            # make call if current cell is not empty
            if(board[row][col] != "."):
                return backtrack(next_row,next_col)
            
            # explore all digit choices
            for i in range(1,10):
                if (is_safe(row,col,str(i))):
                    board[row][col] = str(i)

                    if(backtrack(next_row,next_col)):
                        return True
                    board[row][col] = "."
            return False
        backtrack(0,0)

board = [["5","3",".",".","7",".",".",".","."],["6",".",".","1","9","5",".",".","."],[".","9","8",".",".",".",".","6","."],
         ["8",".",".",".","6",".",".",".","3"],["4",".",".","8",".","3",".",".","1"],["7",".",".",".","2",".",".",".","6"],
         [".","6",".",".",".",".","2","8","."],[".",".",".","4","1","9",".",".","5"],[".",".",".",".","8",".",".","7","9"]]
solveSudoku(board)
board
````

6. **https://algo.monster/liteproblems/254**   
   **https://leetcode.com/problems/factor-combinations/**

````python
def factor_combinations(n):
    result = []
    def backtrack(start,temp_arr,remaining_num):
        # base case
        if(remaining_num == 1):
            # atleast two factor needed
            if(len(temp_arr) >=2):
                result.append(temp_arr[:])
            return

        for factor in range(start,remaining_num+1):
            if(remaining_num % factor == 0):
                temp_arr.append(factor)
                backtrack(factor,temp_arr,remaining_num//factor)
                temp_arr.pop()
    backtrack(2,[],n)
    return result

factor_combinations(12)
````

````python
# optimized approach
class Solution:
    def getFactors(self, n):
        """
        Find all possible combinations of factors that multiply to n.
        Factors must be greater than 1 and less than n.

        Args:
            n: The target number to factorize

        Returns:
            List of lists containing all possible factor combinations
        """

        def dfs(target, factor):
            """
            Depth-first search to find factor combinations.

            Args:
                target: Current number to factorize
                start_factor: Minimum factor to consider (avoids duplicates)
            """
            # If we have factors in current_path, add this combination
            # (current factors + remaining target as last factor)
            if current_path:
                result.append(current_path + [target])

            # Try all possible factors from start_factor up to sqrt(target)
            while factor * factor <= target:
                # If factor divides target evenly
                if target % factor == 0:
                    # Add factor to current path
                    current_path.append(factor)
                    # Recursively find factors for the quotient
                    # Use 'factor' as start to avoid duplicate combinations
                    dfs(target // factor, factor)
                    # Backtrack - remove the factor we just tried
                    current_path.pop()
                factor += 1

        # Initialize tracking variables
        current_path = []  # Stores current combination of factors being explored
        result = []        # Stores all valid factor combinations

        # Start DFS with the original number and minimum factor of 2
        dfs(n, 2)

        return result

sol = Solution()
sol.getFactors(12)
````

7. **https://leetcode.com/problems/split-a-string-into-the-max-number-of-unique-substrings/**

````python
def maxUniqueSplit(s):
        best = 0
        used = set()
        def backtrack(st_index):
            nonlocal best
            # base case
            if(st_index == len(s)):
                best = max(best,len(used))
                return
            
            for end in range(st_index,len(s)):
                sub_string = s[st_index:end+1]

                if sub_string not in used:
                    used.add(sub_string)
                    backtrack(end+1)
                    used.remove(sub_string)
            
        backtrack(0)
        return best
s = "ababccc"
maxUniqueSplit(s)
````

````python
# another way just for recursion learning purpose
def maxUniqueSplit(s):
        best = 0
        used = set()
        def solve(start):
            nonlocal best
            if start == len(s):
                return 0
            best = 0
        
            for end in range(start, len(s)):
                sub = s[start:end+1]
                if sub not in used:
                    used.add(sub)
                    best = max(best, 1 + solve(end+1))
                    used.remove(sub)
            return best
            
        solve(0)
        return best


s = "ababccc"
maxUniqueSplit(s)
````

8. **https://leetcode.com/problems/generate-parentheses/description/**

````python
def generateParenthesis(n):
        result = []
        def backtrack(temp_str,open_c,close_c):
            if(len(temp_str)==2*n):
                result.append(temp_str)
                return
            
            # open bracket
            if(open_c <n):
                backtrack(temp_str + "(",open_c +1,close_c)
            
            # close bracket
            if(close_c < open_c):
                backtrack(temp_str + ")",open_c,close_c + 1)
        backtrack("",0,0)
        return result

generateParenthesis(3)
````

9. **https://leetcode.com/problems/letter-combinations-of-a-phone-number/description/**

````python
def keypad_combination(digits):
    phone = {
        "2": "abc",
        "3": "def",
        "4": "ghi",
        "5": "jkl",
        "6": "mno",
        "7": "pqrs",
        "8": "tuv",
        "9": "wxyz"
    }
    result = []
    def backtrack_dfs(ind,curr_str):
        #base case
        if(len(curr_str) == len(digits)):
            result.append(curr_str)
            return
        key_num = digits[ind]
        letters = phone[key_num]
        for i in range(len(letters)):
            backtrack_dfs(ind+1,curr_str+letters[i])
    backtrack_dfs(0,"")
    return result

s = "23"
keypad_combination(s)
    
````

10. **https://leetcode.com/problems/permutations/description/**

````python
def permutations(nums):
    result = []
    def backtrack_dfs(nums,curr):
        # base case
        if(len(nums) == 0):
            result.append(curr)

        for i in range(len(nums)):
            n = nums[i]
            new_string = nums[:i] + nums[i+1:]
            backtrack_dfs(new_string,curr + [n])
    backtrack_dfs(nums,[])
    return result

a = [1,2,3]
permutations(a)
````

````python
# another using normal backtracking template and set
def perm(nums):
    result = []
    used = set()
    def backtracking_dfs(temp_arr,used):
        if(len(temp_arr) == len(nums)):
            result.append(temp_arr[:])
            return
        for i in range(len(nums)):
            if nums[i] in used:
                continue
            used.add(nums[i])
            temp_arr.append(nums[i])
            backtracking_dfs(temp_arr,used)
            used.remove(nums[i])
            temp_arr.pop()
    backtracking_dfs([],used)
    return result
a = [1,2,3,5]
perm(a)
````

11. **https://leetcode.com/problems/palindrome-partitioning/description/**

````python
def partition(s):
        result = []
        def is_pallindrome(st):
            i = 0
            j = len(st) -1
            while i<=j:
                if(st[i]!=st[j]):
                    return False
                i += 1
                j -= 1
            return True
        def backtrack_dfs(start_ind,temp_str):
            # base case
            if(start_ind == len(s)):
                result.append(temp_str[:])
                return
            
            for end in range(start_ind,len(s)):
                sub_str = s[start_ind:end + 1]
                if(is_pallindrome(sub_str)):
                    temp_str.append(sub_str)
                    backtrack_dfs(end+1,temp_str)
                    temp_str.pop()
        backtrack_dfs(0,[])
        return result

partition("aab")
````
