# Scenario: You need to find paths in a binary tree that sum up to a target value or enumerate all root-to-leaf paths.
# Clue: Explore problems where you traverse the tree while keeping track of the path and its sum, recursively exploring all possible paths.

1. **https://leetcode.com/problems/path-sum/description/**

````python
# time: O(n)
# space: recursive stack O(h)

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def hasPathSum(self, root: Optional[TreeNode], targetSum: int) -> bool:
        

        def dfs(root,cur_sum):
            if(root is None):
                return False
            if(cur_sum + root.val == targetSum and root.left is None and root.right is None):
                return True
            # if(cur_sum + root.val > targetSum):
            #     return False  # this pruning will only work if node value does not have negative numbers
            if(dfs(root.left,cur_sum + root.val)):
                return True
            if(dfs(root.right,cur_sum + root.val)):
                return True
            return False
        return dfs(root,0)
````

2. **https://leetcode.com/problems/path-sum-ii/**

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> List[List[int]]:
        result = []
        def dfs(root,path,cur_sum):
            if(root is None):
                return
            path.append(root.val)
            cur_sum += root.val
            if(cur_sum == targetSum and root.left is None and root.right is None):
                result.append(path[:])
                path.pop()
                # cur_sum -= root.val  # as int is immutable so it does not matter, as parent call stack have its own previous int value
                return
            dfs(root.left,path,cur_sum)
            dfs(root.right,path,cur_sum)
            path.pop()
            # cur_sum -= root.val      # as int is immutable so it does not matter, as parent call stack have its own previous int value
        dfs(root,[],0)
        return result
````

3. **https://leetcode.com/problems/path-sum-iii/**

````python
# time = O(n2)
# space = O(h)

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> int:
        count = 0
        def dfs_sum_check(root,cur_sum):
            nonlocal count
            if(root is None):
                return
            cur_sum += root.val
            if(cur_sum == targetSum):
                count += 1
                return # if chile node have 0 val then it will not count that part so that's why return not needed
                # root = [10,5,-3,3,2,null,11,0,1,null,1,null,0,0,null,0,0,0,null,1,null,0,null,0], targetSum = 8
            dfs_sum_check(root.left,cur_sum)
            dfs_sum_check(root.right,cur_sum)
        
        def dfs(root):
            if(root is None):
                return
            dfs_sum_check(root,0)
            dfs(root.left)
            dfs(root.right)
        dfs(root)
        return count
````

````python
# optimal solution, use prefix sum technique to calculate all subarray which are equal to target sum
# time = O(n)
# space = O(h)

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> int:
        ans_count = 0
        cur_prefix_sum = 0
        prefix_map = {0:1}
        def dfs(root):
            if(root is None):
                return
            nonlocal ans_count,cur_prefix_sum
            cur_prefix_sum += root.val
            if(cur_prefix_sum - targetSum) in prefix_map:
                ans_count += prefix_map[cur_prefix_sum - targetSum]
            prefix_map[cur_prefix_sum] = prefix_map.get(cur_prefix_sum,0) + 1
            dfs(root.left)
            dfs(root.right)
            prefix_map[cur_prefix_sum] = prefix_map.get(cur_prefix_sum,0) - 1
            cur_prefix_sum -= root.val
        dfs(root)
        return ans_count
````

4. **https://leetcode.com/problems/sum-root-to-leaf-numbers/description/**

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def sumNumbers(self, root: Optional[TreeNode]) -> int:
        ans = 0
        def dfs(root,num):
            nonlocal ans
            if(root is None):
                return
            num += str(root.val)
            if(root.left is None and root.right is None):
                ans += int(num)
                return
            dfs(root.left,num)
            dfs(root.right,num)
        dfs(root,"")
        return ans
````

🔴5.  **https://leetcode.com/problems/binary-tree-maximum-path-sum/description/**

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def maxPathSum(self, root: Optional[TreeNode]) -> int:
        ans = float("-inf")
        def dfs(root):
            nonlocal ans
            if(root is None):
                return 0
            left_gain = max(dfs(root.left),0)
            right_gain = max(dfs(root.right),0)
            ans = max(ans,root.val+left_gain+right_gain)
            return root.val+max(left_gain,right_gain)
        dfs(root)
        return ans
````

## Here's the one sentence I want you to remember (don't memorize the formulas)

Every node has **two jobs**:

1. **Update the answer** by imagining *"I am the highest point of the best path."*
2. **Return one side** to its parent because a parent cannot continue through both children.

If you remember those two jobs, you can re-derive the formulas even if you forget them.
