### Scenario: You need to visit each node in the tree exactly once, exploring all nodes in a specific order (e.g., inorder, preorder, postorder).

### Clue: Look for problems where you must visit each node systematically without missing any, often requiring recursion(for DFS) or iteration(for BFS).

1. **https://leetcode.com/problems/binary-tree-inorder-traversal/description/**

```python
# recursive traversing

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        def dfs(root):
            if(root is None):
                return
            dfs(root.left)
            ans.append(root.val)
            dfs(root.right)
        dfs(root)
        return ans
```

```python
# 🔴 iterative traversing,

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        stk = []
        curr = root
        while curr or stk:
            while curr:
                stk.append(curr)
                curr = curr.left
            curr = stk.pop()
            ans.append(curr.val)
            curr = curr.right
        return ans
```

2. **https://leetcode.com/problems/binary-tree-preorder-traversal/**

```python
# recursive 

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        def dfs(root):
            if(root is None):
                return None
            ans.append(root.val)
            dfs(root.left)
            dfs(root.right)
        dfs(root)
        return ans
```

```python
# iterative

# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        stk = []
        curr = root
        while curr or stk:
            while curr:
                stk.append(curr)
                ans.append(curr.val)
                curr = curr.left
            curr = stk.pop()
            curr = curr.right
        return ans
```

3. **https://leetcode.com/problems/binary-tree-postorder-traversal/description/**

```python
# iterative

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def postorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        if(root is None):
            return []
        ans = []
        stk1 = [root]
        stk2 = []
        while stk1:
            node = stk1.pop()
            stk2.append(node)
            if node.left:
                stk1.append(node.left)
            if node.right:
                stk1.append(node.right)
        while stk2:
            ans.append(stk2.pop().val)  
        return ans
```
