1. **https://leetcode.com/problems/lowest-common-ancestor-of-a-binary-tree/**

```python
# todo : compare the time complexity of both below given solution

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def findpath(self, root, target, path):
        if(root == None):
            return False
        path.append(root)
        if root == target:
            return True
        if(self.findpath(root.left,target,path)):
            return True
        if(self.findpath(root.right,target,path)):
            return True
        path.pop()
        return False
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        path1 = []
        path2 = []
        self.findpath(root,p,path1)
        self.findpath(root,q,path2)
        i = 0
        while(i<len(path1) and i <len(path2) and (path1[i] == path2[i])):
            i += 1
        return path1[i-1]
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        if(root == None):
            return None
        
        if (root == p or root == q):
            return root
        l = self.lowestCommonAncestor(root.left,p,q)
        r = self.lowestCommonAncestor(root.right,p,q)

        if(l and r):
            return root
        
        if l:
            return l
        else:
            return r
```

2. **https://leetcode.com/problems/search-in-a-binary-search-tree/description/**

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def searchBST(self, root: Optional[TreeNode], val: int) -> Optional[TreeNode]:
        if (root is None):
            return None
        if(root.val == val):
            return root
        elif(val > root.val):
            return self.searchBST(root.right,val)
        else:
            return self.searchBST(root.left,val)
```

3. **https://leetcode.com/problems/lowest-common-ancestor-of-a-binary-search-tree/description/**

```python
# recursive

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':

        if p.val < root.val and q.val < root.val:
            return self.lowestCommonAncestor(root.left, p, q)
        
        if p.val > root.val and q.val > root.val:
            return self.lowestCommonAncestor(root.right, p, q)
        
        return root
```

```python
# iterative

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        while root:
            if(p.val > root.val and q.val > root.val):
                root = root.right
            elif(p.val < root.val and q.val < root.val):
                root = root.left
            else:
                return root
```

4. **https://leetcode.com/problems/lowest-common-ancestor-of-deepest-leaves/description/**
5. **https://leetcode.com/problems/smallest-subtree-with-all-the-deepest-nodes/description/**

The root of the smallest subtree containing all deepest nodes is exactly the Lowest Common Ancestor (LCA) of all the deepest nodes.

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def lcaDeepestLeaves(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        def dfs(root,depth):
            if(root is None):
                return None, depth
            
            lca_left, left_depth = dfs(root.left,depth+1)
            lca_right, right_depth = dfs(root.right,depth + 1)

            if(left_depth > right_depth):
                return lca_left, left_depth
            elif(right_depth > left_depth):
                return lca_right, right_depth
            else:
                # The second value returned by dfs() is not the current node's depth. It is the deepest leaf depth found anywhere in that subtree.

                # So after comparing the left and right subtrees:
                # If the left is deeper, return the left subtree's deepest depth.
                # If the right is deeper, return the right subtree's deepest depth.
                # If they are equal, the current node is the LCA, but the deepest depth is still the children's deepest depth, not the current node's depth.
                # return root, right_depth will also as left_depth, right_depth both are equal here
                return root, left_depth
        lca , depth = dfs(root,0)
        return lca
```

6. **https://leetcode.com/problems/find-mode-in-binary-search-tree/description/**

```python
# brute force 
# time = O(n)
# space = O(n)

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def findMode(self, root: Optional[TreeNode]) -> List[int]:
        n = {}
        q = deque([root])
        while q:
            for i in range(len(q)):
                node = q.popleft()
                if node.val in n:
                    n[node.val] += 1
                else:
                    n[node.val] = 1
                
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
        
        max_count = max(n.values())
        result = [k for k, v in n.items() if v == max_count]
        return result
```

```python
# optimal , sorted sequence + two pointers approach
# space: O(h), which is recursion call stack, but we are not using any etc list or dict

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def findMode(self, root: Optional[TreeNode]) -> List[int]:
        max_count = 0
        cur_count = 0
        prev_val = None
        ans = []
        def dfs(root):
            nonlocal max_count,cur_count,prev_val
            if(root is None):
                return
            dfs(root.left)
            if prev_val == root.val:
                cur_count += 1
            else:
                cur_count = 1
            prev_val = root.val
            if cur_count > max_count:
                max_count = cur_count
                ans.clear()
                ans.append(root.val)
            elif cur_count == max_count:
                ans.append(root.val)
            dfs(root.right)
        dfs(root)
        return ans
```
