1. **https://leetcode.com/problems/minimum-depth-of-binary-tree/description/**

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def minDepth(self, root: Optional[TreeNode]) -> int:
        def dfs(root):
            if(root is None):
                return 0
            l = dfs(root.left)
            r = dfs(root.right)
            if(root.left is None and root.right is None):
                return 1
            if root.left is None:
                return 1 + r
            if root.right is None:
                return 1 + l
            return min(l,r) + 1
        return dfs(root)
```

2. **https://leetcode.com/problems/maximum-depth-of-binary-tree/description/**

```python
class Solution:
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        if(root == None):
            return 0
        
        lh = self.maxDepth(root.left)
        rh = self.maxDepth(root.right)
        return max(lh,rh) + 1
```

3. **https://leetcode.com/problems/balanced-binary-tree/description/**

```python
class Solution:
    def isBalanced(self, root: Optional[TreeNode]) -> bool:
        def dfs(root):
            if(root is None):
                return 0
            l = dfs(root.left)
            if l is False:
                return False
            r = dfs(root.right)
            if r is False:
                return False
            diff = abs(l-r)
            if diff > 1:
                return False
            return max(l,r) + 1
        return dfs(root) is not False
```

4. **https://leetcode.com/problems/diameter-of-binary-tree/description/**

```python
class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        self.ans = 0
        def height(root):
            if(root == None):
                return 0
            
            lh = height(root.left)
            rh = height(root.right)
            self.ans = max(self.ans, lh+rh)
            return max(lh,rh) + 1
        height(root)
        return self.ans
```

5. **https://leetcode.com/problems/binary-tree-tilt/description/**

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
'''
tilt total needs to go up
subtree sum needs to go up

tilt is just accumulation so we could use nonlocal tilt or we can return a tuple (int, int)
'''
class Solution:
    def findTilt(self, root: Optional[TreeNode]) -> int:
        
        def dfs(node) -> (int, int):  # returns subtree sum at node, running tilt total
            if not node:
                return 0, 0
            
            leftSum, leftTilt = dfs(node.left)
            rightSum, rightTilt = dfs(node.right)
            tilt = abs(leftSum - rightSum)  # accumulates from running subtree sum 

            return (leftSum+rightSum+node.val, tilt+leftTilt+rightTilt)

        _, tilt_total = dfs(root)
        return tilt_total
```

6. **https://leetcode.com/problems/check-completeness-of-a-binary-tree/description/**

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
from collections import deque
class Solution:
    def isCompleteTree(self, root: Optional[TreeNode]) -> bool:
        found_null = False
        q = deque()
        q.append(root)
        while q:
            node = q.popleft()
            if node.left is None:
                found_null = True
            if node.left:
                if found_null:
                    return False
                else:
                    q.append(node.left)
            if node.right is None:
                found_null = True
            if node.right:
                if found_null:
                    return False
                else:
                    q.append(node.right)
        return True



# class Solution:
#     def isCompleteTree(self, root: Optional[TreeNode]) -> bool:

#         queue = deque([root])
#         found_null = False

#         while queue:

#             node = queue.popleft()

#             if node is None:
#                 found_null = True

#             else:

#                 if found_null:
#                     return False

#                 queue.append(node.left)
#                 queue.append(node.right)

#         return True
```

🔴 7. **https://leetcode.com/problems/validate-binary-search-tree/description/**

```python
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        def dfs(node,low,high):
            if node is None:
                return True
            
            if node.val >= high or node.val <= low:
                return False
            
            return (dfs(node.left,low,node.val) and dfs(node.right,node.val,high))
        return dfs(root,float("-inf"),float("inf"))
```

8. **https://leetcode.com/problems/recover-binary-search-tree/description/**

```python
# brute solution

# Time complexity: O(n log n) because sorting dominates.
# Space: O(n) , because of inorder array, sorted array, recursion stack (worst case)

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def recoverTree(self, root: Optional[TreeNode]) -> None:
        """
        Do not return anything, modify root in-place instead.
        """
        first = None
        second = None
        inorder_arr = []
        def dfs(root):
            if(root is None):
                return 
            dfs(root.left)
            inorder_arr.append(root.val)
            dfs(root.right)
        dfs(root)
        
        sorted_arr = sorted(inorder_arr)
        for i in range(len(sorted_arr)):
            if(sorted_arr[i] != inorder_arr[i]):
                first = inorder_arr[i]
                second = sorted_arr[i]
                break

        def dfs_recover(root):
            if(root is None):
                return 
            dfs_recover(root.left)
            if(root.val == first):
                root.val = second
            elif(root.val == second):
                root.val = first
            dfs_recover(root.right)
        dfs_recover(root)
        
        
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def recoverTree(self, root: Optional[TreeNode]) -> None:
        """
        Do not return anything, modify root in-place instead.
        """
        first = None
        prev = None
        second = None
        def dfs(root):
            nonlocal first
            nonlocal prev
            nonlocal second
            if(root is None):
                return 
            dfs(root.left)
            if(prev and prev.val > root.val):
                if(not first):
                    first = prev
                    # The first value that is "too large" is always prev. Save it once.
                    # The last value that is "too small" is always node. Keep updating it until traversal finishes.
                second = root
            prev = root
            dfs(root.right)
        dfs(root)

        first.val, second.val = second.val, first.val
        
```
