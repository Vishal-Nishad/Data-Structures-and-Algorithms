# To solve a problem, follow this structure:
- What is the brute-force idea
- Why does it fail or seem inefficient
- What observation changes the problem
- How does that observation lead naturally to the final algo(optal if any)

### Scenario: You need to visit each node in the tree exactly once, exploring all nodes in a specific order (e.g., inorder, preorder, postorder).

### Clue: Look for problems where you must visit each node systematically without missing any, often requiring recursion(for DFS) or iteration(for BFS).

1. **https://leetcode.com/problems/binary-tree-inorder-traversal/description/**

````python
# recursive traversing

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        def dfs(root):
            if(root is None):
                return
            dfs(root.left)
            ans.append(root.val)
            dfs(root.right)
        dfs(root)
        return ans
````

````python
# 🔴 iterative traversing,

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        stk = []
        curr = root
        while curr or stk:
            while curr:
                stk.append(curr)
                curr = curr.left
            curr = stk.pop()
            ans.append(curr.val)
            curr = curr.right
        return ans
````

2. **https://leetcode.com/problems/binary-tree-preorder-traversal/**

````python
# recursive 

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        def dfs(root):
            if(root is None):
                return None
            ans.append(root.val)
            dfs(root.left)
            dfs(root.right)
        dfs(root)
        return ans
````

````python
# iterative

# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        stk = []
        curr = root
        while curr or stk:
            while curr:
                stk.append(curr)
                ans.append(curr.val)
                curr = curr.left
            curr = stk.pop()
            curr = curr.right
        return ans
````

3. **https://leetcode.com/problems/binary-tree-postorder-traversal/description/**

````python
# iterative

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def postorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        if(root is None):
            return []
        ans = []
        stk1 = [root]
        stk2 = []
        while stk1:
            node = stk1.pop()
            stk2.append(node)
            if node.left:
                stk1.append(node.left)
            if node.right:
                stk1.append(node.right)
        while stk2:
            ans.append(stk2.pop().val)  
        return ans
````

4. **https://leetcode.com/problems/binary-tree-level-order-traversal/description/**

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
from collections import deque
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        if root is None:
            return []
        q = deque()
        ans = []
        q.append(root)
        while q:
            size = len(q)
            curr_ans = []
            for i in range(size):
                node = q.popleft()
                curr_ans.append(node.val)
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            ans.append(curr_ans)
        return ans
````

5. **https://leetcode.com/problems/binary-tree-zigzag-level-order-traversal/description/**

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
from collections import deque
class Solution:
    def zigzagLevelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        if(root is None):
            return []
        ans = []
        q = deque()
        q.append(root)
        left_to_right = True
        while q:
            curr_ans = []
            size = len(q)
            for j in range(size):
                node = q.popleft()
                curr_ans.append(node.val)
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            if not left_to_right:
                curr_ans.reverse()
            if not left_to_right:
                left_to_right = True
            else:
                left_to_right = False
            ans.append(curr_ans)
        return ans
````

6. **https://leetcode.com/problems/binary-tree-level-order-traversal-ii/description/**

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
from collections import deque
class Solution:
    def levelOrderBottom(self, root: Optional[TreeNode]) -> List[List[int]]:
        if(root is None):
            return []
        ans = []
        q = deque()
        q.append(root)

        while q:
            curr_level = []
            for i in range(len(q)):
                node = q.popleft()
                curr_level.append(node.val)
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            ans.append(curr_level)
        ans = ans[::-1]
        return ans
````

**Absolutely.** In fact, this is how experienced engineers think.

They **don't memorize 20 traversal problems**.

They memorize **4 templates**, and every traversal problem is a small variation of one of them.

# DFS Recursive Master Template

Every recursive DFS is literally this:

```python
def dfs(root):

    if root is None:
        return

    # PREORDER

    dfs(root.left)

    # INORDER

    dfs(root.right)

    # POSTORDER
```

Now just decide where to do your work.

---

## Preorder

Root Left Right

```python
def preorder(root):

    if root is None:
        return

    ans.append(root.val)

    preorder(root.left)

    preorder(root.right)
```

Only difference:

```python
ans.append(root.val)
```

comes first.

---

## Inorder

Left Root Right

```python
def inorder(root):

    if root is None:
        return

    inorder(root.left)

    ans.append(root.val)

    inorder(root.right)
```

---

## Postorder

Left Right Root

```python
def postorder(root):

    if root is None:
        return

    postorder(root.left)

    postorder(root.right)

    ans.append(root.val)
```

---

# This is the ONLY recursive template you need.

Just move

```python
ans.append(root.val)
```

around.

---

# DFS Iterative Master Templates

These are different.

---

## Iterative Preorder

Think

> Visit immediately.

Template

```python
stack = [root]

while stack:

    node = stack.pop()

    visit(node)

    if node.right:
        stack.append(node.right)

    if node.left:
        stack.append(node.left)
```

Notice

Right first

Left second

Because stack reverses.

---

## Iterative Inorder

Think

> Keep going left.

Template

```python
stack = []

curr = root

while curr or stack:

    while curr:

        stack.append(curr)

        curr = curr.left

    curr = stack.pop()

    visit(curr)

    curr = curr.right
```

This template is used everywhere.

---

## Iterative Postorder (2-stack)

Think

> Root Right Left

then reverse.

Template

```python
stack1 = [root]

stack2 = []

while stack1:

    node = stack1.pop()

    stack2.append(node)

    if node.left:
        stack1.append(node.left)

    if node.right:
        stack1.append(node.right)

while stack2:

    visit(stack2.pop())
```

---

# BFS Master Template

This one solves **80% of BFS problems**.

```python
queue = deque([root])

while queue:

    size = len(queue)

    level = []

    for _ in range(size):

        node = queue.popleft()

        level.append(node.val)

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

    answer.append(level)
```

Everything else modifies only one line.

# Variations

---

## Level Order

```python
answer.append(level)
```

---

## Zigzag

```python
if reverse:

    level.reverse()

answer.append(level)

reverse = not reverse
```

Only two extra lines.

---

## Level Order II

Instead of

```python
answer.append(level)
```

do

```python
answer.insert(0, level)
```

or reverse at the end.

---

## Right Side View

Instead of

```python
answer.append(level)
```

do

```python
answer.append(level[-1])
```

---

## Left Side View

```python
answer.append(level[0])
```

---

## Average of Levels

```python
answer.append(sum(level)/len(level))
```

---

See?

Same BFS.

Only one line changes.
