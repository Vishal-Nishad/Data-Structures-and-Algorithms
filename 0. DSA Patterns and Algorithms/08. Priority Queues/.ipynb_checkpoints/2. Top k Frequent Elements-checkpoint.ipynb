# Scenario: You're analyzing data and need to identify the most frequently occurring elements.
# Clue: Look for problems where you need to track element frequencies and select the top K elements based on occurrence count.

1. **https://leetcode.com/problems/top-k-frequent-elements/description/**

```python
# brute using sorting
# time: nlogn
# space: n

class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        hash_map = {}
        for i in range(len(nums)):
            hash_map[nums[i]] = hash_map.get(nums[i],0) + 1
        arr = []
        for val, freq in hash_map.items():
            arr.append([val,freq])
        arr.sort(key = lambda x:x[1])
        arr.reverse()
        return [arr[i][0] for i in range(k)]
```

```python
# heap
# time: nlogk
# space: n

class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        hash_map = {}
        for i in range(len(nums)):
            hash_map[nums[i]] = hash_map.get(nums[i],0) + 1
        heap = []
        
        for val,freq in hash_map.items():
            heapq.heappush(heap,[freq,val])
            if(len(heap)>k):
                heapq.heappop(heap)
        return [val for freq,val in heap]
```

2. **https://leetcode.com/problems/sort-characters-by-frequency/description/**

```python
class Solution:
    def frequencySort(self, s: str) -> str:
        hash_map = {}
        for i in range(len(s)):
            hash_map[s[i]] = hash_map.get(s[i],0) + 1
        heap = []
        
        for val,freq in hash_map.items():
            heapq.heappush(heap,[-freq,val])

        ans = []
        while heap:
            freq,val = heapq.heappop(heap)
            for i in range(-freq):
                ans.append(val)
            
        return "".join(ans)
```

```python
🔴 do bucket sort approach to achieve linear time complexity
class Solution:
    def frequencySort(self, s: str) -> str:
        freq = {}

        for c in s:
            freq[c] = freq.get(c, 0) + 1
        bucket = [[] for _ in range(len(s) + 1)]
        for ch, f in freq.items():
            bucket[f].append(ch)

        ans = []
        for f in range(len(s), 0, -1):
            for ch in bucket[f]:
                ans.append(ch * f)
        return "".join(ans)
```

3. **https://leetcode.com/problems/top-k-frequent-words/description/**

```python
# optimal sol, manual custom heap implementation using arr of arr
# time: n logk
# space: n


class Solution:
    def topKFrequent(self, words: List[str], k: int) -> List[str]:
        def is_worse(a,b):
            if(a[0] != b[0]):
                return a[0] < b[0]

            return a[1] > b[1]
        def min_down_heapify(arr, parent_index, n):
            while True:
                smallest = parent_index
                left_indx = 2 * parent_index + 1
                right_indx = 2 * parent_index + 2
                if left_indx < n and is_worse(arr[left_indx],arr[smallest]):
                    smallest = left_indx
                if right_indx < n and is_worse(arr[right_indx],arr[smallest]):
                    smallest = right_indx
                if smallest == parent_index:
                    break
                arr[smallest], arr[parent_index] = (
                    arr[parent_index],
                    arr[smallest]
                )
                parent_index = smallest

        hash_map = {}
        for i in range(len(words)):
            hash_map[words[i]] = hash_map.get(words[i],0) + 1
        
        arr = []
        for words, freq in hash_map.items():
            arr.append([freq,words])
        
        heap = arr[:k]

        for i in range((k//2)-1,-1,-1):
            min_down_heapify(heap,i,k)

        for i in range(k,len(arr)):
            if(is_worse(heap[0],arr[i])):
                heap[0] = arr[i]
                min_down_heapify(heap,0,k)
        
        def remove_el(arr):
            first = arr[0]
            arr[0] = arr[-1]
            arr.pop()
            min_down_heapify(heap,0,len(arr))
            return first

        arr = []
        while heap:
            ele = remove_el(heap)
            arr.append(ele[1])
        arr.reverse()
        return arr
```
