# Scenario: You have a large dataset, and you need to efficiently find the Kth largest or smallest element.
# Clue: Look for problems where you need to track the K largest or smallest elements while processing the dataset.

1. **https://leetcode.com/problems/kth-largest-element-in-an-array/description/**

```python
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        def min_down_heapify(nums, parent_index, n):
            while True:
                smallest = parent_index
                left_indx = 2 * parent_index + 1
                right_indx = 2 * parent_index + 2
                if left_indx < n and nums[left_indx] < nums[smallest]:
                    smallest = left_indx
                if right_indx < n and nums[right_indx] < nums[smallest]:
                    smallest = right_indx
                if smallest == parent_index:
                    break
                nums[smallest], nums[parent_index] = (
                    nums[parent_index],
                    nums[smallest]
                )
                parent_index = smallest

        # Build min heap using first k elements
        for i in range((k // 2) - 1, -1, -1):
            min_down_heapify(nums, i, k)

        # Process remaining elements
        for i in range(k, len(nums)):
            if nums[i] > nums[0]:
                nums[i], nums[0] = nums[0], nums[i]
                min_down_heapify(nums, 0, k)

        # Root = kth largest
        return nums[0]
```

2. **https://leetcode.com/problems/kth-smallest-element-in-a-sorted-matrix/description/**

```python
class Solution:
    def kthSmallest(self, matrix: List[List[int]], k: int) -> int:
        # heap = [value,row,col]
        def min_down_heapify(heap,parent_index):
            n = len(heap)
            while True:
                smallest = parent_index
                left_indx = 2*smallest + 1
                right_indx = left_indx + 1
                if(left_indx < n and heap[left_indx][0] < heap[smallest][0]):
                    smallest = left_indx
                if(right_indx < n and heap[right_indx][0] < heap[smallest][0]):
                    smallest = right_indx
                if(smallest == parent_index):
                    break
                heap[smallest],heap[parent_index] = heap[parent_index],heap[smallest]
                parent_index = smallest
        
        def remove_element(heap):
            ele = heap[0]
            heap[0] = heap[-1]
            heap.pop()
            min_down_heapify(heap,0)
            return ele
        
        def insert_element(heap,value):
            heap.append(value)
            child = len(heap)-1

            while child > 0:
                parent_indx = (child - 1)//2
                if(heap[parent_indx][0] <= heap[child][0]):
                    break
                heap[parent_indx], heap[child] = heap[child],heap[parent_indx]
                child = parent_indx
        heap = []
        for i in range(len(matrix)):
            heap.append([matrix[i][0],i,0])
        
        for i in range((len(heap)//2)-1,-1,-1):
            min_down_heapify(heap,i)
        
        count = 0
        while count != k:
            element = remove_element(heap)
            count += 1
            value = element[0]
            row = element[1]
            col = element[2]
            if col + 1 < len(matrix):
                insert_element(heap,[matrix[row][col+1],row,col+1])
        return value
```

3. **https://leetcode.com/problems/kth-largest-element-in-a-stream/description/**

```python
class KthLargest:
    def __init__(self, k: int, nums: List[int]):
        self.k = k
        self.heap = nums[:k]
        for i in range((len(self.heap)//2)-1,-1,-1):
            self.min_down_hepify(self.heap,i)
        for i in range(k,len(nums)):
            if(self.heap[0]<nums[i]):
                self.heap[0],nums[i] = nums[i],self.heap[0]
                self.min_down_hepify(self.heap,0)
        for i in range(len(self.heap)-1,k-1,-1):
            self.heap.pop()
        
    def add(self, val: int) -> int:
        if(len(self.heap) < self.k):
            self.heap.append(val)
            child = len(self.heap)-1
            while(child > 0):
                parent_indx = (child - 1) // 2
                if(self.heap[parent_indx] <= self.heap[child]):
                    break
                self.heap[parent_indx],self.heap[child] = self.heap[child],self.heap[parent_indx]
                child = parent_indx
        else:
            if(val > self.heap[0]):
                self.heap[0] = val
                self.min_down_hepify(self.heap,0)
        return self.heap[0]

    def min_down_hepify(self,arr,parent_index):
        n = len(arr)
        while True:
            smallest = parent_index
            left_child = parent_index * 2 + 1
            right_child = parent_index * 2 + 2
            if(left_child < n and arr[left_child] < arr[smallest]):
                smallest = left_child
            if(right_child < n and arr[right_child] < arr[smallest]):
                smallest = right_child
            if(smallest == parent_index):
                break
            arr[smallest],arr[parent_index] = arr[parent_index],arr[smallest]
            parent_index = smallest
        


# Your KthLargest object will be instantiated and called as such:
# obj = KthLargest(k, nums)
# param_1 = obj.add(val)
```

```python
import heapq
class KthLargest:

    def __init__(self, k, nums):
        self.k = k
        self.nums = nums
        heapq.heapify(self.nums)  # Convert nums into a heap
        # If the heap is larger than k, remove the smallest elements until it has exactly k elements
        while len(self.nums) > k:
            heapq.heappop(self.nums)

    def add(self, val):
        heapq.heappush(self.nums, val)
        # If after adding the new value, the heap has more than k elements, pop the smallest
        if len(self.nums) > self.k:
            heapq.heappop(self.nums)
        # The smallest element in the heap is now the k-th largest element
        return self.nums[0]

        


# Your KthLargest object will be instantiated and called as such:
# obj = KthLargest(k, nums)
# param_1 = obj.add(val)
```

4. **https://leetcode.com/problems/find-median-from-data-stream/**

```python
# brute force
# Time: nlogn, due to sorting
# Space: n
class MedianFinder:

    def __init__(self):
        self.nums = []
        
    def addNum(self, num: int) -> None:
        self.nums.append(num)
        
    def findMedian(self) -> float:
        self.nums.sort()
        if(len(self.nums) % 2 != 0):
            return self.nums[len(self.nums)//2]
        else:
            return (self.nums[len(self.nums)//2] + self.nums[(len(self.nums)//2)-1])/2
        
# Your MedianFinder object will be instantiated and called as such:
# obj = MedianFinder()
# obj.addNum(num)
# param_2 = obj.findMedian()
```

5. **https://leetcode.com/problems/k-closest-points-to-origin/description/**

```python
class Solution:
    def kClosest(self, points: List[List[int]], k: int) -> List[List[int]]:
        def max_down_heapify(arr,parent_index):
            while True:
                largest = parent_index
                left_child = 2 * parent_index + 1
                right_child = 2 * parent_index + 2
                if(left_child < k and arr[left_child][0] > arr[largest][0]):
                    largest = left_child
                if(right_child < k and arr[right_child][0] > arr[largest][0]):
                    largest = right_child
                if(largest == parent_index):
                    break
                arr[largest],arr[parent_index] = arr[parent_index],arr[largest]
                parent_index = largest
        dis = []
        for i in range(len(points)):
            d = (points[i][0] ** 2) + (points[i][1] ** 2)
            dis.append([d,i]) 
        
        for i in range((k//2)-1,-1,-1):
            max_down_heapify(dis,i)
        
        for i in range(k,len(dis)):
            if(dis[0][0] > dis[i][0]):
                dis[0],dis[i] = dis[i],dis[0]
                max_down_heapify(dis,0)
        ans = []
        for i in range(k):
            ans.append(points[dis[i][1]])
        return ans
```

```python
class Solution:
    def kClosest(self, points: List[List[int]], k: int) -> List[List[int]]:
        heap = []
        for i in range(len(points)):
            x = points[i][0]
            y = points[i][1]
            heapq.heappush(heap,[-(x*x + y*y), points[i]])
            if(len(heap) > k):
                heapq.heappop(heap)
        return [point for d,point in heap]
```
