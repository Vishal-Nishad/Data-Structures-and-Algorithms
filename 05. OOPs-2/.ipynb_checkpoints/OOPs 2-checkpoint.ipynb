```python
class Vehicle:
    def __init__(self,color,max_speed):
        self._color=color
        # MAKING a data member private and then 
        # using getter  setter function to access it 
        self.__max_speed=max_speed
    def getMaxSpeed(self):     # function for accessing
        return self.__max_speed

    def setMaxSpeed(self,max_speed):  # function for setting
        self.__max_speed=max_speed
    # but to private member within same class getter setter is not required
    def print(self):
        print("Max_Speed = ",self.__max_speed)
class Car(Vehicle):
    def __init__(self,color,max_speed,num_gears,is_convertible):
        super().__init__(color,max_speed)
        self.num_gears=num_gears
        self.is_convertible=is_convertible
    def print_car(self):
        # here _ underscore is protected member we can access it outside the class
        # and also modify it, it is only to give hint to the user don't try to modify it coz it kind of protected or private
        print(f"Color is : {self._color}")
        # TO ACCESS private member getter method is required
        print(f"Max Speed : {self.getMaxSpeed()}")
        print(f"Number of gears are : {self.num_gears}")
        print(f"Is convertible : {self.is_convertible}")
        # we can use both syntax to print class method i.e
        self.print()
        super().print() ## use super() for direct accessing by object name i.e c.print()
        
c=Car("White",200,6,False)
c.print_car()
```
