```python
class Node:
    def __init__(self,value = None):
        self.value = value
        self.next = None

class CircularSLL:
    def __init__(self):
        self.head = None
        self.tail = None

    def __iter__(self):
        node = self.head
        while node:
            yield node
            if node.next == self.head:
                break
            node = node.next
    # Creation of circular singly linked list
    def create_csll(self,value):
        node = Node(value)
        node.next = node
        self.head = node
        self.tail = node
    
    def insertion_at_beginning(self,value):
        if self.head is None:
            return create_csll(value)
        new_node = Node(value)
        else:
            self.tail.next = new_node
            new_node.next = self.head
            self.head = new_node
            self.tail = new_node
    def insertion_at_end(self, value):
        if self.head is None:
            return create_csll(value)
        new_node = Node(value)
        
circular_sll = CircularSLL()
circular_sll.create_csll(1)
print([node.value for node in circular_sll])
```
