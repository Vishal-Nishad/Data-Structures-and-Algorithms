```python
class Node:
    def __init__(self,value = None):
        self.value = value
        self.next = None

class CircularSLL:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def __iter__(self):
        node = self.head
        while node:
            yield node
            if node.next == self.head:
                break
            node = node.next
    # Creation of circular singly linked list
    def create_csll(self,value):
        node = Node(value)
        node.next = node
        self.head = node
        self.tail = node
        self.size = 1
    
    def insertion_at_beginning(self,value):
        if self.head is None:
            return self.create_csll(value)
        else:
            new_node = Node(value)
            self.tail.next = new_node
            new_node.next = self.head
            self.head = new_node
            self.size += 1
    def insertion_at_end(self, value):
        if self.head is None:
            return self.create_csll(value)
        else:
            new_node = Node(value)
            new_node.next = self.tail.next
            self.tail.next = new_node
            self.tail = new_node
            self.size += 1
    def insertion_at_middle(self,value,position):
        if position < 0 or position > self.size:
            raise IndexError("Index out of bounds")
        elif position == 0:
            return self.insertion_at_beginning(value)
        elif position == self.size:
            return self.insertion_at_end(value)
        else:
            new_node = Node(value)
            index = 0
            tempnode = self.head
            while index < position - 1:
                tempnode = tempnode.next
                index += 1
            nextnode = tempnode.next
            tempnode.next = new_node
            new_node.next = nextnode
            self.size += 1
    def traverse_csll(self):
        curr = self.head
        while curr:
            print(curr.value)
            if curr.next == self.head:
                break
            curr = curr.next

    def search_csll(self,position):
        if position < 0 or position >= self.size:
            raise IndexError("Index out of bound")
        else:
            index = 0
            tar_node = self.head
            while index < position:
                tar_node = tar_node.next
                index += 1
            print(tar_node.value)
                
        
circular_sll = CircularSLL()
circular_sll.create_csll(1)
circular_sll.insertion_at_beginning(9)
circular_sll.insertion_at_middle(10,1)
circular_sll.insertion_at_middle(50,0)
circular_sll.insertion_at_middle(30,4)
circular_sll.insertion_at_middle(20,3)
circular_sll.insertion_at_middle(32,0)
print([node.value for node in circular_sll])

circular_sll.traverse_csll()
circular_sll.search_csll()
```
