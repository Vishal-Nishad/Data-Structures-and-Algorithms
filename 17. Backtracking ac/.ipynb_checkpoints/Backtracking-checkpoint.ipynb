# HOW TO IDENTIFY RECURSION / BACKTRACKING
### Whenever problem says:
- generate all
- find all

THINK:
"Can I make a choice and recursively solve smaller problem?"

If YES:
recursion/backtracking is likely.

# Q. Backtracking Array, 

````python
def change_array(arr,i,val):
    if(i==len(arr)):
        print(arr)
        return
    # // recursion work
    arr[i] = val
    change_array(arr,i+1,val+1) # fnx call step
    arr[i] = arr[i] - 2 #backtracking step

a = [0]*5
change_array(a,0,1)
a
````

# Print all subset of given string Recursively

````python
'''
Whenever you generate all subsets/power set:
Number of subsets = 2^n
Time complexity is almost always:O(n⋅2^n)

because printing/building each subset costs extra time.
'''
# space complexity = O(n)
def rec_str_subset(str,new_str,i):
    if(i == len(str)):
        if(len(new_str)==0):
            print("Null")
        else:
            print(new_str)
        return
    rec_str_subset(str,new_str + str[i],i+1)
    rec_str_subset(str,new_str,i+1)

s = "abc"
rec_str_subset(s,"",0)
````

````python
General Formula for Recursive Complexity
Whenever analyzing recursion:

TIME
Ask:
1. How many recursive calls?

(branching factor)
2. Height/depth of recursion tree?
3. Work done inside each call?

Then:
Total Work=nodes×work per node


SPACE
Ask:
1. Maximum recursion depth?
2. Any extra data structure?
3. Are copies being created?
(strings/lists/slices)
````

Excellent. This is exactly the right way to learn recursion complexity.

Let's apply the framework directly on YOUR program step-by-step.

---

# Program

```python id="yj8os0"
def rec_str_subset(str,new_str,i):

    if(i == len(str)):
        print(new_str)
        return

    rec_str_subset(str,new_str + str[i],i+1)
    rec_str_subset(str,new_str,i+1)
```

Suppose:

```python id="n4t6qb"
n = len(str)
```

---

# PART 1 — TIME COMPLEXITY

We apply your formula:

---

# STEP 1 → How many recursive calls from each function?

Inside one function call:

```python id="ry0tb8"
rec_str_subset(...)
rec_str_subset(...)
```

Two recursive calls.

So:

# Branching Factor

[
2
]

---

# STEP 2 → Height / Depth of recursion tree

Look at:

```python id="85azv0"
i+1
```

Every recursive call increases `i` by 1.

Recursion stops when:

```python id="ygh4hq"
i == len(str)
```

Meaning:

```text id="3iy0om"
0 → 1 → 2 → 3 → ... → n
```

So maximum depth:

# Height

[
n
]

---

# STEP 3 → Total Nodes in recursion tree

General recursion tree formula:

[
\text{Total Nodes} =
1 + b + b^2 + b^3 + ... + b^h
]

where:

* (b) = branching factor
* (h) = height

Here:

[
b = 2
]
[
h = n
]

So:

[
1 + 2 + 2^2 + 2^3 + ... + 2^n
]

This becomes:

[
O(2^n)
]

---

# STEP 4 → Work done inside each node

Now inspect carefully:

```python id="4vrrrz"
new_str + str[i]
```

This is NOT O(1).

Python strings are immutable.

So:

```python id="e1zjz6"
"ab" + "c"
```

creates a NEW string `"abc"`.

Cost depends on current string size.

Worst case:

```python id="8v4e3l"
O(n)
```

because subset length can become `n`.

---

# STEP 5 → Final Time Complexity

Formula:

# Total Work

[
\text{nodes} \times \text{work per node}
]

So:

[
O(2^n) \times O(n)
]

Final:

# Time Complexity

[
O(n \cdot 2^n)
]

---

# PART 2 — SPACE COMPLEXITY

Now apply second framework.

---

# STEP 1 → Maximum recursion depth

At one moment, how many function calls exist simultaneously?

Path:

```text id="s0c3o5"
0 → 1 → 2 → 3 → ... → n
```

So maximum stack frames:

[
n
]

Thus:

# Recursion Stack

[
O(n)
]

---

# STEP 2 → Extra Data Structures?

Do we use:

* array?
* hashmap?
* queue?
* stack?

No.

So:

```text id="5l32c5"
No major extra DS
```

---

# STEP 3 → Are copies being created?

YES.

This line:

```python id="eqcv85"
new_str + str[i]
```

creates new strings repeatedly.

At depth:

* 1 → string size 1
* 2 → string size 2
* 3 → string size 3
* ...
* n → string size n

So simultaneously recursion stack may hold:

```text id="qnd6zf"
1 + 2 + 3 + ... + n
```

Which is:

[
O(n^2)
]

---

# Final Space Complexity

## Common Interview Answer

(interviewers usually expect)

[
O(n)
]

because they count recursion stack only.

---

# Strict Python Analysis

Including string copies:

[
O(n^2)
]

---

# The MAIN Skill You Should Learn

When analyzing recursion:

---

# TIME

Always ask:

| Question          | Your Program     |
| ----------------- | ---------------- |
| branching factor? | 2                |
| recursion height? | n                |
| total nodes?      | (2^n)            |
| work per node?    | (O(n))           |
| final?            | (O(n \cdot 2^n)) |

---

# SPACE

Always ask:

| Question           | Your Program              |
| ------------------ | ------------------------- |
| recursion depth?   | n                         |
| extra DS?          | none                      |
| copying happening? | yes, strings              |
| final?             | (O(n)) or strict (O(n^2)) |

---

# Very Important Interview Pattern

Most recursive problems are solved exactly like this.

---

# Example Patterns

## Binary recursion

```python id="3w2pqj"
f(n-1)
f(n-1)
```

Usually:

[
O(2^n)
]

---

## Ternary recursion

```python id="f3pyn5"
f(n-1)
f(n-1)
f(n-1)
```

Usually:

[
O(3^n)
]

---

## Divide by 2 recursion

```python id="u6vxuq"
f(n/2)
f(n/2)
```

Usually:

[
O(n \log n)
]

like Merge Sort.

---

# Fast Mental Formula

For recursion trees:

[
\text{Time}
===========

(\text{branching factor})^{(\text{height})}
\times
(\text{work per node})
]

Your case:

[
2^n \times n
============

O(n2^n)
]

This is the fast mental model top candidates use.


# Q Find Permutations

````python
def find_permutations(str,ans_str):
    if(len(str)==0):
        print(ans_str)
        return
    for i in range(len(str)):
        curr = str[i]
        new_str = str[:i] + str[i+1:]
        find_permutations(new_str,ans_str + curr)
s = "ab"
st = "abc"
find_permutations(s,"")
print()
find_permutations(st,"")
````

# Q. N Queens Problem

````python
# ignoring queens move, only placing queens in each row in all possible way
def n_queens(board,row):
    if(row == len(board)):
        print_board(board)
        return
    for i in range(len(board)):
        board[row][i] = "Q"
        n_queens(board,row+1)
        board[row][i] = "X"

def print_board(board):
    print("_____________________________")
    for i in range(len(board)):
        for j in range(len(board)):
            print(board[i][j], end=" ")
        print()
        
        
        
def create_chess_board(n):
    board = []
    for i in range(n):
        board.append([])
        for j in range(n):
            board[i].append("X")
    return board

board = create_chess_board(2)
n_queens(board,0)
````

````python
# actual printed valid nqueens
count = 0
def n_queens(board,row):
    global count
    if(row == len(board)):
        print_board(board)
        count += 1
        return
    # column loop
    for i in range(len(board)):
        if is_safe(board,row,i):
            board[row][i] = "Q"
            n_queens(board,row+1)
            board[row][i] = "X"

def is_safe(board,row,col):
    # vertical safe
    r = row
    while(r>=0):
        if(board[r][col]=="Q"):
            return False
        r -= 1
        
    # diagonal left safe
    r = row
    c = col
    while(r>=0 and c>=0):
        if(board[r][c]=="Q"):
            return False
        r -= 1
        c -= 1
    
    # diagonal right safe
    r = row
    c = col
    while(r>=0 and c<len(board)):
        if(board[r][c]=="Q"):
            return False
        r -= 1
        c += 1
    return True

def print_board(board):
    print("_____________________________")
    for i in range(len(board)):
        for j in range(len(board)):
            print(board[i][j], end=" ")
        print()
        
        
        
def create_chess_board(n):
    board = []
    for i in range(n):
        board.append([])
        for j in range(n):
            board[i].append("X")
    return board

board = create_chess_board(4)
print(count)
n_queens(board,0)
print()
print(count)
````

# N-Queens Print 1 solution
- Check if problem can be solved and print only 1 solution to N queens problem.

````python
# actual printed valid nqueens
count = 0
def n_queens(board,row):
    global count
    if(row == len(board)):
        print_board(board)
        count += 1
        return True
    # column loop
    for col in range(len(board)):
        if is_safe(board,row,col):
            board[row][col] = "Q"
            
            if (n_queens(board,row+1)):
                return True
            board[row][col] = "X"

def is_safe(board,row,col):
    # vertical safe
    r = row
    while(r>=0):
        if(board[r][col]=="Q"):
            return False
        r -= 1
        
    # diagonal left safe
    r = row
    c = col
    while(r>=0 and c>=0):
        if(board[r][c]=="Q"):
            return False
        r -= 1
        c -= 1
    
    # diagonal right safe
    r = row
    c = col
    while(r>=0 and c<len(board)):
        if(board[r][c]=="Q"):
            return False
        r -= 1
        c += 1
    return True

def print_board(board):
    print("_____________________________")
    for i in range(len(board)):
        for j in range(len(board)):
            print(board[i][j], end=" ")
        print()
        
        
        
def create_chess_board(n):
    board = []
    for i in range(n):
        board.append([])
        for j in range(n):
            board[i].append("X")
    return board

board = create_chess_board(4)
print(count)
n_queens(board,0)
print()
print(count)
````

# Grid Ways
- Find number of ways to reach from (0,0) to (N-1,M-1) in a N*M Grid. Allowed moves - right or down.

````python
# time = O(2^n*m)
def grid_ways(i,j,n,m):
    if(i==n-1 and j == m-1):
        return 1
    elif(i==n or j == n):
        return 0
    # recursion work
    w1 = grid_ways(i+1,j,n,m)
    w2 = grid_ways(i,j+1,n,m)
    return w1 + w2

grid_ways(0,0,3,3)
````

# Q. Sudoku Solver

````python
def sudoku_solver(sudoku,row,col):
    # base case
    if(row == 0):
        return True

    # recursion work
    next_row = row
    next_col = col + 1
    if(col + 1 == 9):
        next_row = row + 1
        next_col = 0

    if(sudoku[row][col] !=0):
        return sudoku_solver(sudoku,next_row,next_col)
    
    for digit in range(1,10):
        if(is_safe(sudoku,row,col,digit)):
            sudoku[row][col] = digit
            if(sudoku_solver(sudoku,next_row,next_col)):
                return True
            sudoku[row][col] = 0
    return False
````
