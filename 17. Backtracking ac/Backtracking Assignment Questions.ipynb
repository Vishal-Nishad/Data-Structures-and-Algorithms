# Q. Rat in Maze Problem

## STAGE 1
- Only:
- Down + Right
- One path

* Understand:
    - how TRUE propagates upward

## STAGE 2
- Down + Right
- All paths

## STAGE 3
- UDLR
- One path

## STAGE 4
- UDLR
- All paths

* This becomes the complete backtracking version.

```python
# choices = 2 i.e down, right
# and only have to print one solution so instantly return True after getting first solution

# stage one
def rat_maze(maze):
    n = len(maze)
    sol = [[0 for i in range(n)]for r in range(n)]

    def solve_maze(row,col):
        # base case
        if(row == n-1 and col == n-1 and maze[row][col]==1):
            sol[row][col] = 1
            return True

        # invalid case
        if(row >= n or col >=n or maze[row][col]==0):
            return False

        # Do recursion work
        sol[row][col] = 1

        # recurse call
        # down move
        if(solve_maze(row+1,col)):
            return True

        # right move
        if(solve_maze(row,col+1)):
            return True

        # backtrack and undo
        sol[row][col] = 0
        return False
    if(solve_maze(0,0)):
        for r in sol:
            print(r)
    else:
        return "solution does not exist"

maze = [
    [1,1,1],
    [1,1,1],
    [0,1,1]
]
rat_maze(maze)
```
