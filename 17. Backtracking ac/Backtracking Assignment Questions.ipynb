````python
# choices = 2 i.e down, right
# and only have to print one solution so instantly return True after getting first solution

# stage one
def rat_maze(maze):
    n = len(maze)
    sol = [[0 for i in range(n)]for r in range(n)]

    def solve_maze(row,col):
        # base case
        if(row == n-1 and col == n-1 and maze[row][col]==1):
            sol[row][col] = 1
            return True

        # invalid case
        if(row >= n or col >=n or maze[row][col]==0):
            return False

        # Do recursion work
        sol[row][col] = 1

        # recurse call
        # down move
        if(solve_maze(row+1,col)):
            return True

        # right move
        if(solve_maze(row,col+1)):
            return True

        # backtrack and undo
        sol[row][col] = 0
        return False
    if(solve_maze(0,0)):
        for r in sol:
            print(r)
    else:
        return "solution does not exist"

maze = [
    [1,1,1],
    [1,1,1],
    [0,1,1]
]
rat_maze(maze)
````

# Q. knight tour problem

````python
def knight_tour(n):
    board = [[-1 for _ in range(n)]for _ in range(n)]
    x_move = [2,2,-2,-2,-1,1,-1,1]
    y_move = [1,-1,-1,1,2,2,-2,-2]

    def is_safe(row,col):
        return(
            0 <= row < n and
            0 <= col < n and
            board[row][col] == -1
        )

    def solve_knight_tour(row,col,move):
        # invalid case
        if not is_safe(row,col):
            return False

        # place the value
        board[row][col] = move
        # base case
        if(move == (n*n)-1):
            return True

        # recurse
        for i in range(8):
            next_row = row + x_move[i]
            next_col = col + y_move[i]
            if(solve_knight_tour(next_row,next_col,move+1)):
                return True
        board[row][col] = -1
        return False
    solve_knight_tour(0,0,0)
    for i in board:
        print(i)

knight_tour(8)
````

This is one of the most important recursion/backtracking concepts.

The position of:

```python id="h8j7h6"
board[...] = -1
```

depends entirely on:

> “Where was the value placed?”

That is the key rule.

---

# FIRST STYLE

This was the first style:

```python id="m7k4ae"
for i in range(8):

    next_row = ...
    next_col = ...

    if is_safe(next_row, next_col):

        board[next_row][next_col] = move

        if solve(...):
            return True

        board[next_row][next_col] = -1
```

---

# What Is Happening Here?

Inside loop, YOU place:

```python id="z4i8fp"
board[next_row][next_col] = move
```

So this loop iteration is responsible for that placement.

Therefore SAME loop iteration must undo it:

```python id="otgt44"
board[next_row][next_col] = -1
```

---

# Dry Run

Suppose:

```text id="3ff9d6"
Current = (0,0)
```

Loop tries:

```text id="bljl9d"
Move 1 → (2,1)
```

You place:

```python id="8pt8qe"
board[2][1] = 1
```

Then recurse.

Suppose recursion fails.

Now before trying another move:

```text id="q3pd4d"
Move 2 → (1,2)
```

you MUST clean previous move:

```python id="w8z8fi"
board[2][1] = -1
```

Otherwise board remains polluted.

So undo must happen INSIDE loop because:

* every iteration creates a different placement
* every iteration must clean its own placement

---

# SECOND STYLE

Second style:

```python id="rzl7ai"
def solve(row, col, move):

    if not safe:
        return False

    board[row][col] = move

    for i in range(8):

        if solve(next_row, next_col, move+1):
            return True

    board[row][col] = -1
```

---

# What Changed?

Now placement happens HERE:

```python id="7nbeho"
board[row][col] = move
```

This is OUTSIDE loop.

Meaning:

* current function call owns this placement
* not individual loop iterations

So undo must happen when entire function finishes.

That is why backtracking is outside loop.

---

# VERY IMPORTANT THINKING RULE

# Rule

> Backtracking happens in the same scope where placement happened.

---

# Style 1 Scope

Placement:

```python id="4n3y53"
inside loop
```

Undo:

```python id="v3o8q9"
inside loop
```

Correct.

---

# Style 2 Scope

Placement:

```python id="9dx8cf"
outside loop
```

Undo:

```python id="9w5f2m"
outside loop
```

Correct.

# Visual Comparison

# STYLE 1

```text id="y8rt1g"
LOOP ITERATION
    place
    recurse
    undo
```

Each iteration manages its own move.

---

# STYLE 2

```text id="l6u8h5"
FUNCTION CALL
    place

    LOOP:
        recurse
        recurse
        recurse

    undo
```

Entire function call manages one placement.

---

# Extremely Important Recursion Insight

A recursive call represents a "state".

Question is:

> “Who owns the state change?”

---

# Style 1

Loop iteration owns state.

---

# Style 2

Function call owns state.

---

# Analogy

Imagine entering rooms.

# Style 1

Each door handler:

* opens door
* explores
* closes door

---

# Style 2

Whole room handler:

* enters room once
* explores all doors
* leaves room once

---

# This Same Pattern Appears Everywhere

You will see this in:

* N Queens
* Sudoku
* Maze
* Subsets
* Permutations
* Graph DFS backtracking

Understanding this deeply is a major recursion milestone.


# Keypad Combinations**
Given a string containing digits from 2-9 inclusive, print all possible letter combinations that
the number could represent. You can print the answer in any order.
A mapping of digits to letters (just like on the telephone buttons) is given below. Note that 1
does not map to any let**
e* rs.
Sample Input 1 : digits =     - "23"
Sample Output 1 : "ad", "ae", "af", "bd", "be", "bf", "cd", "ce",*  "cf"
Sample Input 2 : digits    -  = "2"
Sample Output 2 : "a", "* b", "c"
Sample Input 3 : dig    - its = ""
Sample Output 3 : ””

````python
def keypd_combination(digits):
    if not digits:
        return []
    phone = {
        "2": "abc",
        "3": "def",
        "4": "ghi",
        "5": "jkl",
        "6": "mno",
        "7": "pqrs",
        "8": "tuv",
        "9": "wxyz"
    }

    result = []
    def backtrack(ind,current_str):
        # base case
        if(ind == len(digits)):
            result.append(current_str)
            return

        # work
        digit = digits[ind]
        letters = phone[digit]

        # recursion on choices
        for i in range(len(letters)):
            backtrack(ind+1,current_str + letters[i])

    backtrack(0,"")
    return result

keypd_combination("2345")
````
