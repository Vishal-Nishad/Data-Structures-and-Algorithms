# Q. Backtracking Array, 

```python
def change_array(arr,i,val):
    if(i==len(arr)):
        print(arr)
        return
    # // recursion work
    arr[i] = val
    change_array(arr,i+1,val+1) # fnx call step
    arr[i] = arr[i] - 2 #backtracking step

a = [0]*5
change_array(a,0,1)
a
```

# Print all subset of given string Recursively

```python
def rec_str_subset(str,new_str,i):
    if(i == len(str)):
        if(len(new_str)==0):
            print("Null")
        else:
            print(new_str)
        return
    rec_str_subset(str,new_str + str[i],i+1)
    rec_str_subset(str,new_str,i+1)

s = "abc"
rec_str_subset(s,"",0)
```
