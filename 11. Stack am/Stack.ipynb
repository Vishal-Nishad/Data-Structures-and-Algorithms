**creating stack**

```python
class Stack:
    def __init__(self):
        self.list = []

    def __str__(self):
        values = self.list.reverse()
        values = [str(i) for i in self.list]
        # print(values)
        return '\n'.join(values)

    def isempty(self):
        return len(self.list) == 0
    def push(self,value):
        self.list.append(value)
        return "The element has been successfully inserted"
    def pop(self):
        if self.isempty():
            return "there is no element in the stack"
        else:
            return self.list.pop()
    def peek(self):
        # returning/checking last element of stack
        if self.isempty():
            return "There is no elelment in the stack"
        else:
            return self.list[len(self.list) - 1]

    def delete(self):
        self.list = None  # time complexity O(1)
st = Stack()
st.list.append(4)
st.list.append(1)
st.list.append(5)
st.push(7)
st.pop()
st.pop()
print(st.peek())

st.pop()
print(st.isempty())
print(st)
st.pop()   # after this stack become empty
st.pop()
```
