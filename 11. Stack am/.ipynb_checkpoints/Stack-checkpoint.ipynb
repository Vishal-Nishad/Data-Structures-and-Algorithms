# 🔍 The Truth (In Python)

## 1️⃣ Stack Using List (Dynamic Array)

```python
stack = []
stack.append(x)
stack.pop()
```

### Performance

| Operation | Time Complexity |
| --------- | --------------- |
| Push      | O(1) amortized  |
| Pop       | O(1)            |
| Peek      | O(1)            |

Python list resizing is amortized O(1), not slow.

When resizing happens:

* It doubles capacity.
* So expensive resize happens rarely.

So it is NOT slow in practice.

In fact:

👉 Python list stack is usually **faster than linked list stack**.

Because:

* Better CPU cache locality
* No pointer chasing
* Implemented in optimized C code

---

## 2️⃣ Stack Using Linked List

Push and pop are O(1) always.

But:

* Each node has extra pointer memory.
* Memory is scattered in heap.
* Worse cache performance.
* More Python object overhead.

So in Python, linked list stack is usually **slower**.

---

# 🔥 Correct Comparison

| Factor                    | Array (Python list) | Linked List  |
| ------------------------- | ------------------- | ------------ |
| Push                      | O(1) amortized      | O(1)         |
| Pop                       | O(1)                | O(1)         |
| Memory per element        | Less                | More         |
| Cache performance         | Better              | Worse        |
| Implementation            | Simple              | More complex |
| Real-world speed (Python) | Faster              | Slower       |

---

# 💡 Important Concept: Amortized O(1)

Sometimes list resizing takes O(n).

But it happens rarely.

Over many operations, average cost is constant.

That’s called amortized O(1).

**creating stack**

````python
class Stack:
    def __init__(self):
        self.list = []

    def __str__(self):
        values = self.list.reverse()
        values = [str(i) for i in self.list]
        # print(values)
        return '\n'.join(values)

    def isempty(self):
        return len(self.list) == 0
    def push(self,value):
        self.list.append(value)
        return "The element has been successfully inserted"
    def pop(self):
        if self.isempty():
            return "there is no element in the stack"
        else:
            return self.list.pop()
    def peek(self):
        # returning/checking last element of stack
        if self.isempty():
            return "There is no elelment in the stack"
        else:
            return self.list[len(self.list) - 1]

    def delete(self):
        self.list = None  # time complexity O(1)
st = Stack()
st.list.append(4)
st.list.append(1)
st.list.append(5)
st.push(7)
st.pop()
st.pop()
print(st.peek())

st.pop()
print(st.isempty())
print(st)
st.pop()   # after this stack become empty
st.pop()
````

````python
# implementing stack by defining maxsize 
class StackM:
    def __init__(self,max_size):
        self.max_size = max_size
        self.list = []
    
    def __str__(self):
        reverse_values = reversed(self.list)
        values = [str(i) for i in reverse_values]
        return "\n".join(values)

    def isempty(self):
        return len(self.list) == 0

    def isfull(self):
        return len(self.list) == self.max_size
        
    def push(self,value):
        if self.isfull():
            return "the stack limit is full"
        self.list.append(value)
        print(self.list)
        return "the element inserted successfully"
        
    def pop(self):
        if self.isempty():
            return "stack is empty"
        else:
            self.list.pop()
        print()
        print(self.list)

    def peek(self):
        # returning/checking last element of stack
        if self.isempty():
            return "There is no elelment in the stack"
        else:
            return self.list[len(self.list) - 1]

    def delete(self):
        self.list = None  # time complexity O(1)

st = StackM(5)
st.push(5)
st.push(3)
st.push(0)
st.push(9)
st.push(0)
print(st.push(9))
print(st)

st.pop()
````

**implement stack using linked list**

````python
class Node:
    def __init__(self,value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

class Stack:
    def __init__(self):
        self.linkedlist = LinkedList()   # we can write here directly self.head = None, but separate linked list class provide etc abstraction layer
        self.count = 0
    def isempty(self):
        if self.linkedlist.head is None:
            return True
        else:
            return False
    def push(self,value):
        new_node = Node(value)
        new_node.next = self.linkedlist.head
        self.linkedlist.head = new_node
        self.count += 1

    def pop(self):
        if self.isempty():
            return "linked list is empty"
        self.linkedlist.head = self.linkedlist.head.next
        self.count -= 1

    def printstack(self):
        if self.isempty():
            return "linked list is empty"
        curr = self.linkedlist.head
        while curr:
            print(curr.value)
            curr = curr.next

    def peek(self):
        if self.isempty():
            return "linked list is empty"
        return self.linkedlist.head.value

    def delete(self):
        self.linkedlist.head = None
        self.count = 0

    def size(self):
        return self.count
        
st = Stack()
st.push(4)
st.push(6)
st.push(9)
st.printstack()
st.pop()
print()
st.printstack()
print()
print(st.peek())
st.size()
````
