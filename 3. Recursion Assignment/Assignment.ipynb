```python
# 1. Geometric sum
def g_s(k):
    if k==0:
        return 1
    return g_s(k-1)+1/(2**k)

g_s(3)
```

```python
# 2. is_pallindrom
def is_pallindrome(st:str,start,end):
    if start>=end:
        return True
    if st[start]!=st[end]:
        return False
    return is_pallindrome(st,start+1,end-1)
s='abcaba'
is_pallindrome(s,0,len(s)-1)
```

```python
# 3. 
def sum_of_digit(n,start,end):
    if start>end:
        return 0
    return int(n[start])+sum_of_digit(n,start+1,end)
li="12345"
sum_of_digit(li,0,4)
```

```python
# 4. Mutliplication without using product operator
def multi(m,n):
    if n==0:
        return 0
    return m+multi(m,n-1)

multi(3,1000)
```
