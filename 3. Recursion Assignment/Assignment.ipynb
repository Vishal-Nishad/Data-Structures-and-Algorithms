```python
# 1. Geometric sum
def g_s(k):
    if k==0:
        return 1
    return g_s(k-1)+1/(2**k)

g_s(3)
```

```python
# 2. is_pallindrom
def is_pallindrome(st:str,start,end):
    if start>=end:
        return True
    if st[start]!=st[end]:
        return False
    return is_pallindrome(st,start+1,end-1)
s='abcaba'
is_pallindrome(s,0,len(s)-1)
```

```python
# 3. 
def sum_of_digit(n,start,end):
    if start>end:
        return 0
    return int(n[start])+sum_of_digit(n,start+1,end)
li="12345"
sum_of_digit(li,0,4)
```

```python
# 4. Mutliplication without using product operator
def multi(m,n):
    if n==0:
        return 0
    return m+multi(m,n-1)

multi(3,1000)
```

```python
# 5. Count Zeroes
def count_zeroes(n):
    if n==0:
        return 1
    if n<10:
        return 0
    last_digit=n%10
    if last_digit==0:
        return 1+count_zeroes(n//10)
    return count_zeroes(n//10)

count_zeroes(1001001)
```

```python
# 6. String to integer
def st_to_int(st,index=0):
    if index==len(st):
        return 0
    current_digit=ord(st[index])-ord('0')
    return current_digit*(10**(len(st)-index-1))+st_to_int(st,index+1)

st='00001231'
st_to_int(st)
```

```python
# 7. Pair star
def pair_star(st):
    if len(st)<=1:
        return st
    if st[0]==st[1]:
        return st[0]+'*'+pair_star(st[1:])
    return st[0]+pair_star(st[1:])

st="helllo"
pair_star(st)
```

```python
# 8. Check ab
def check_ab(st,index=0):
    if index==len(st):
        return True
    if st[index]!='a':
        return False
    if len(st)>index+1 and st[index+1]=='b':
        if len(st)>index+2 and st[index+2]=='b':
            return check_ab(st,index+3)
        else:
            return False
    return check_ab(st,index+1)

st='aaabbaabb'
check_ab(st)
```

```python
#. 9. Staircase problem
Step 1: Understanding the Problem
A child is climbing N steps, and they can take either 1 step, 2 steps, or 3 steps at a time.

Example (N = 4)
The child can reach step 4 by:

Taking 1 step from step 3
Taking 2 steps from step 2
Taking 3 steps from step 1
So, the total ways to reach step 4 = ways to reach step 3 + ways to reach step 2 + ways to reach step 1

W(4)=W(3)+W(2)+W(1)
Step 2: Understanding the Base Cases
We need to handle the smallest cases manually:

N = 1 → Only 1 way (just 1 step).
Ways: (1)
So, W(1) = 1

N = 2 → Two ways: (1+1) or (2).
Ways: (1,1), (2)
So, W(2) = 2

N = 3 → Four ways: (1+1+1), (1+2), (2+1), (3).
Ways: (1,1,1), (1,2), (2,1), (3)
So, W(3) = 4
```

```python
def staircase(n):
    if n==1:
        return 1
    if n==2:
        return 2
    if n==3:
        return 4
    return staircase(n-1)+staircase(n-2)+staircase(n-3)

staircase(5)
```
