### Q. For a given integer array of size N. you have to find all the occurrences of(indices) of given element(key) and print them. Use a recursive function to solve this problem

```python
def print_rec(arr,key,i):
    if(i==len(arr)):
        return
    if(arr[i]==key):
        print(i,end=" ")
    print_rec(arr,key,i+1)

a=[3,2,4,5,6,2,7,2,2]
print_rec(a,2,0)
```

### Q 2. You are given a number(2019), convert it into english like string like (two zero one nine). use a recursive function to solve this problem
- Note: the digits of the number only be in the range of 0-9 and the last digit can't be 0

```python
digit_str = {}
digit_str[0] = "zero"
digit_str[1] = "one"
digit_str[2] = "two"
digit_str[3] = "three"
digit_str[4] = "four"
digit_str[5] = "five"
digit_str[6] = "six"
digit_str[7] = "seven"
digit_str[8] = "eight"
digit_str[9] = "nine"
def rec_num_str(n):
    if(n==0):
        return
    rec_num_str(n//10)
    num = n%10
    print(digit_str[num], end=" ")

rec_num_str(2019)
```

### Q 3. write a program to find length of the string using recursion

```python
def rec_str_len(str,i):
    if(i==len(str)):
        return 0
    return 1+ rec_str_len(str,i+1)

rec_str_len("vishal",0)
```

### Q4. you are given a string, you have to find the count of all the substrings starting and ending with the same character

```python
def helper(s, start, end):
    if start > end:
        return 0

    count = 0

    for i in range(start, end + 1):
        if s[start] == s[i]:
            count += 1

    return count + helper(s, start + 1, end)


print(helper("aba", 0, 2))
```
