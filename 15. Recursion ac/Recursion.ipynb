# Recursion
- Recursion is a method of solving a computational problem where the solution depends on solutions to smaller instances of the same problem.

- Base Case
- What work we have to do
- arguments or how we gonna call our inner function to solve problems in part

# Q 1. Print number from n to 1

````python
def rec_print_no(n):
    if n==0:
        return
    print(n,end=" ")
    rec_print_no(n-1)

rec_print_no(5)
````

# Q 2. factorial of number n

````python
def fact_rec(n):
    if(n==1):
        return 1
    prev_fact = fact_rec(n-1)
    return n*prev_fact  # n * fact_rec(n-1)
fact_rec(5)
````

# Q 3. print sum of first n natural numbers

````python
def rec_sum_n(n):
    if(n==1):
        return 1
````
````python
    prev_sum = rec_sum_n(n-1)
    total_sum = n + prev_sum
    return total_sum

rec_sum_n(5)
````

# Q 4. print nth fibonacci number

````python
def fib_iterative(n):
    if(n==1 or n==2):
        return 1
    a = 0
    b = 1
    for i in range(1,n):
        c = a+b
        a = b
        b = c
    return b
fib_iterative(5)
````

````python
def rec_fib(n):
    if(n==1 or n==2):
        return 1
    return rec_fib(n-1)+rec_fib(n-2)
rec_fib(10)
````

# Q 5. Check if array is sorted or not

````python
def rec_is_sorted(arr,i):
    if(i == len(arr)-1):
        return True
    if(arr[i]>arr[i+1]):
        return False
    return rec_is_sorted(arr,i+1)

a = [1,4,2,3]
rec_is_sorted(a,0)
````

# Q 6. find first occurence of given key return its index

````python
def rec_first_occ(arr,key,i):
    if(i == len(arr)):
        return -1
    if(arr[i] == key):
        return i
    return rec_first_occ(arr,key,i+1)

a = [1,3,2,3]
rec_first_occ(a,5,0)
````

# Q 7. find last occurence of given key return its index

````python
def rec_first_occ(arr,key,i):
    if(i == -1):
        return -1
    if(arr[i] == key):
        return i
    return rec_first_occ(arr,key,i-1)

a = [1,3,2,3]
rec_first_occ(a,3,len(a)-1)
````

````python
# another 
def rec_last_occ(arr,key,i):
    if(i == len(arr)):
        return -1
    is_found = rec_last_occ(arr,key,i+1)
    if(is_found == -1 and arr[i] == key):
        return i
    return is_found

a = [1,3,2,3]
rec_last_occ(a,3,0)
````

# Q 8. Print x to the power n

````python
def rec_x_pow_n(x,n):
    if(n==0):
        return 1
    call_value = rec_x_pow_n(x,n-1)
    ans = x * call_value
    return ans
rec_x_pow_n(2,5)
````

````python
def rec_x_pow_n(x,n):
    if(n==0):
        return 1
    return x * rec_x_pow_n(x,n-1)
rec_x_pow_n(2,5)
````

````python
# optimize, logn
def rec_x_pow_n(x,n):
    if(n==0):
        return 1
    halfpower = rec_x_pow_n(x,n//2)
    halfpower = halfpower * halfpower
    if(n%2 != 0):
        halfpower = x * halfpower
    return halfpower

rec_x_pow_n(2,10)
````

# Q 9. Tiling problem

````python
def rec_tiling_problem(n): # 2*n floor size ,,, and 2*1 tile size
    if(n==0 or n==1):
        return 1
    # work we have to do
    # vertical choice
    vertical = rec_tiling_problem(n-1)

    # horizontal choice
    horizontal = rec_tiling_problem(n-2)

    total_ways = horizontal + vertical 
    return total_ways

rec_tiling_problem(4)
````

# Q 11. Friends Pairing Problem
- Given n friends, each one can remain single or can be paired up with some other friend. Each friend can be paired only once. Find out the total number of ways in which friends can remain single or can be paired up.

````python
def friends_pairing(n):
    if(n==1 or n==2):
        return n
    single_choice = friends_pairing(n-1)
    pair_choice = friends_pairing(n-2)
    total_ways = single_choice + (n-1)*pair_choice
    return total_ways
friends_pairing(4)
````

This is one of the best recursion/DP problems because it teaches the **core skill**:

> “Fix one person, then count all valid remaining possibilities.”

Right now don’t focus on formulas.

Suppose there are:

```text id="j0ct6v"

```

Each friend has only 2 possibilities:

* stay single
* pair with exactly one other friend



```text id="k9x5jd"

```

# Choice 2: A pairs with someone

A can pair with:

```text id="j9m49v"
B OR C
```

For general `n` friends:

A has:


* 2 friends are used


Ways to arrange them:

```text id="gfik0c"
f(n-2)
```

And since A had `(n-1)` partner choices:

```text id="j5v9j9"
(n-1) * f(n-2)
```

---

# 🧠 Final recurrence

So total ways:

```text id="pwdd50"
f(n) = f(n-1) + (n-1) * f(n-2)
```

---

# 💡 Why multiplication?

This is VERY important.

Suppose:

```text id="t0cjlwm"
A can pair with B
A can pair with C
A can pair with D
```

That’s:

```text id="hzzlga"
(n-1) choices
```

For EACH choice:

* remaining friends can be arranged in:

```text id="6vkizk"
f(n-2) ways
```

So:

```text id="fbozls"
total = (n-1) * f(n-2)
```

# ✅ Recursive code

```python id="g0vq3e"
def friends_pairing(n):
    if n == 0 or n == 1:
        return 1

    # single
    single = friends_pairing(n-1)

    # pair
    pair = (n-1) * friends_pairing(n-2)

    return single + pair
```

# Q 12. Binary Strings Problem
- Print binary string without consecutive ones.

````python
def rec_bin_str(st,n_st,i):
    if(i==len(st)):
        return n_st
    if(st[i]=="0"):
        n_st += st[i]
    elif(i<len(st)-1 and st[i]!=st[i+1]):
        n_st += st[i]
    elif(i==len(st)-1):
        n_st += st[i]
    return rec_bin_str(st,n_st,i+1)

st = "01101"
n_st = ""
rec_bin_str(st,n_st,0)
````

# Q 13. Print all binary strings of size n without consecutive ones

````python
def rec_bin_st_n(n,lp,n_str):
    if(n==0):
        print(n_str)
        return
    rec_bin_st_n(n-1,0,n_str + "0")
    if(lp==0):
        rec_bin_st_n(n-1,1,n_str + "1")

rec_bin_st_n(3,0,"")
````
