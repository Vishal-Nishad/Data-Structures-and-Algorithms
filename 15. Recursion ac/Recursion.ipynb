# Recursion
- Recursion is a method of solving a computational problem where the solution depends on solutions to smaller instances of the same problem.

## Steps to solve recursion problems
- Base Case
- What work we have to do
- arguments or how we gonna call our inner function to solve problems in part

# Q 1. Print number from n to 1

```python
def rec_print_no(n):
    if n==0:
        return
    print(n,end=" ")
    rec_print_no(n-1)

rec_print_no(5)
```

```python
# 1 to n
def rec_print_no(n):
    if n==0:
        return
    rec_print_no(n-1)
    print(n,end=" ")

rec_print_no(5)
```

# Q 2. factorial of number n

```python
def fact_rec(n):
    if(n==1):
        return 1
    prev_fact = fact_rec(n-1)
    return n*prev_fact  # n * fact_rec(n-1)
fact_rec(5)
```

# Q 3. print sum of first n natural numbers

```python
def rec_sum_n(n):
    if(n==1):
        return 1
    return n + rec_sum_n(n-1)
rec_sum_n(5)
```

```python
def rec_sum_n(n):
    if(n==1):
        return 1
    prev_sum = rec_sum_n(n-1)
    total_sum = n + prev_sum
    return total_sum

rec_sum_n(5)
```

# Q 4. print nth fibonacci number

```python
def fib_iterative(n):
    if(n==1 or n==2):
        return 1
    a = 0
    b = 1
    for i in range(1,n):
        c = a+b
        a = b
        b = c
    return b
fib_iterative(5)
```

```python
def rec_fib(n):
    if(n==1 or n==2):
        return 1
    return rec_fib(n-1)+rec_fib(n-2)
rec_fib(10)
```

```python
def rec_fib(n):
    if(n==1 or n==2):
        return 1
    fib1 = rec_fib(n-1)
    fib2 = rec_fib(n-2)
    fib_sum = fib1 + fib2
    return fib_sum
rec_fib(10)
```

# Q 5. Check if array is sorted or not

```python
def rec_is_sorted(arr,i):
    if(i == len(arr)-1):
        return True
    if(arr[i]>arr[i+1]):
        return False
    return rec_is_sorted(arr,i+1)

a = [1,4,2,3]
rec_is_sorted(a,0)
```

# Q 6. find first occurence of given key return its index

```python
def rec_first_occ(arr,key,i):
    if(i == len(arr)):
        return -1
    if(arr[i] == key):
        return i
    return rec_first_occ(arr,key,i+1)

a = [1,3,2,3]
rec_first_occ(a,5,0)
```

# Q 7. find last occurence of given key return its index

```python
def rec_first_occ(arr,key,i):
    if(i == -1):
        return -1
    if(arr[i] == key):
        return i
    return rec_first_occ(arr,key,i-1)

a = [1,3,2,3]
rec_first_occ(a,3,len(a)-1)
```

```python
# another 
def rec_last_occ(arr,key,i):
    if(i == len(arr)):
        return -1
    is_found = rec_last_occ(arr,key,i+1)
    if(is_found == -1 and arr[i] == key):
        return i
    return is_found

a = [1,3,2,3]
rec_last_occ(a,3,0)
```
