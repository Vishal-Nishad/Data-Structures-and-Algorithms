# Recursion
- Recursion is a method of solving a computational problem where the solution depends on solutions to smaller instances of the same problem.

## Steps to solve recursion problems
- Base Case
- What work we have to do
- arguments or how we gonna call our inner function to solve problems in part

# Print number from n to 1

```python
def rec_print_no(n):
    if n==0:
        return
    print(n,end=" ")
    rec_print_no(n-1)

rec_print_no(5)
```

```python
# 1 to n
def rec_print_no(n):
    if n==0:
        return
    rec_print_no(n-1)
    print(n,end=" ")

rec_print_no(5)
```

# factorial of number n

```python
def fact_rec(n):
    if(n==1):
        return 1
    prev_fact = fact_rec(n-1)
    return n*prev_fact  # n * fact_rec(n-1)
fact_rec(5)
```
