## 8. Self Parameter / instance methods - in which self is passed

```python
class Student:
    passin_percentage=40 # class attribute
    def student_details(self):
        self.name="vishal"
        self.percentage=80
        print("name = ", self.name)
        print("percentage = ", self.percentage)
        print("Passin percentage is : ",self.passin_percentage)

    def is_passed(self):
        if self.percentage>Student.passin_percentage:
            print("Student is passed.")
        else:
            print("Student is not passed.")

s1=Student()
s1.student_details()
s1.is_passed()
```

## 11. Static method

```python
class Student:
    passin_percentage=40 # class attribute
    def student_details(self):
        self.name="vishal"
        self.percentage=80
        print("name = ", self.name)
        print("percentage = ", self.percentage)
        print("Passin percentage is : ",self.passin_percentage)

    def is_passed(self):
        if self.percentage>Student.passin_percentage:
            print("Student is passed.")
        else:
            print("Student is not passed.")
            
    # static method doesn't require self argument
    @staticmethod
    def welcome_to_school():
        print("Hey! Welcome to School.")

s1=Student()
s1.student_details()
s1.is_passed()
s1.welcome_to_school()
```

## 12. init method

```python
class Student:
    def __init__(self,name,roll_no):
        self.name=name
        self.rollnumber=roll_no

    def print_student_details(self):
        print(self.name, end=" ")
        print(self.rollnumber)
        
```

```python
s1=Student("vishal",1)
s1.__dict__
```

```python
s2=Student("rahul",2)
s2.__dict__
```

```python
s2.print_student_details()
```

## Public and Private Modifiers

```python
class Student:
    def __init__(self,name,age):
        self.__name=name
        self.age=age
    def print_student_details(self):
        print(self.__name,end=" ")
        print(self.age)

s=Student("vishal",20)
print(s.age)
print(s.name)
```

```python
class Student:
    def __init__(self,name,age):
        self.__name=name
        self.age=age
    def print_student_details(self):
        print(self.__name,end=" ")
        print(self.age)

s=Student("vishal",20)
s.print_student_details()
```
