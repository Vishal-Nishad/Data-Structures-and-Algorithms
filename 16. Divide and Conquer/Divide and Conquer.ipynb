# merge sort

```python
# creating new array not inplace sorting
# time = nlogn
# space = n
def mergeSort(arr):
    def divide(arr):
        if(len(arr)<=1):
            return arr
        mid = len(arr)//2
        left = divide(arr[:mid])
        right = divide(arr[mid:])
        return merge(left,right)

    def merge(left,right):
        i = 0
        j = 0
        tem = []
        while(i<len(left) and j<len(right)):
            if left[i]<=right[j]:
                tem.append(left[i])
                i += 1
            else:
                tem.append(right[j])
                j += 1
        while(i<len(left)):
            tem.append(left[i])
            i += 1
        while(j<len(right)):
            tem.append(right[j])
            j += 1
        return tem
    return divide(arr)

a = [5,-1,2]
b = mergeSort(a)
print(b)
a
```
