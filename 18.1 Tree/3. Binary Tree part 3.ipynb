1. **https://www.geeksforgeeks.org/dsa/print-all-nodes-of-given-binary-tree-at-the-kth-level/**

```python
# using bfs
from collections import deque

class TreeNode:
    def __init__(self,val):
        self.data = val
        self.left = None
        self.right = None
        
def bfs(k,root):
    if(root == None):
        return []
    q = deque()
    q.append(root)
    level = 0
    while q:
        size = len(q)
        ans = []
        if level == k:
            for node in q:
                ans.append(node.data)
            return ans

        for _ in range(size):
            node = q.popleft()
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        level += 1
    return []
root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(6)
root.right.left = TreeNode(6)
root.left.left = TreeNode(11)
root.left.right = TreeNode(141)
k = 2
bfs(2,root)
```

```python
# dfs
def dfs(root,k,ans,level):
    if(root is None):
        return

    if level == k:
        ans.append(root.data)
        return
    dfs(root.left,k,ans,level+1)
    dfs(root.right,k,ans,level + 1)

def klevel(root,k):
    ans = []
    dfs(root,k,ans,0)
    return ans

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(6)
root.right.left = TreeNode(6)
root.left.left = TreeNode(11)
root.left.right = TreeNode(141)
k = 2

klevel(root,k)
```

2. **https://leetcode.com/problems/lowest-common-ancestor-of-a-binary-tree/description/**

```python
# brute force
# time complexity = O(n)
# space complexity = O(h)

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def findpath(self, root, target, path):
        if(root == None):
            return False
        path.append(root)
        if root == target:
            return True
        if(self.findpath(root.left,target,path)):
            return True
        if(self.findpath(root.right,target,path)):
            return True
        path.pop()
        return False
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        path1 = []
        path2 = []
        self.findpath(root,p,path1)
        self.findpath(root,q,path2)
        i = 0
        while(i<len(path1) and i <len(path2) and (path1[i] == path2[i])):
            i += 1
        return path1[i-1]
```

```python
Whenever you analyze complexity, ask these three questions in order:

Question 1
How many times is each node visited?
Once → O(n)
Twice → Still O(n)
Three times → Still O(n)
Nested traversal of all nodes → Maybe O(n²)
For findPath(), each node is visited at most once, so it's O(n).

Question 2
Am I storing anything extra?
Here:
path1
path2
Both store a root-to-node path.
Each has at most h nodes.

Question 3
How deep can the recursion go?

Recursion depth equals the height of the tree:
Balanced tree → O(log n)
Skewed tree → O(n)
    
One important interview habit
When asked for space complexity, don't just say O(h). Clarify what h means:
"The auxiliary space is O(h) due to the recursion stack and the two path arrays, where h is the height of the tree. 
In a balanced tree this is O(log n), while in the worst-case skewed tree it becomes O(n)."
```

```python
# time: O(n), but only one dfs, no while loop comparison 
# space = Still O(h), but without the two explicit path arrays.

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        if(root == None):
            return None
        
        if (root == p or root == q):
            return root
        l = self.lowestCommonAncestor(root.left,p,q)
        r = self.lowestCommonAncestor(root.right,p,q)

        if(l and r):
            return root
        
        if l:
            return l
        else:
            return r
```

3. **https://www.geeksforgeeks.org/problems/min-distance-between-two-given-nodes-of-a-binary-tree/1**
   **https://algo.monster/liteproblems/1740**

```python
# time : O(n)
# space : O(h)

# A binary tree node
# class Node:
#     def __init__(self, val):
#         self.data = val
#         self.left = None
#         self.right = None

class Solution:
    def lca(self, root,a,b):
        if(root == None or (root.data == a or root.data == b)):
            return root
        
        l = self.lca(root.left,a,b)
        r = self.lca(root.right,a,b)
        
        if l and r:
            return root
        if l:
            return l
        else:
            return r
        
    def findedges(self,node,target):
        if(node is None):
            return -1
        if(node.data == target):
            return 0
        l = self.findedges(node.left,target)
        r = self.findedges(node.right,target)
        if l == -1 and r == -1:
            return -1
        return 1 + max(l,r)
        
    def findDist(self, root, a, b):
        # code here
        lca_node = self.lca(root,a,b)
        return self.findedges(lca_node,a) + self.findedges(lca_node,b)
```
