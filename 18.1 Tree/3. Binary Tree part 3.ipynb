```python
# using bfs
from collections import deque

class TreeNode:
    def __init__(self,val):
        self.data = val
        self.left = None
        self.right = None
        
def bfs(k,root):
    if(root == None):
        return []
    q = deque()
    q.append(root)
    level = 0
    while q:
        size = len(q)
        ans = []
        if level == k:
            for node in q:
                ans.append(node.data)
            return ans

        for _ in range(size):
            node = q.popleft()
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        level += 1
    return []
root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(6)
root.right.left = TreeNode(6)
root.left.left = TreeNode(11)
root.left.right = TreeNode(141)
k = 2
bfs(2,root)
```

```python
# dfs
def dfs(root,k,ans,level):
    if(root is None):
        return

    if level == k:
        ans.append(root.data)
        return
    dfs(root.left,k,ans,level+1)
    dfs(root.right,k,ans,level + 1)

def klevel(root,k):
    ans = []
    dfs(root,k,ans,0)
    return ans

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(6)
root.right.left = TreeNode(6)
root.left.left = TreeNode(11)
root.left.right = TreeNode(141)
k = 2

klevel(root,k)
```
