## Finding smallest number and its index

```python
def smallestNum(arr):
    smallest_num=arr[0]
    smallest_index=0
    for i in range(1,len(arr)):
        if arr[i]<smallest_num:
            smallest_num=arr[i]
            smallest_index=i
    print('smallest number is: ',smallest_num )
    print('and its index number is: ',smallest_index)
    return 

arr=[4,6,2,1,7]
print(smallestNum(arr))
    
```

```python
def smallest_num(arr):
    smallest=arr[0]
    smallest_index=0
    for i in range(1,len(arr)):
        if smallest>arr[i]:
            smallest=arr[i]
            smallest_index=i
    return smallest,smallest_index

ar=[3,5,2,1]
smallest_num(ar)
```

## Selection Sort algorithm

```python
def selectionSort(arr):
    for i in range(len(arr)):
        min_index=i
        for j in range (i+1,len(arr)):
            if arr[j]<arr[min_index]:
                min_index=j

        arr[i],arr[min_index]=arr[min_index],arr[i]
    return arr

print(selectionSort([7,3,9,4,2,0,1]))
```

```python
def selection_sort(arr):
    for i in range(len(arr)):
        smallest_index=i
        for j in range (i+1, len(arr)):
            if arr[smallest_index]>arr[j]:
                smallest_index=j
        arr[i],arr[smallest_index]=arr[smallest_index],arr[i]
    return arr

ar=[7,34,234,4,3,2,1,454,4]
selection_sort(ar)
```

```python
def sel_sort(arr):
    for i in range(len(arr)):
        smallest_ind=i
        for j in range(i+1,len(arr)):
            if arr[j]<arr[smallest_ind]:
                smallest_ind=j
        arr[i],arr[smallest_ind]=arr[smallest_ind],arr[i]
    return arr
ar=[7,34,234,4,3,2,1,454,4]
sel_sort(ar)
```

## With separate smallest number funtion

```python
def smallest(arr,i):
    smallest_index=i
    for j in range(i+1,len(arr)):
        if arr[j]<arr[smallest_index]:
            smallest_index=j

    return smallest_index

def selectionSort(arr):
    for i in range(len(arr)):
        min_index=smallest(arr,i)

        arr[i],arr[min_index]=arr[min_index],arr[i]
    return arr

print(selectionSort([3,5,2,0]))
```
